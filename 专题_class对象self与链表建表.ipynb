# A. 为什么会出现 `class`

## A01. 普通变量只能分别保存信息

假设要描述一个学生：

```python
name = '小明'
```

```python
name1, age1
```

我们希望把“属于同一个学生的数据”装在一起：

```text
学生对象
├── name

```

`class` 就是用来定义这种新类型的。

````python
# 信息分散的写法
name = '小明'
age = 20
print(name, age)

# 稍后我们会把它们放进同一个 Student 对象
````

这里的 `pass` 表示暂时什么都不写，让语法保持完整。

# B. 类与对象

## B01. 类像设计说明，对象是实际创建出来的东西

可以先这样类比：

```text
类 class             对象 object
饼干模具              用模具做出的某块饼干
房屋设计图            按设计图建出的某套房

```

使用类名加圆括号创建对象：

```python
student1 = Student()
student2 = Student()
```

两次调用创建的是两个不同对象。

````python
class Student:
    pass


student1 = Student()
student2 = Student()

print(student1)
print(student2)
print('是同一个对象吗：', student1 is student2)  # False
print('类型相同吗：', type(student1) is type(student2))  # True
````

## B02. 变量与对象不是同一件事

执行：

```python
student1 = Student()
```

可以理解为两步：

1. Python 创建一个 `Student` 对象；
2. 变量 `student1` 指向这个对象。

文字图：

```text
student1  ──────>  Student 对象
变量                 实际对象
```

变量更像对象的“标签”或“遥控器”，不是对象本身。链表中的 `head`、`current`、`next_node` 也都是指向节点对象的变量。

````python
class Student:
    pass


student1 = Student()
another_name = student1

print(student1 is another_name)  # True：两个变量指向同一个对象
````

# C. `__init__`：创建对象时自动初始化

## C01. 不希望每次创建后再手动补数据

如果学生对象需要姓名和年龄，我们可以在创建后手动设置：

```python
student.name = '小明'
student.age = 20
```

但更自然的是创建时直接传入：

```python
student = Student('小明', 20)
```

这就需要 `__init__`。

`__init__` 常被称为初始化方法。创建对象后，Python 会自动调用它，为对象准备初始属性。

````python
class Student:
    def __init__(self, name, age):
        print('__init__ 正在执行')
        self.name = name
        self.age = age


student = Student('小明', 20)
print(student.name)
print(student.age)
````

## C02. `__init__` 前后为什么是双下划线

正确拼写是：

```python
__init__
```

- 前面两个下划线；
- `init`；
- 后面两个下划线。

它是 Python 约定的特殊方法名。只有拼写完全正确，创建对象时才会自动执行。

你写的消息中 `**init**` 是 Markdown 加粗形式；Python 代码必须写成 `__init__`。

现阶段先把它理解为：

> 新对象创建后，用传进来的数据给它设置初始属性。

````python
class Demo:
    def __init__(self):
        print('创建 Demo 对象时自动执行')


first = Demo()
second = Demo()  # 创建第二个对象，又执行一次
````

## C03. 默认参数

链表节点常写：

```python
def __init__(self, val=0, next=None):
```

`val=0` 和 `next=None` 是默认值。

- `ListNode()`：使用默认值 `val=0、next=None`；
- `ListNode(5)`：`val=5、next=None`；
- `ListNode(5, another_node)`：`val=5、next=another_node`。

`None` 表示“这里暂时没有指向任何下一个节点”。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode()
node2 = ListNode(5)

print(node1.val, node1.next)
print(node2.val, node2.next)
````

# D. `self` 到底是谁

## D01. `self` 表示“当前正在操作的这个对象”

看下面两次创建：

```python
student1 = Student('小明', 20)
student2 = Student('小红', 19)
```

第一次执行 `__init__` 时：

```text
self 指向 student1 对象
```

第二次执行时：

```text
self 指向 student2 对象
```

因此，同一份类代码可以给不同对象保存各自的数据。

````python
class Student:
    def __init__(self, name):
        print('当前 self：', self)
        self.name = name


student1 = Student('小明')
student2 = Student('小红')

print('student1：', student1)
print('student2：', student2)
print(student1.name, student2.name)
````

## D02. 为什么调用时没有手动传 `self`

定义方法时写：

```python
def introduce(self):
```

调用时写：

```python
student.introduce()
```

Python 会自动把 `student` 这个对象作为第一个参数传给 `self`。

可以近似理解为：

```python
Student.introduce(student)
```

所以你不用写：

```python
student.introduce(student)  # 错误，会多传一次
```

````python
class Student:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print('我是', self.name)


student = Student('小明')
student.introduce()

# 两种调用本质接近
Student.introduce(student)
````

## D03. 拆解 `self.val = val`

这是最容易看晕的一句：

```python
self.val = val
```

左右两个 `val` 职责不同：

- 左边 `self.val`：当前对象的属性；
- 右边 `val`：调用 `__init__` 时传进来的参数。

例如：

```python
node = ListNode(8)
```

执行过程近似为：

```text
参数 val 收到 8
把 8 保存到当前 node 对象的 val 属性
```

可以把参数换个名字，意义不变：

```python
def __init__(self, value):
    self.val = value
```

````python
class ClearNode:
    def __init__(self, incoming_value):
        # 左边是对象属性，右边是本次传入的参数
        self.val = incoming_value


node = ClearNode(8)
print(node.val)
````

## D04. 拆解 `self.next = next`

同样：

- 左边 `self.next`：当前节点对象的属性；
- 右边 `next`：创建节点时传入的下一个节点，默认是 `None`。

属性 `next` 保存的不是“下一个节点的值”，而是**对下一个节点对象的引用**。

```text
node1.next  ──────>  node2 对象
```

这正是链表能连接起来的核心。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node2 = ListNode(2)
node1 = ListNode(1, node2)

print(node1.val)       # 1
print(node1.next)      # node2 对象
print(node1.next.val)  # 2
print(node1.next is node2)  # True
````

# F. 对象引用：链表真正的基础

## F01. 两个变量可以指向同一个对象

```python
node1 = ListNode(1)
another = node1
```

不是复制了一个新节点，而是两个变量指向同一个节点：

```text
node1    ─┐
          ├────> 同一个节点对象(val=1)
another  ─┘
```

通过任何一个变量修改属性，另一个变量再看时也会发现改变。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode(1)
another = node1
another.val = 99

print(node1.val)            # 99
print(node1 is another)     # True
````

## F02. `is` 与 `==`

- `is`：是否是同一个对象；
- `==`：值是否相等，具体行为由类型决定。

两个不同节点可以保存同样的值：

```python
a = ListNode(8)
b = ListNode(8)
```

它们的 `val` 相等，但不是同一个节点。

相交链表、环形链表需要判断节点身份，通常使用 `is`。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


a = ListNode(8)
b = ListNode(8)
c = a

print(a.val == b.val)  # True：值相等
print(a is b)          # False：不同节点
print(a is c)          # True：同一个节点
````

## F03. `None` 表示没有下一个节点

链表最后一个节点：

```python
tail.next = None
```

文字图：

```text
[3 | next] ─────> None
```

遍历链表常写：

```python
while current is not None:
```

只要当前变量还指向节点，就继续；变成 `None` 时说明走到链表末尾之后。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


tail = ListNode(3)
print(tail.next is None)

current = tail
while current is not None:
    print('访问：', current.val)
    current = current.next

print('循环结束时 current：', current)
````

# G. 从一个节点到一条链表

## G01. 先创建一个节点

```python
node1 = ListNode(1)
```

内存关系：

```text
node1 ──> [val=1 | next=None]
```

此时只有一个节点，也可以看成长度为 1 的链表。`node1` 同时是头节点和尾节点。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode(1)
print(node1.val)
print(node1.next)
````

## G02. 创建两个节点并连接

```python
node1 = ListNode(1)
node2 = ListNode(2)
node1.next = node2
```

最后一句不是把数字 2 保存进去，而是让 `node1.next` 指向 `node2` 对象：

```text
node1                  node2
  │                      │
  v                      v
[1 | next] ----------> [2 | None]
```

从 `node1` 出发可以到达两个节点，所以 `node1` 是这条链表的头节点。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode(1)
node2 = ListNode(2)
node1.next = node2

print(node1.val)
print(node1.next.val)
print(node1.next is node2)
````

## G03. 手动创建 `1 -> 2 -> 3`

```python
node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)

node1.next = node2
node2.next = node3
```

结果：

```text
head
 │
 v
[1] -> [2] -> [3] -> None
```

变量 `head` 只需要指向第一个节点。沿着 `next` 就能找到后面所有节点。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)

node1.next = node2
node2.next = node3

head = node1
print(head.val)
print(head.next.val)
print(head.next.next.val)
print(head.next.next.next)
````

## G04. 也可以在创建时直接传入 `next`

因为 `ListNode` 的第二个参数是 `next`，所以：

```python
node3 = ListNode(3)
node2 = ListNode(2, node3)
node1 = ListNode(1, node2)
```

需要从后往前创建，因为创建 `node2` 时要先拿到 `node3`。

还可以嵌套写：

```python
head = ListNode(1, ListNode(2, ListNode(3)))
```

嵌套写法短，但初学时不建议使用，分开写更容易观察对象关系。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node3 = ListNode(3)
node2 = ListNode(2, node3)
node1 = ListNode(1, node2)
head = node1

print(head.val, head.next.val, head.next.next.val)
````

# H. 怎样遍历链表

## H01. `current` 是负责移动的指针变量

不要直接让 `head = head.next` 来观察，先保留头节点：

```python
current = head
while current is not None:
    print(current.val)
    current = current.next
```

每一轮：

1. `current` 指向当前节点；
2. 读取 `current.val`；
3. `current = current.next` 让变量改为指向下一个节点。

这不会修改节点的 `next`，只是在移动变量。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


head = ListNode(1, ListNode(2, ListNode(3)))
current = head

while current is not None:
    print('当前节点：', current.val)
    current = current.next

print('head 仍然指向第一个节点：', head.val)
````

## H02. 移动变量与修改箭头的区别

### 移动变量

```python
current = current.next
```

只是让 `current` 改为指向另一个节点，链表结构不变。

### 修改箭头

```python
current.next = previous
```

改变当前节点的 `next` 属性，链表结构发生变化。

反转链表中两种操作都会出现，所以必须分清等号左边：

- 左边只有变量名：移动变量；
- 左边是 `对象.属性`：修改对象内部属性。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


node1 = ListNode(1)
node2 = ListNode(2)
node1.next = node2

current = node1
current = current.next  # 移动 current
print('current 现在是：', current.val)
print('链表仍然是 1 -> 2：', node1.next.val)

node1.next = None       # 修改 node1 的箭头
print('修改后 node1.next：', node1.next)
````

# I. 从手动建表到 `build_list`

## I01. 手动建表的重复部分

创建 `[1,2,3,4]` 时，我们反复做：

1. 创建新节点；
2. 让当前尾节点的 `next` 指向新节点；
3. 更新尾节点。

```text
新节点 = ListNode(value)
tail.next = 新节点
tail = 新节点
```

`build_list` 只是把这三步放进 `for` 循环，并没有新的神秘概念。

## I02. 为什么使用虚拟头节点 `dummy`

刚开始时还没有真正的第一个节点。

如果不用 `dummy`，第一次添加需要特殊处理：

```python
if head is None:
    head = new_node
```

使用虚拟节点：

```python
dummy = ListNode()
tail = dummy
```

每次都统一执行：

```python
tail.next = new_node
tail = new_node
```

最终真正头节点是 `dummy.next`。虚拟节点本身不属于输入数据。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


dummy = ListNode()  # 临时的虚拟节点
tail = dummy

new_node = ListNode(1)
tail.next = new_node
tail = new_node

print('dummy 的值不是数据重点：', dummy.val)
print('真正第一个节点：', dummy.next.val)
print('tail 指向当前最后节点：', tail.val)
````

## I03. 逐行拆解 `build_list`

```python
def build_list(values):
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next

    return dummy.next
```

逐行含义：

1. 创建虚拟头节点；
2. `tail` 一开始也指向虚拟节点；
3. 遍历普通列表中的每个值；
4. 创建新节点，并接到当前尾节点后面；
5. `tail` 移动到刚创建的新尾节点；
6. 返回虚拟节点后面的真正头节点。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list_with_trace(values):
    dummy = ListNode()
    tail = dummy

    print('开始：dummy 和 tail 指向同一个虚拟节点')
    for value in values:
        tail.next = ListNode(value)
        print(f'创建节点 {value}，接在 tail 后面')
        tail = tail.next
        print(f'tail 移动到节点 {tail.val}')

    return dummy.next


head = build_list_with_trace([1, 2, 3])
print('真正头节点：', head.val)
````

## I04. `tail.next = ListNode(value)` 到底分几步

这一句可以拆成：

```python
new_node = ListNode(value)
tail.next = new_node
```

然后：

```python
tail = tail.next
```

让 `tail` 从旧尾节点移动到新尾节点。

注意：

- `tail.next = new_node` 修改链表连接；
- `tail = tail.next` 只移动 `tail` 变量；
- 它们不是同一种操作。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


dummy = ListNode()
tail = dummy

new_node = ListNode(100)
tail.next = new_node     # 修改旧 tail 的 next 属性
tail = tail.next         # 移动 tail 变量

print(dummy.next.val)
print(tail.val)
print(dummy.next is tail)
````

## I05. 为什么返回 `dummy.next`，不能返回 `dummy` 或 `tail`

- `dummy`：虚拟节点，不属于真实数据；
- `dummy.next`：真正第一个节点，也就是 `head`；
- `tail`：最后一个节点。

对于 `1 -> 2 -> 3`：

```text
dummy -> 1 -> 2 -> 3 -> None
         ^              ^
      dummy.next       tail
```

所以建表函数返回 `dummy.next`。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


dummy = ListNode()
dummy.next = ListNode(1, ListNode(2, ListNode(3)))
tail = dummy.next.next.next

print('dummy：', dummy.val)
print('dummy.next，真正头节点：', dummy.next.val)
print('tail，最后节点：', tail.val)
````

# J. 完整建表、遍历和输出

## J01. 你在 Day 05 中需要看懂的完整工具

```python
class ListNode:
    ...

def build_list(values):
    ...

def list_values(head):
    ...
```

- `ListNode`：定义一个节点对象有哪些属性；
- `build_list`：把普通列表转换成链表；
- `list_values`：把链表值收集回普通列表，方便打印测试。

这三个部分属于学习和本地测试工具。LeetCode 提交时，平台通常已经提供 `ListNode`，你主要提交题目要求的函数。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next

    return dummy.next


def list_values(head):
    values = []
    current = head

    while current is not None:
        values.append(current.val)
        current = current.next

    return values


head = build_list([1, 2, 3, 4])
print(list_values(head))
````

## J02. 用变量跟踪建表过程

输入 `[1,2,3]`：

| 时刻 | 新建节点 | `dummy.next` | `tail` |
|---|---|---|---|
| 开始 | 无 | `None` | dummy |
| 第1轮 | 1 | 节点1 | 节点1 |
| 第2轮 | 2 | 节点1 | 节点2 |
| 第3轮 | 3 | 节点1 | 节点3 |

`dummy.next` 一旦指向节点1，后续一直保持真正头节点；`tail` 每轮向后移动，始终指向最后节点。

### 忘记 `self`


实例方法第一个参数需要接收当前对象，通常命名为 `self`。

### 只写 `val = val`

```python
self.val = val
```

### 把 `next` 保存成值

```python

```

链表的 `next` 应指向节点对象或 `None`，通常不是直接保存整数 2。

### 创建了值相同节点就以为相交

相交要求共享同一个节点对象。

## K03. 为什么属性名前面总有点号

点号表示“属于这个对象”：

```python
node.val
node.next
student.name
```

读作：

- `node.val`：node 对象的 val 属性；
- `node.next`：node 对象的 next 属性。

链式写法：

```python
head.next.next.val
```

按从左到右理解：

1. 找到 `head` 指向的节点；
2. 找它的下一个节点；
3. 再找下一个节点；
4. 读取该节点的值。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


head = ListNode(1, ListNode(2, ListNode(3)))
print(head.next.next.val)  # 3
````
