# 专题：下标、`range` 与切片

## 从 `n - 1`、`k - 1` 到 `nums[-k:]`

这个专题专门解决下面这些容易混在一起的问题：

- 明明有 `n` 个元素，为什么最后下标是 `n - 1`？
- 第 `k` 个元素为什么写成 `nums[k - 1]`？
- “1 到 n”为什么有时写 `range(1, n + 1)`？
- `range(n)`、`range(1, n)` 到底包含哪些数？
- `nums[:k]`、`nums[k:]`、`nums[-k:]`、`nums[:-k]` 分别是什么？
- 为什么切片右端不包含，自己写的 `reverse(left, right)` 却可能包含？

学习目标不是背符号，而是每次都能画出下标，自己判断边界。

## 建议学习顺序

1. **数量和下标**：先分清“有几个”和“编号是多少”。
2. **第 k 个与下标 k**：这是 `k - 1` 最常出现的地方。
3. **`range`**：理解右端为什么不包含。
4. **切片**：把 `range` 的左闭右开迁移到列表。
5. **负数下标**：学会从末尾数。
6. **区间边界**：分清 `[left, right]` 与 `[left, right)`。
7. **轮转数组应用**：解释 `n - 1`、`k - 1`、`nums[-k:]` 为什么同时出现。

每一小节先读图，再运行代码，最后做小练习。

## 先做一个判断：你混乱的是哪一层

看到下面四句话时，先不要急着写代码：

1. 列表有 7 个元素。
2. 列表最后一个元素的下标是 6。
3. 前 3 个元素的下标是 0、1、2。
4. 从数字 1 数到数字 7，一共有 7 个数字。

它们都正确。混乱通常来自把“**第几个**”“**下标**”“**普通数字范围**”当成同一件事。

这个专题会一直使用下面的列表：

| 第几个 | 第1个 | 第2个 | 第3个 | 第4个 | 第5个 | 第6个 | 第7个 |
|---|---:|---:|---:|---:|---:|---:|---:|
| 下标 | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
| 元素值 | 10 | 20 | 30 | 40 | 50 | 60 | 70 |

# A. 数量、下标与 `n - 1`

## A01. `n` 表示数量，不是最后下标

```python
nums = [10, 20, 30, 40, 50, 60, 70]
n = len(nums)
```

这里 `n = 7`，表示列表里有 7 个元素。

Python 下标从 0 开始，所以 7 个下标是：

```text
0, 1, 2, 3, 4, 5, 6
```

最后一个下标是 6，也就是 `n - 1`。

### 一句话记忆

> 有 n 个元素时，下标从 0 到 n-1。

````python
nums = [10, 20, 30, 40, 50, 60, 70]
n = len(nums)

print('元素数量 n：', n)
print('第一个下标：', 0)
print('最后一个下标 n-1：', n - 1)
print('最后一个元素 nums[n-1]：', nums[n - 1])
print('全部下标：', list(range(n)))
````

## A02. 第 `k` 个元素为什么是下标 `k - 1`

“第几个”通常从 1 开始说，但列表下标从 0 开始。

| 说法 | 第1个 | 第2个 | 第3个 | 第k个 |
|---|---:|---:|---:|---:|
| 下标 | 0 | 1 | 2 | `k-1` |

例如 `k = 3`：

```python
nums[k - 1]  # nums[2]，得到第 3 个元素
```

### 特别容易混淆的两句话

- **第 k 个元素**：下标是 `k - 1`。
- **下标 k 的元素**：它其实是第 `k + 1` 个元素。

所以“第 3 个”和“下标 3”不是同一个位置。

````python
nums = [10, 20, 30, 40, 50, 60, 70]
k = 3

print('第 k 个元素 nums[k-1]：', nums[k - 1])  # 30
print('下标 k 的元素 nums[k]：', nums[k])      # 40
print('它们不是同一个位置')
````

## A03. “前 k 个”为什么对应下标 `0` 到 `k - 1`

还是令 `k = 3`：

```text
前 3 个元素：10  20  30
对应下标：    0   1   2
```

因此：

- 前 `k` 个的第一个下标是 `0`。
- 前 `k` 个的最后下标是 `k - 1`。
- 后面的剩余部分从下标 `k` 开始。

画成两段就是：

```text
前 k 个              剩余部分
0 ... k-1            k ... n-1
```

````python
nums = [10, 20, 30, 40, 50, 60, 70]
n = len(nums)
k = 3

front_indices = list(range(0, k))
remaining_indices = list(range(k, n))

print('前 k 个的下标：', front_indices)
print('剩余部分的下标：', remaining_indices)
print('前 k 个的值：', [nums[i] for i in front_indices])
print('剩余部分的值：', [nums[i] for i in remaining_indices])
````

# B. `range`：右端为什么不包含

## B01. `range(start, stop)` 是左闭右开

```python
range(start, stop)
```

会生成从 `start` 开始、到 `stop` 前面结束的整数。

数学上常写成：

```text
[start, stop)
```

方括号表示包含，圆括号表示不包含。

例如：

```python
range(2, 6)  # 2, 3, 4, 5
```

包含左端 2，不包含右端 6。

````python
print('range(2, 6)：', list(range(2, 6)))
print('包含 2：', 2 in range(2, 6))
print('不包含 6：', 6 in range(2, 6))
print('一共有几个数：', len(range(2, 6)))
````

## B02. `range(n)`、`range(1, n)`、`range(1, n + 1)`

这是最值得放在一起比较的三种写法：

| 写法 | 产生的数 |
|---|---|
| `range(n)` | `0` 到 `n-1` |
| `range(1, n)` | `1` 到 `n-1` |
| `range(1, n+1)` | `1` 到 `n` |

假设 `n = 5`：

```python
range(5)       # 0,1,2,3,4
range(1, 5)    # 1,2,3,4
range(1, 6)    # 1,2,3,4,5
```

### 为什么遍历下标常用 `range(n)`

有 `n` 个元素时，合法下标正好是 `0` 到 `n-1`，与 `range(n)` 完全对应。

````python
n = 5

print('range(n)：', list(range(n)))
print('range(1, n)：', list(range(1, n)))
print('range(1, n+1)：', list(range(1, n + 1)))
````

## B03. 倒着遍历：`range(n - 1, -1, -1)`

完整形式是：

```python
range(start, stop, step)
```

倒着走时 `step = -1`。

如果要遍历全部下标 `n-1, n-2, ..., 0`：

```python
range(n - 1, -1, -1)
```

为什么 `stop` 写 `-1`？因为右端不包含。要想最后拿到 0，就必须让停止边界再往左一格写成 -1。

````python
nums = [10, 20, 30, 40]
n = len(nums)

print('倒序下标：', list(range(n - 1, -1, -1)))
print('倒序元素：')
for i in range(n - 1, -1, -1):
    print(f'下标 {i} -> {nums[i]}')
````

# C. 切片：把 `range` 的规则搬到列表

## C01. `nums[start:stop]` 也是左闭右开

切片：

```python
nums[start:stop]
```

取得下标 `start` 到 `stop-1` 的元素：

- 包含 `start`。
- 不包含 `stop`。

例如：

```python
nums[2:5]
```

取得下标 2、3、4，不取得下标 5。

切片和 `range(2, 5)` 使用的是同一套左闭右开规则。

````python
nums = [10, 20, 30, 40, 50, 60, 70]

print('nums[2:5]：', nums[2:5])
print('对应下标：', list(range(2, 5)))
print('对应元素：', [nums[i] for i in range(2, 5)])
````

## C02. 为什么 `nums[:k]` 是前 k 个

省略 `start` 表示从开头开始：

```python
nums[:k]
```

取得下标：

```text
0, 1, 2, ..., k-1
```

一共正好 `k` 个元素。

对应关系：

| 要什么 | 下标 | 切片 |
|---|---|---|
| 前 `k` 个 | `0` 到 `k-1` | `nums[:k]` |
| 下标 `k` 到末尾 | `k` 到 `n-1` | `nums[k:]` |

注意：`nums[k:]` 从下标 `k` 开始，所以它是从第 `k+1` 个元素开始。

````python
nums = [10, 20, 30, 40, 50, 60, 70]
k = 3

print('前 k 个 nums[:k]：', nums[:k])
print('从下标 k 开始 nums[k:]：', nums[k:])
print('两段拼回原列表：', nums[:k] + nums[k:])
````

## C03. 负数下标：从末尾开始数

| 负下标 | 含义 |
|---|---|
| `-1` | 最后一个 |
| `-2` | 倒数第二个 |
| `-3` | 倒数第三个 |
| `-k` | 倒数第 `k` 个 |

负下标和正下标可以指向同一个位置。列表长度为 `n` 时：

```python
nums[-k] == nums[n-k]
```

例如 `n=7、k=3`：`-3` 和 `7-3=4` 都指向值 50。

````python
nums = [10, 20, 30, 40, 50, 60, 70]
n = len(nums)
k = 3

print('nums[-1]：', nums[-1])
print('nums[-k]：', nums[-k])
print('nums[n-k]：', nums[n-k])
print('两种下标指向同一个值：', nums[-k] == nums[n-k])
````

## C04. `nums[-k:]` 与 `nums[:-k]`

这两段正好把列表切成“最后 `k` 个”和“它前面的部分”：

| 写法 | 含义 |
|---|---|
| `nums[-k:]` | 从倒数第 `k` 个取到末尾，也就是最后 `k` 个 |
| `nums[:-k]` | 从开头取到倒数第 `k` 个之前 |

它们也可以用正下标写：

```python
nums[-k:] == nums[n-k:]
nums[:-k] == nums[:n-k]
```

画下标比背表达式更可靠：

```text
前面部分                 最后 k 个
0 ... n-k-1              n-k ... n-1
```

````python
nums = [10, 20, 30, 40, 50, 60, 70]
n = len(nums)
k = 3

print('最后 k 个 nums[-k:]：', nums[-k:])
print('前面部分 nums[:-k]：', nums[:-k])
print('正下标写法 nums[n-k:]：', nums[n-k:])
print('正下标写法 nums[:n-k]：', nums[:n-k])
````

## C05. 省略位置和步长 `step`

完整切片是：

```python
sequence[start:stop:step]
```

常见写法：

| 写法 | 含义 |
|---|---|
| `nums[:]` | 从头到尾，得到浅拷贝 |
| `nums[::2]` | 每隔 2 个位置取一个 |
| `nums[1::2]` | 从下标 1 开始，每隔 2 个取一个 |
| `nums[::-1]` | 步长为 -1，倒序复制 |

切片通常会创建新列表，所以需要 `O(切片长度)` 的额外空间。

````python
nums = [10, 20, 30, 40, 50, 60, 70]

print('完整复制 nums[:]：', nums[:])
print('偶数下标 nums[::2]：', nums[::2])
print('奇数下标 nums[1::2]：', nums[1::2])
print('倒序 nums[::-1]：', nums[::-1])
print('nums[:] 是新列表：', nums[:] is nums)
````

# D. 最容易混淆的地方：右端到底包不包含

## D01. 先认清两种区间

### 左闭右开 `[left, right)`

包含 `left`，不包含 `right`：

```python
range(left, right)
nums[left:right]
```

### 左闭右闭 `[left, right]`

左右都包含。自己写的双指针函数经常采用这种定义：

```python
reverse_part(nums, left, right)
```

它到底包不包含右端，不是 Python 自动决定的，而是函数作者决定的。一定要看代码或注释。

````python
nums = [10, 20, 30, 40, 50]
left, right = 1, 3

print('左闭右开 nums[left:right]：', nums[left:right])
print('想让切片包含 right：', nums[left:right + 1])

print('range(left, right)：', list(range(left, right)))
print('包含 right 的下标：', list(range(left, right + 1)))
````

## D02. 为什么有时出现 `right + 1`

假设题目说“访问下标 `left` 到 `right`，两端都要包含”。

因为 `range` 和切片不包含右端，所以需要把停止位置多写 1：

```python
range(left, right + 1)
nums[left:right + 1]
```

`right + 1` 不是神秘公式，它只是把“不包含的停止线”放到目标右端的后一格。

````python
nums = [10, 20, 30, 40, 50]
left, right = 1, 3

for i in range(left, right + 1):
    print(f'访问下标 {i}，值 {nums[i]}')

print('同一段切片：', nums[left:right + 1])
````

## D03. 为什么反转前 k 个写 `reverse(0, k - 1)`

假设自己定义的反转函数使用左右都包含的区间 `[left, right]`。

前 `k` 个元素的下标是：

```text
0, 1, ..., k-1
```

所以调用：

```python
reverse_part(nums, 0, k - 1)
```

如果使用切片表达同一段，则写：

```python
nums[:k]
```

对照：

| 同一段“前 k 个” | 写法 |
|---|---|
| 左右都包含的反转函数 | `reverse_part(nums, 0, k-1)` |
| 右端不包含的切片 | `nums[:k]` |

````python
def reverse_part(nums, left, right):
    # 这个函数约定：left 和 right 都包含
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1


nums = [10, 20, 30, 40, 50]
k = 3
print('前 k 个切片：', nums[:k])

reverse_part(nums, 0, k - 1)
print('反转前 k 个后：', nums)
````

# E. 原地修改与列表拷贝

## E01. `nums = ...` 与 `nums[:] = ...`

这两个写法看起来相似，但作用不同：

```python
nums = new_values
```

让变量 `nums` 改为指向另一个列表。

```python
nums[:] = new_values
```

替换原列表中的全部内容，列表对象本身仍是同一个。

LeetCode 要求“原地修改”时，经常需要第二种效果。

````python
nums1 = [1, 2, 3]
outside1 = nums1
nums1 = [3, 2, 1]
print('重新赋值后 nums1：', nums1)
print('外面的旧列表：', outside1)
print('还是同一个对象吗：', nums1 is outside1)

print()

nums2 = [1, 2, 3]
outside2 = nums2
nums2[:] = [3, 2, 1]
print('切片写回后 nums2：', nums2)
print('外面看到的列表：', outside2)
print('还是同一个对象吗：', nums2 is outside2)
````

# F. 综合应用：轮转数组中的所有边界

## F01. 切片版本

向右轮转 `k` 位，就是把最后 `k` 个元素搬到前面：

```python
k %= len(nums)
nums[:] = nums[-k:] + nums[:-k]
```

以 `n=7、k=3` 为例：

```text
前面部分                 最后 k 个
下标 0 ... 3             下标 4 ... 6
     0 ... n-k-1              n-k ... n-1
```

- 最后 `k` 个：`nums[-k:]`。
- 前面部分：`nums[:-k]`。
- 拼接后用 `nums[:]` 写回原列表。

切片版本时间复杂度 `O(n)`，额外空间复杂度 `O(n)`。

````python
def rotate_with_slice(nums, k):
    k %= len(nums)
    nums[:] = nums[-k:] + nums[:-k]


nums = [1, 2, 3, 4, 5, 6, 7]
rotate_with_slice(nums, 3)
print(nums)  # [5, 6, 7, 1, 2, 3, 4]
````

## F02. 三次反转版本中的 `n - 1` 与 `k - 1`

反转函数约定左右端点都包含：

```python
reverse_part(nums, 0, n - 1)  # 整个数组：0 到 n-1
reverse_part(nums, 0, k - 1)  # 前 k 个：0 到 k-1
reverse_part(nums, k, n - 1)  # 剩余部分：k 到 n-1
```

这些表达式都来自同一张下标图：

```text
整个数组：0 ........................ n-1
前 k 个： 0 ........ k-1
剩余部分：             k ........... n-1
```

### `k %= n` 的作用

长度为 `n` 时，轮转 `n` 步会回到原样，所以只保留余数。

- `k > n`：去掉完整转圈的部分。
- `k` 是 `n` 的倍数：余数为 0，不需要移动。

````python
def rotate_with_reversal(nums, k):
    n = len(nums)
    k %= n
    reverse_part(nums, 0, n - 1)
    reverse_part(nums, 0, k - 1)
    reverse_part(nums, k, n - 1)


nums = [1, 2, 3, 4, 5, 6, 7]
rotate_with_reversal(nums, 10)  # 10 % 7 = 3
print(nums)  # [5, 6, 7, 1, 2, 3, 4]
````

## F03. `k = 0` 的特殊观察

在切片里 `-0` 就是 `0`：

```python
nums[-0:] == nums[0:] == nums[:]
nums[:-0] == nums[:0] == []
```

所以两段拼起来仍然是原列表内容。

在三次反转里：

```python
reverse_part(nums, 0, k - 1)
```

会变成 `reverse_part(nums, 0, -1)`。因为循环条件 `left < right` 一开始就是假，不会交换任何元素。

虽然两种代码都能工作，但实际写代码时可以在 `k == 0` 时提前返回，让意图更清楚。

````python
nums = [1, 2, 3]
k = 0

print('nums[-0:]：', nums[-0:])
print('nums[:-0]：', nums[:-0])
print('拼接结果：', nums[-0:] + nums[:-0])
````

# G. 高频写法速查表

| 想要什么 | 写法 |
|---|---|
| 列表元素数量 | `n = len(nums)` |
| 最后一个下标 | `n - 1` |
| 第 `k` 个元素 | `nums[k - 1]` |
| 下标 `k` 的元素 | `nums[k]`，它是第 `k+1` 个 |
| 全部下标 | `range(n)` |
| 普通数字 1 到 n | `range(1, n + 1)` |
| 前 `k` 个元素 | `nums[:k]` |
| 下标 `k` 到末尾 | `nums[k:]` |
| 最后 `k` 个元素 | `nums[-k:]` |
| 除最后 `k` 个以外 | `nums[:-k]` |
| 下标 `left` 到 `right`，包含两端 | `nums[left:right+1]` |
| 列表倒序副本 | `nums[::-1]` |
| 全部内容写回原列表 | `nums[:] = new_values` |

### 真正需要记住的不是整张表

1. 有 `n` 个元素，下标是 `0` 到 `n-1`。
2. `range` 和切片都不包含右端。
3. 遇到 `k-1`、`n-1` 时，先画下标，不硬背。
