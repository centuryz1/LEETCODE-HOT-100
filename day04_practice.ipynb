## 题目描述（改写版）

### 输入是什么

- `intervals` 是一个非空二维列表。
- 每个元素都是 `[start, end]`，并保证 `start <= end`。
- 输入区间不保证已经排序。

### 输出是什么

返回合并后的二维列表，通常按区间起点从小到大排列。

### 关键限制和易漏细节

- `[1, 4]` 和 `[4, 5]` 也算重叠，因为它们在位置 `4` 相接。

- 排序会改变 `intervals` 中区间的顺序；本题允许这样做。
- `1 <= len(intervals) <= 10000`。

输入：`[[1,3],[2,6],[8,10],[15,18]]`

输出：`[[1,6],[8,10],[15,18]]`

解释：两个区间端点相接，也需要合并。
