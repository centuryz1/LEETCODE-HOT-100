## 题目 1 做题前先补：把“今天卖出”拆开看

第 i 天卖出的利润 = 第 i 天的价格 - 之前某一天的最低价格。

所以只要在遍历时维护两个变量：

- `min_price`：目前为止见过的最低价；
- `best`：目前为止的最大利润。

每走到一天，先看看今天是不是新的最低价，然后算一下“如果今天卖，能赚多少”，和历史最佳比一比。

**关键点：卖出必须在买入之后。** 因为我们是从左往右遍历，`min_price` 只包含“今天之前”的价格，所以这个顺序自动满足，不会出现“用后面的价格当买入价”这种错误。


````python
# 手工走一遍 [7, 1, 5, 3, 6, 4]
prices = [7, 1, 5, 3, 6, 4]
min_price = prices[0]
best = 0

for price in prices:
    if price < min_price:
        min_price = price
    profit = price - min_price
    if profit > best:
        best = profit
    print("价格", price, "：最低价变成", min_price, "，今天卖能赚", profit, "，历史最佳", best)

print("最大利润：", best)
````

## 题目 2 做题前先补：只关心“最远能到哪”

先明确题意：你在下标 0，每个位置 `i` 有一个跳跃长度 `nums[i]`，表示从这里最多能往前跳 `nums[i]` 步（可以少跳）。问能不能到达最后一个下标。

**关键洞察：站在位置 i，我不关心“具体跳几步”，只关心“从当前位置能到达的最远下标是多少”。**

策略：一路往右走，维护一个变量 `farthest` = 目前为止能到达的最远下标。

- 走到某个位置 i 时，如果 `i > farthest`，说明这个位置根本到不了，直接返回 `False`；
- 否则用 `i + nums[i]` 尝试更新 `farthest`。

走完全程都没被卡住，就返回 `True`。

**为什么这样贪心是对的？** 因为我们维护的是“所有可达位置里能到达的最远处”，它覆盖了所有可能的选择，不会漏掉任何一条路。


````python
# 手工走一遍 [2, 3, 1, 1, 4]
nums = [2, 3, 1, 1, 4]
farthest = 0

for i in range(len(nums)):
    if i > farthest:
        print("走到", i, "时已经够不到了，失败")
        break
    if i + nums[i] > farthest:
        farthest = i + nums[i]
    print("位置", i, "：能跳到的最远下标变成", farthest)

print("能到达最后一个位置吗：", farthest >= len(nums) - 1)
````

## 题目 3 做题前先补：把跳跃过程看成“一段一段”

第二题只问能不能到，这一题要问**最少几步**。

换个角度理解：站在起点，第一步能覆盖一个区间 `[0, nums[0]]`；在这些位置上，我们能算出“第二步最远能覆盖到哪里”，这又是新的一段。

所以思路是：

- `current_end`：当前这一步的覆盖范围右边界；
- `farthest`：在覆盖范围内，下一步最远能到哪；
- 遍历时不断更新 `farthest`；当 `i` 走到 `current_end` 时，说明当前这一步已经用完了，必须再跳一步，于是 `jumps` 加一，并把 `current_end` 更新成 `farthest`。

**为什么这样是最少的？** 因为我们每次都把“下一步能覆盖的全部位置”都用上，才决定再跳，这相当于每跳一步都尽可能覆盖最远的范围，也就是贪心地扩大可达区间。

注意循环只走到 `len(nums) - 2`：因为一旦到达最后一个下标就不用再跳了。


````python
# 手工走一遍 [2, 3, 1, 1, 4]
nums = [2, 3, 1, 1, 4]
jumps = 0
current_end = 0
farthest = 0

for i in range(len(nums) - 1):
    if i + nums[i] > farthest:
        farthest = i + nums[i]
    print("位置", i, "：下一步最远能到", farthest, "，当前这一步的边界是", current_end)
    if i == current_end:
        jumps = jumps + 1
        current_end = farthest
        print("    必须再跳一步，已跳", jumps, "步，新的边界是", current_end)

print("最少步数：", jumps)
````

## 优化思路

两趟扫描：

1. 第一趟：字典 `last` 记录每个字母最后出现的下标；
2. 第二趟：维护 `start` 和 `end`。每走到 `i`，用 `last[s[i]]` 更新 `end`；当 `i == end` 时切一段，长度是 `end - start + 1`，然后 `start = i + 1`。

为什么这样切出的片段最多？因为只要能切就切——每次 `i == end` 时，当前片段内的字母已经全部收齐，再往后放任何一个字母都会违反规则。时间 O(n)，空间 O(1)（字母表固定 26 个）。

面试表达：

我先遍历一遍字符串，用字典记录每个字母最后出现的下标。然后第二次遍历，维护当前片段的起点 `start` 和右边界 `end`。每走到一个字符，就用它最后出现的位置去更新 `end`，取较大值——这保证了片段会一直延伸到内部所有字母的最远处。当遍历下标正好等于 `end` 时，说明这个片段里的字母都收齐了，不可能再往后扩，于是切一刀，记录长度，并把起点移到下一个位置。时间 O(n)，空间是字母表大小。


````python


def partition_labels_answer(s):
    last = {}
    for i in range(len(s)):
        last[s[i]] = i

    result = []
    start = 0
    end = 0
    for i in range(len(s)):
        if last[s[i]] > end:
            end = last[s[i]]
        if i == end:
            result.append(end - start + 1)
            start = i + 1
    return result


print(partition_labels_answer("ababcbacadefegdehijhklij"))   # [9, 7, 8]
print(partition_labels_answer("eccbbbbdec"))                # [10]
print(partition_labels_answer("a"))                         # [1]
print(partition_labels_answer("ab"))                        # [1, 1]
````
