# Git 常用命令速查

面向 Git 初学者和日常项目使用。

> 本专题中的 Git 命令应在 **VS Code 终端** 中运行，不是在 Markdown 单元格中运行。打开终端：`Ctrl + 反引号`。


## 1. 先记住四个位置

```text
工作区 -> 暂存区 -> 本地仓库 -> GitHub 远程仓库
       git add    git commit    git push
```

| 位置 | 含义 |
|---|---|
| 工作区 | 电脑上正在编辑的文件 |
| 暂存区 | 准备放进下一次提交的修改 |
| 本地仓库 | 已经提交到本机 Git 历史的版本 |
| 远程仓库 | GitHub 上的版本 |


## 2. 每天最常用的流程

```powershell
git status
git add -A
git diff --cached
git commit -m "Add day 05 linked list notes"
git pull --rebase
git push
```

| 命令 | 人话解释 |
|---|---|
| `git status` | 看哪些文件发生了变化 |
| `git add -A` | 把全部新增、修改和删除放入暂存区 |
| `git diff --cached` | 提交前检查即将提交的内容 |
| `git commit -m "说明"` | 在本地保存一个版本 |
| `git pull --rebase` | 把 GitHub 新提交接到本地提交之前 |
| `git push` | 把本地提交上传到 GitHub |

自己的学习仓库通常掌握这一组就能完成大部分操作。


## 3. 第一次使用 Git

### 配置身份

```powershell
git config --global user.name "你的名字"
git config --global user.email "你的邮箱"
```

查看配置：

```powershell
git config --global --list
```


## 4. 创建或下载仓库

### 把当前文件夹变成 Git 仓库

```powershell
git init
```

### 从 GitHub 下载已有仓库

```powershell
git clone https://github.com/用户名/仓库名.git
cd 仓库名
```

已经克隆的仓库不需要再次执行 `git init`。


## 5. 查看当前状态

```powershell
git status
git status -sb
```

- `git status`：详细状态。
- `git status -sb`：简洁状态，并显示当前分支。

常见标记：

| 标记 | 含义 |
|---|---|
| `M` | 文件被修改 |
| `A` | 新文件已加入暂存区 |
| `D` | 文件被删除 |
| `??` | Git 还没有跟踪的新文件 |


## 6. 查看具体修改

```powershell
git diff
git diff --cached
git diff HEAD
```

| 命令 | 查看什么 |
|---|---|
| `git diff` | 还没有 `add` 的修改 |
| `git diff --cached` | 已经 `add`、即将提交的修改 |
| `git diff HEAD` | 工作区和最近一次提交的全部差异 |


## 7. 把修改放进暂存区

```powershell
git add 文件名
git add day05_practice.ipynb
git add -A
```

- 只提交一个文件：使用 `git add 文件名`。
- 确认全部修改都属于本次任务：使用 `git add -A`。
- 不确定时先执行 `git status`，不要闭眼 `add`。


## 8. 创建本地提交

```powershell
git commit -m "Add Git command reference"
```

好的提交说明应该简短说明“做了什么”：

```text
Add day 05 linked list exercises
Fix longest substring example
Update notebook generation rules
```

`commit` 只保存到本地，不等于已经上传 GitHub。


## 9. 查看提交历史

```powershell
git log
git log --oneline -10
git log --oneline --graph --decorate --all
```

- `--oneline`：每次提交只显示一行。
- `-10`：只看最近 10 次。
- `--graph`：显示分支关系。


## 10. 从 GitHub 获取更新

```powershell
git fetch
git pull
git pull --rebase
```

| 命令 | 含义 |
|---|---|
| `git fetch` | 只下载远程信息，不修改当前文件 |
| `git pull` | 下载并合并到当前分支 |
| `git pull --rebase` | 下载后把本地提交接到远程提交之后，历史更直 |

拉取前先运行 `git status`。工作区很乱时，先提交或暂存修改。


## 11. 上传到 GitHub

```powershell
git push
```

第一次推送新分支：

```powershell
git push -u origin 分支名
```

`-u` 会建立跟踪关系，以后直接运行 `git push` 即可。


## 12. 远程仓库

```powershell
git remote -v
git remote add origin https://github.com/用户名/仓库名.git
git remote set-url origin 新地址
```

- `origin` 是远程仓库的常用别名，不是固定关键词。
- 已经存在 `origin` 时，不要重复执行 `remote add origin`。


## 13. 分支操作

```powershell
git branch
git switch -c feature/git-notes
git switch main
git merge feature/git-notes
git branch -d feature/git-notes
```

| 命令 | 含义 |
|---|---|
| `git branch` | 查看本地分支 |
| `git switch -c 新分支` | 创建并进入新分支 |
| `git switch 分支名` | 切换分支 |
| `git merge 分支名` | 把目标分支合并到当前分支 |
| `git branch -d 分支名` | 删除已经合并的本地分支 |


## 14. 安全撤销

### 取消暂存，但保留文件修改

```powershell
git restore --staged 文件名
```

### 丢弃尚未暂存的修改

```powershell
git restore 文件名
```

第二条会丢失代码，运行前一定先看 `git diff`。


## 15. 撤销提交

### 最近一次提交说明写错了

```powershell
git commit --amend -m "新的提交说明"
```

### 撤销最近一次本地提交，但保留文件修改

```powershell
git reset --soft HEAD~1
```

### 撤销已经推送的提交

```powershell
git revert 提交编号
```

已经推送或多人共享的提交优先使用 `git revert`，因为它会创建新的反向提交，不会改写公共历史。


## 16. 临时收起修改：`stash`

```powershell
git stash push -m "未完成的 day05 修改"
git stash list
git stash pop
```

适合：当前修改还不想提交，但临时需要切换分支。

- `stash push`：收起修改。
- `stash list`：查看暂存记录。
- `stash pop`：恢复最近一次修改，并删除对应 stash。


## 17. 解决冲突

发生冲突后：

```powershell
git status
```

打开冲突文件，会看到：

```text
<<<<<<< HEAD
本地内容
=======
远程内容
>>>>>>> other-branch
```

处理步骤：

1. 决定最终保留的内容。
2. 删除 `<<<<<<<`、`=======`、`>>>>>>>` 标记。
3. 保存文件。
4. 执行 `git add 文件名`。
5. 按终端提示继续提交或执行 `git rebase --continue`。


## 18. `.gitignore`

`.gitignore` 用于告诉 Git 不要跟踪哪些文件：

```gitignore
__pycache__/
*.pyc
.venv/
.ipynb_checkpoints/
.env
```

已经被 Git 跟踪的文件，后来写进 `.gitignore` 不会自动停止跟踪。

不要提交：密码、Token、虚拟环境、缓存、大型临时数据和无用运行产物。


## 19. 危险命令

看到下面命令先停一下：

```text
git reset --hard
git clean -fd
git push --force
```

它们可能删除本地修改、未跟踪文件或改写远程历史。初学阶段不要根据网上的一行答案直接执行；先运行 `git status` 和 `git diff`，确认影响范围。


## 20. 当前学习仓库的推荐推送流程

```powershell
git status
git diff
git add -A
git diff --cached
git commit -m "Add today's study notes"
git pull --rebase
git push
```

如果 `git status` 显示了不认识的修改，先不要提交，先确认这些文件来自哪里。


## 21. 高频问题

### `nothing to commit`

没有新修改，或者修改还没有保存。

### `non-fast-forward`

GitHub 上有本地没有的提交。先运行：

```powershell
git pull --rebase
git push
```

### `not a git repository`

当前终端不在 Git 仓库目录中。使用 `cd` 进入项目目录。

### `remote origin already exists`

已经配置过 `origin`。使用 `git remote -v` 查看，不要重复添加。


## 22. 一页速查

| 目的 | 命令 |
|---|---|
| 查看状态 | `git status -sb` |
| 查看未暂存修改 | `git diff` |
| 暂存一个文件 | `git add 文件名` |
| 暂存全部变化 | `git add -A` |
| 查看待提交内容 | `git diff --cached` |
| 本地提交 | `git commit -m "说明"` |
| 拉取更新 | `git pull --rebase` |
| 上传 GitHub | `git push` |
| 查看历史 | `git log --oneline -10` |
| 创建分支 | `git switch -c 分支名` |
| 切换分支 | `git switch 分支名` |
| 取消暂存 | `git restore --staged 文件名` |
| 临时收起修改 | `git stash push -m "说明"` |
| 查看远程地址 | `git remote -v` |

最重要的一条习惯：**执行会修改历史或删除内容的命令前，先看 `git status` 和 `git diff`。**
