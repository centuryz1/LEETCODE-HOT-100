## B02. 列表 `list`

列表是刷算法题最常用的数据结构之一：有顺序、允许重复、可以修改。

```python
nums = [3, 1, 4]
```

常用操作：

| 操作 | 作用 |
|---|---|
| `nums.append(x)` | 在末尾添加一个元素 |
| `nums.extend(items)` | 添加多个元素 |
| `nums.pop()` | 删除并返回最后一个元素 |
| `nums.pop(i)` | 删除并返回下标 `i` 的元素 |
| `nums.remove(x)` | 删除第一个值为 `x` 的元素 |
| `nums.insert(i, x)` | 在下标 `i` 插入元素 |
| `nums.reverse()` | 原地反转 |
| `nums.sort()` | 原地排序 |
| `len(nums)` | 元素数量 |

`append([3, 4])` 会把整个列表作为一个元素；`extend([3, 4])` 会分别加入 `3` 和 `4`。

````python
nums = [3, 1, 4]
nums.append(2)
print("append 后：", nums)

last = nums.pop()
print("pop 出来：", last)

nums.extend([5, 9])
print("extend 后：", nums)

assert nums == [3, 1, 4, 5, 9]
````

## B03. 切片 `sequence[start:stop:step]`

切片适用于字符串、列表和元组。

```python
nums = [0, 1, 2, 3, 4, 5]

nums[1:4]    # [1, 2, 3]，不包含下标 4
nums[:3]     # [0, 1, 2]
nums[3:]     # [3, 4, 5]
nums[::2]    # [0, 2, 4]
nums[::-1]   # [5, 4, 3, 2, 1, 0]
```

记住：切片通常**左闭右开**，包含 `start`，不包含 `stop`。

切片会创建新对象，长度为 `k` 的列表切片通常需要 `O(k)` 时间和空间。

````python
nums = [0, 1, 2, 3, 4, 5]

print(nums[1:4])
print(nums[::2])
print(nums[::-1])

assert nums[1:4] == [1, 2, 3]
````

## B04. 元组 `tuple` 与解包

元组有顺序，但创建后不能修改。

```python
point = (3, 5)
x, y = point
```

解包在算法题中很常见：

```python
left, right = 0, len(nums) - 1
left, right = right, left  # 交换
```

只有一个元素的元组必须写逗号：

```python
single = (1,)
```

元组可以作为字典键或集合元素，前提是其中所有元素都可哈希。

````python
point = (3, 5)
x, y = point
print(x, y)

left, right = 1, 9
left, right = right, left
print(left, right)

assert (left, right) == (9, 1)
````

## B05. 字典 `dict`

字典保存“键 → 值”的映射。

```python
scores = {"小明": 90, "小红": 95}
```

常用操作：

```python
scores["小明"]                 # 读取，键不存在会报错
scores.get("小刚", 0)          # 不存在时返回默认值 0
scores["小刚"] = 88            # 新增或修改
"小明" in scores               # 判断键是否存在
scores.pop("小明")              # 删除并返回对应值
scores.keys()                   # 所有键
scores.values()                 # 所有值
scores.items()                  # 所有键值对
```

遍历键值对：

```python
for name, score in scores.items():
    print(name, score)
```

字典查询、添加和删除平均为 `O(1)`。

````python
words = ["a", "b", "a", "c", "a", "b"]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
for word, count in counts.items():
    print(word, count)

assert counts == {"a": 3, "b": 2, "c": 1}
````

## B06. 集合 `set`

集合只保存不重复元素，适合去重和快速判断是否存在。

```python
seen = {1, 2, 3}
```

常用操作：

```python
seen.add(4)
seen.remove(2)     # 不存在会报错
seen.discard(2)    # 不存在也不报错
3 in seen
```

集合运算：

```python
a | b   # 并集
a & b   # 交集
a - b   # 差集
```

空集合必须写成 `set()`；`{}` 表示空字典。

集合查询和添加平均为 `O(1)`。

````python
nums = [1, 2, 2, 3, 3, 3]
unique = set(nums)

a = {1, 2, 3}
b = {3, 4, 5}

print("去重：", unique)
print("交集：", a & b)
print("并集：", a | b)

assert unique == {1, 2, 3}
````

## B07. 如何选择容器？

| 需求 | 优先考虑 |
|---|---|
| 需要顺序、重复、按下标访问 | `list` |
| 固定的一组值、不希望修改 | `tuple` |
| 保存键和值的对应关系 | `dict` |
| 去重或快速判断存在 | `set` |
| 频繁从队头和队尾操作 | `collections.deque` |

刷题时可以先问：

1. 是否需要保留顺序？
2. 是否允许重复？
3. 是否需要按下标访问？
4. 是否需要快速判断元素是否出现过？
5. 是否需要保存“某个值对应什么信息”？

# C. 条件和循环

## C01. `if`、`elif`、`else`

```python
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"
```

一条 `if-elif-else` 分支链只执行第一个满足条件的分支。

多个独立 `if` 可能执行多次：

```python
if number > 0:
    print("正数")
if number % 2 == 0:
    print("偶数")
```

选择原则：

- 条件互斥，只想选一个结果：`if/elif/else`。
- 多个条件需要分别检查：多个独立 `if`。

````python
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "D"


assert get_grade(95) == "A"
assert get_grade(85) == "B"
print(get_grade(72))
````

## C02. 真值与短路判断

以下值在条件中会被看作假：

```python
False
None
0
0.0
""
[]
{}
set()
```

所以：

```python
if nums:
    print("列表不为空")
```

短路判断：

```python
if left < len(nums) and nums[left] == target:
    ...
```

Python 先判断左边。左边为假时，不会再执行右边，因此可以避免下标越界。

````python
empty_values = [False, None, 0, 0.0, "", [], {}, set()]

for value in empty_values:
    print(repr(value), "->", bool(value))

assert all(not value for value in empty_values)
````

## C03. `for` 与 `range`

`for` 适合遍历已有序列或明确次数：

```python
for number in nums:
    print(number)
```

`range` 常见形式：

```python
range(5)        # 0, 1, 2, 3, 4
range(2, 5)     # 2, 3, 4
range(5, 0, -1) # 5, 4, 3, 2, 1
```


## C04. `while`：条件满足就继续

```python
left = 0
right = len(nums) - 1

while left < right:
    ...
```

`while` 适合循环次数事先不确定，但知道继续条件的场景。

与 `if` 的区别：

- `if`：条件满足，执行一次；
- `while`：条件满足，反复执行。

写 `while` 时检查三件事：

1. 初始状态是什么？
2. 什么时候停止？
3. 每轮怎样靠近停止条件？

第三点缺失时容易死循环。

````python
def digits_count(number):
    if number == 0:
        return 1

    count = 0
    number = abs(number)
    while number > 0:
        number //= 10
        count += 1
    return count


assert digits_count(0) == 1
assert digits_count(12345) == 5
print(digits_count(-987))
````

## C05. `break`、`continue`、`pass`

### `break`

立即结束整个循环。

### `continue`

跳过本轮剩余代码，进入下一轮。

### `pass`

什么也不做，只是语法占位，不会结束或跳过循环。

```python
for number in nums:
    if number < 0:
        continue
    if number == target:
        break
    print(number)
```

注意：在 `while` 中使用 `continue` 前，必须确认循环变量已经更新，否则可能死循环。

````python
nums = [-2, 1, 3, 5, 7]
kept = []

for number in nums:
    if number < 0:
        continue
    if number == 5:
        break
    kept.append(number)

print(kept)
assert kept == [1, 3]
````

## C06. `enumerate` 与 `zip`

### 同时获得下标和元素

```python
for i, value in enumerate(nums):
    print(i, value)
```

### 同时遍历多个序列

```python
for name, score in zip(names, scores):
    print(name, score)
```

`zip` 会在最短的序列结束时停止。

这两个工具通常比手动维护下标更清楚。

````python
names = ["小明", "小红", "小刚"]
scores = [90, 95, 88]

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(index, name, score)

assert list(zip([1, 2], ["a", "b"])) == [(1, "a"), (2, "b")]
````

```python
def add(a, b):
```

## D02. 参数

```python
def greet(name, prefix="你好"):
    return f"{prefix}，{name}"
```

- `name` 是必传参数；
- `prefix` 有默认值，是默认参数；
- 调用时可以使用关键字参数：`greet(name="小明", prefix="早上好")`。

危险写法：

```python
def add_item(item, items=[]):
    items.append(item)
    return items
```

默认列表会在多次调用之间共享。安全写法：

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

````python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items


first = add_item("a")
second = add_item("b")
print(first, second)
assert first == ["a"] and second == ["b"]
````

## D03. 作用域

函数内部创建的变量通常是局部变量：

```python
def calculate():
    result = 10

```

函数外不能直接访问内部的 `result`。

刷题时推荐：

- 尽量通过参数传入数据；
- 通过 `return` 返回结果；
- 少依赖全局变量；
- 递归题使用外层变量时，理解 `nonlocal` 后再使用。

专业名称：**局部作用域（Local Scope）**、**全局作用域（Global Scope）**。

# E. 刷题常用 Python 写法

## E01. 列表、集合与字典推导式

普通循环：

```python
squares = []
for number in nums:
    squares.append(number * number)
```

列表推导式：

```python
squares = [number * number for number in nums]
```

带筛选条件：

```python
even_squares = [x * x for x in nums if x % 2 == 0]
```

集合和字典推导式：

```python
unique_lengths = {len(word) for word in words}
index_map = {value: i for i, value in enumerate(nums)}
```

推导式过长时应改回普通循环，清晰比炫技重要。

````python
nums = [1, 2, 3, 4, 5]

squares = [number * number for number in nums]
even_squares = [number * number for number in nums if number % 2 == 0]
index_map = {value: index for index, value in enumerate(nums)}

print(squares)
print(even_squares)
print(index_map)

assert even_squares == [4, 16]
````

## E02. 排序：`sort`、`sorted`、`key`

```python
nums.sort()          # 原地修改，返回 None
new_nums = sorted(nums)  # 返回新列表
```

倒序：

```python
sorted(nums, reverse=True)
```

按指定规则排序：

```python
words = ["pear", "a", "banana"]
sorted(words, key=len)
```

对二元组排序：

```python
pairs = [(1, 5), (3, 2), (2, 8)]
sorted(pairs, key=lambda item: item[1])
```

`lambda` 是短小的匿名函数。上面表示“用每个二元组的第二项作为排序键”。

Python 排序是稳定排序：排序键相同时，原来的相对顺序会保留。

````python
words = ["pear", "a", "banana", "kiwi"]
pairs = [(1, 5), (3, 2), (2, 8)]

print(sorted(words, key=len))
print(sorted(pairs, key=lambda item: item[1]))

assert sorted(words, key=len)[0] == "a"
assert sorted(pairs, key=lambda item: item[1])[0] == (3, 2)
````

## E03. 常用内置函数

```python
len(nums)
sum(nums)
min(nums)
max(nums)
abs(number)
```

带生成表达式：

```python
sum(x * x for x in nums)
max(len(word) for word in words)
```

条件汇总：

```python
any(x < 0 for x in nums)  # 是否至少一个满足

```

同时获得最大值所在下标：

```python
index = max(range(len(nums)), key=lambda i: nums[i])
```

````python
nums = [3, -1, 8, 2]

print(sum(nums), min(nums), max(nums))
print(any(number < 0 for number in nums))
print(all(number > 0 for number in nums))

max_index = max(range(len(nums)), key=lambda i: nums[i])
print("最大值下标：", max_index)
assert max_index == 2
````

## E04. 正无穷与负无穷

```python
answer = float("inf")
```

```python
answer = float("-inf")
```

这样不必猜一个“足够大”或“足够小”的数字。

````python
nums = [8, 3, 12, -5]

minimum = float("inf")
maximum = float("-inf")

for number in nums:
    minimum = min(minimum, number)
    maximum = max(maximum, number)

print(minimum, maximum)
assert (minimum, maximum) == (-5, 12)
````

## F02. `defaultdict`：自动提供默认值

普通字典分组：

```python
groups = {}
for key, value in items:
    if key not in groups:
        groups[key] = []
    groups[key].append(value)
```

使用 `defaultdict`：

```python
from collections import defaultdict

groups = defaultdict(list)
for key, value in items:
    groups[key].append(value)
```

常用形式：

```python
defaultdict(int)   # 默认 0
defaultdict(list)  # 默认 []
defaultdict(set)   # 默认 set()
```

括号中传的是函数 `list`，不是已经创建好的 `list()`。

````python
from collections import defaultdict

items = [("水果", "苹果"), ("蔬菜", "白菜"), ("水果", "香蕉")]
groups = defaultdict(list)

for category, item in items:
    groups[category].append(item)

print(dict(groups))
assert groups["水果"] == ["苹果", "香蕉"]
````

## F03. `deque`：双端队列

```python
from collections import deque

queue = deque()
queue.append(x)       # 右侧加入
queue.appendleft(x)   # 左侧加入
queue.pop()           # 右侧弹出
queue.popleft()       # 左侧弹出
```

为什么队列不用列表的 `pop(0)`？

- `list.pop(0)` 需要移动后续元素，时间复杂度 `O(n)`；
- `deque.popleft()` 是 `O(1)`。

广度优先搜索（BFS）和单调队列经常使用 `deque`。

````python
from collections import deque

queue = deque([1, 2, 3])
queue.append(4)
first = queue.popleft()

print(first, list(queue))
assert first == 1
assert list(queue) == [2, 3, 4]
````

## F04. `heapq`：最小堆

堆适合反复获取当前最小值。

```python
import heapq

heap = []
heapq.heappush(heap, 5)
heapq.heappush(heap, 2)
smallest = heapq.heappop(heap)
```

Python 的 `heapq` 默认是最小堆。

模拟最大堆时，常见做法是存负数：

```python
heapq.heappush(heap, -value)
largest = -heapq.heappop(heap)
```

常用于 Top K、合并多个有序序列、优先队列。

````python
import heapq

heap = []
for number in [5, 2, 8, 1, 3]:
    heapq.heappush(heap, number)

ordered = []
while heap:
    ordered.append(heapq.heappop(heap))

print(ordered)
assert ordered == [1, 2, 3, 5, 8]
````

## F05. 栈、队列、哈希表在 Python 中怎么表示？

| 数据结构 | Python 常用实现 |
|---|---|
| 栈 Stack | `list`，使用 `append` 和 `pop` |
| 队列 Queue | `collections.deque` |
| 哈希表 Hash Table | `dict`、`set` |
| 最小堆 Min Heap | `heapq` + `list` |
| 计数器 | `collections.Counter` |

```python
a = [1, 2]

```

此时 `a` 也变成 `[1, 2, 3]`，因为 `a` 和 `b` 指向同一个列表。

## G02. 浅拷贝和深拷贝

一维列表常用浅拷贝：

```python
copy_a = nums[:]
copy_b = nums.copy()
copy_c = list(nums)
```

嵌套列表要小心：

```python
matrix_copy = matrix.copy()
```

只复制了最外层，内部行列表仍然共享。

完全独立复制嵌套对象：

```python
import copy
matrix_copy = copy.deepcopy(matrix)
```

专业名称：**浅拷贝（Shallow Copy）**、**深拷贝（Deep Copy）**。

````python
import copy

matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
deep = copy.deepcopy(matrix)

shallow[0][0] = 99
print("原列表：", matrix)
print("深拷贝：", deep)

assert matrix[0][0] == 99
assert deep[0][0] == 1
````

## G03. 二维列表的经典坑

错误写法：

```python
matrix = [[0] * 3] * 3
```

三行实际上指向同一个内部列表，修改一个位置可能影响多行。

正确写法：

```python
matrix = [[0] * 3 for _ in range(3)]
```

每次循环都会创建新的行列表。

````python
wrong = [[0] * 3] * 3
wrong[0][0] = 1
print("错误创建：", wrong)

correct = [[0] * 3 for _ in range(3)]
correct[0][0] = 1
print("正确创建：", correct)

assert wrong == [[1, 0, 0], [1, 0, 0], [1, 0, 0]]
assert correct == [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
````

# H. LeetCode 与 ACM 形式

## H01. LeetCode 函数形式

```python
class Solution:
    def twoSum(self, nums, target):
        seen = {}
        for i, number in enumerate(nums):
            need = target - number
            if need in seen:
                return [seen[need], i]
            seen[number] = i
```

## I02. 用变量跟踪调试

调试循环时重点观察：

1. 初始值是否正确？
2. 条件是否应该继续？
3. 每轮哪些变量发生变化？

````python
def find_first_even(nums, verbose=False):
    for i, number in enumerate(nums):
        if verbose:
            print(f"i={i}, number={number}")
        if number % 2 == 0:
            return i
    return -1


index = find_first_even([1, 3, 7, 8, 10], verbose=True)
print("答案下标：", index)
assert index == 3
````

先独立完成，再运行下一单元。

````python
def count_even_answer(nums):
    count = 0
    for number in nums:
        if number % 2 == 0:
            count += 1
    return count


def char_count_answer(text):
    counts = {}
    for char in text:
        counts[char] = counts.get(char, 0) + 1
    return counts


def unique_in_order_answer(nums):
    seen = set()
    answer = []
    for number in nums:
        if number not in seen:
            seen.add(number)
            answer.append(number)
    return answer


def sort_by_second_answer(pairs):
    return sorted(pairs, key=lambda item: item[1])


def valid_parentheses_answer(text):
    balance = 0
    for char in text:
        if char == "(":
            balance += 1
        else:
            balance -= 1
        if balance < 0:
            return False
    return balance == 0


assert count_even_answer([1, 2, 4, 7]) == 2
assert char_count_answer("aba") == {"a": 2, "b": 1}
assert unique_in_order_answer([3, 1, 3, 2, 1]) == [3, 1, 2]
assert sort_by_second_answer([("a", 3), ("b", 1), ("c", 2)]) == [("b", 1), ("c", 2), ("a", 3)]
assert valid_parentheses_answer("(())") is True
assert valid_parentheses_answer("(()") is False
print("综合练习参考答案测试通过")
````
