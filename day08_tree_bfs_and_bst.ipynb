## 题目 1 做题前先补：`deque` 与“一层有多少节点”

树的层序遍历使用队列，先进队的节点先处理：

- `queue = deque([root])`：根节点入队。
- `queue.popleft()`：从队首取节点。
- `queue.append(child)`：孩子从队尾入队。

每轮开始先记录 `level_size = len(queue)`。接下来恰好弹出 `level_size` 次，这些节点就属于同一层；循环过程中加入的孩子留给下一层。

```python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(values):
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    index = 1
    while queue and index < len(values):
        node = queue.popleft()
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)
        index += 1
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)
        index += 1
    return root

queue = deque([build_tree([3,9,20,None,None,15,7])])
level_size = len(queue)
print("当前层节点数：", level_size)
node = queue.popleft()
queue.append(node.left)
queue.append(node.right)
print("下一层节点数：", len(queue))
```

# 题目 1：二叉树的层序遍历

LeetCode 102. Binary Tree Level Order Traversal

## 题目描述（改写版）

给你一棵二叉树，按照从上到下、每层从左到右的顺序访问节点，并把每一层的节点值单独放进一个列表返回。

### 输入、输出与限制

- 空树返回空列表。
- 输出是二维列表，外层顺序代表层数。
- 同一层必须保持从左到右。

### 示例

输入 `[3,9,20,None,None,15,7]`，返回 `[[3],[9,20],[15,7]]`。

## 解法名称

专业名称：**广度优先搜索（Breadth-First Search, BFS）**，也叫层序遍历。

## 思路

用队列保存待访问节点。每轮固定处理开始时队列里的 `level_size` 个节点，把它们的值放进当前层，并把非空孩子加入队尾。

时间复杂度 `O(n)`；队列最宽时空间复杂度 `O(w)`，`w` 是树的最大宽度。

```python
root = build_tree([3,9,20,None,None,15,7])
queue = deque([root])
while queue:
    level = []
    level_size = len(queue)
    print("本层开始时队列长度：", level_size)
    for _ in range(level_size):
        node = queue.popleft()
        level.append(node.val)
        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)
    print("本层结果：", level)
```

## 题目 2 做题前先补：BST 不是只比较父子节点

二叉搜索树（BST）要求：

- 左子树所有值都严格小于当前节点。
- 右子树所有值都严格大于当前节点。
- 左右子树本身也必须是 BST。

因此右子树里的某个深层节点不仅要大于它的父节点，还必须大于更上层祖先给出的下界。递归时传递 `(lower, upper)` 才能保存所有祖先约束。

Python 可用 `float('-inf')` 和 `float('inf')` 表示初始无限边界。

```python
lower = float("-inf")
upper = float("inf")
value = 5
print(lower < value < upper)  # True

# 进入节点 5 的左子树后，上界收紧为 5
upper = 5
print(lower < 4 < upper)      # True
print(lower < 6 < upper)      # False
```

# 题目 2：验证二叉搜索树

LeetCode 98. Validate Binary Search Tree

## 题目描述（改写版）

给你一棵二叉树，判断它是否满足严格的二叉搜索树规则：任意节点左子树所有值都更小，右子树所有值都更大，并且所有子树也满足同样规则。

### 输入、输出与限制

- 返回布尔值。
- BST 中不允许出现与当前节点相等的值。
- 不能只检查直接父子关系，还要满足祖先范围。

### 示例

`[2,1,3]` 返回 `True`；`[5,1,4,None,None,3,6]` 返回 `False`，因为节点 3 位于根节点 5 的右子树，却小于 5。

## 解法名称

专业名称：**递归上下界约束（Recursive Bounds Validation）**。

## 容易出错的思路

只比较 `node.left.val < node.val < node.right.val` 会漏掉深层节点违反祖先约束的情况。

## 正确思路

递归函数接收当前节点允许的开区间 `(lower, upper)`。访问左子树时把上界收紧为当前值；访问右子树时把下界收紧为当前值。

时间复杂度 `O(n)`，递归栈空间复杂度 `O(h)`。

```python
def valid_trace(node, lower, upper, depth=0):
    if node is None:
        return True
    print("  " * depth + f"检查 {node.val} 是否在 ({lower}, {upper})")
    if not lower < node.val < upper:
        return False
    return valid_trace(node.left, lower, node.val, depth + 1) and valid_trace(node.right, node.val, upper, depth + 1)

print(valid_trace(build_tree([5,1,4,None,None,3,6]), float("-inf"), float("inf")))
```

# 题目 3：二叉搜索树中第 K 小的元素

LeetCode 230. Kth Smallest Element in a BST

## 题目描述（改写版）

给你一棵合法二叉搜索树的根节点 `root` 和整数 `k`，返回整棵树中第 `k` 小的节点值。

### 输入、输出与限制

- `1 <= k <= 节点总数`，保证答案存在。
- BST 节点值互不相同。
- 返回整数，不需要修改树。

### 示例

BST `[3,1,4,None,2]` 的中序序列为 `[1,2,3,4]`；`k=1` 时返回 1。

## 解法名称

专业名称：**BST 中序遍历（BST Inorder Traversal）**。

## 收集排序

遍历所有节点并排序，时间 `O(n log n)`、空间 `O(n)`，没有利用 BST 性质。

## 优化思路

迭代中序遍历得到递增顺序。每访问一个节点就减少 `k`，到第 `k` 个时立即返回，不必访问剩余节点。

最坏时间 `O(n)`，空间 `O(h)`。如果很早找到答案，实际只访问部分节点。

```python
bst = build_tree([5,3,6,2,4,None,None,1])
stack, current, count = [], bst, 3
while current is not None or stack:
    while current is not None:
        stack.append(current)
        current = current.left
    current = stack.pop()
    count -= 1
    print("访问：", current.val, "还差：", count)
    if count == 0:
        print("答案：", current.val)
        break
    current = current.right
```

```python
def kth_smallest_answer(root, k):
    stack = []
    current = root
    while current is not None or stack:
        while current is not None:
            stack.append(current)
            current = current.left
        current = stack.pop()
        k -= 1
        if k == 0:
            return current.val
        current = current.right
```

- 返回节点对象，不是节点值。
- 这不是 BST，不能根据数值大小选择方向。

### 示例

先记录每个节点的父节点，再向上寻找共同祖先，需要 `O(n)` 额外映射。
