## 题目 1 做题前先补：把数组当成链表

这题有个非常巧的转化。数组里每个数字都在 1 到 n 之间，长度是 n+1，所以**必然有重复**。

现在把数组看成一个“链表”：

- 下标 `i` 是一个节点；
- `nums[i]` 是“下一个节点的下标”。

比如 `nums = [1, 3, 4, 2, 2]`，从下标 0 出发：

```text
0 -> nums[0] = 1 -> nums[1] = 3 -> nums[3] = 2 -> nums[2] = 4 -> nums[4] = 2 -> ...
```

走到 2 之后就一直在 `2 -> 4 -> 2 -> 4` 里打转，形成一个环。

**关键结论：环的入口下标，就是那个重复的数字。**

为什么？因为有两个不同的下标都指向同一个值，也就是“两个节点都指向同一个下一跳”，这正是环入口的特征。

于是问题变成 Day 05 学过的“环形链表 II”：**用快慢指针找环的入口。**

复习一下那两步：

1. 快指针每次走两步、慢指针每次走一步，直到相遇；
2. 让一个指针回到起点，两个指针每次各走一步，再次相遇的位置就是环的入口。


````python
# 手工走一遍快慢指针
nums = [1, 3, 4, 2, 2]
slow = nums[0]
fast = nums[0]

print("阶段一：找相遇点")
while True:
    slow = nums[slow]
    fast = nums[nums[fast]]
    print("    slow =", slow, "，fast =", fast)
    if slow == fast:
        break

print("阶段二：找环入口（也就是重复数字）")
slow = nums[0]
while slow != fast:
    slow = nums[slow]
    fast = nums[fast]
    print("    slow =", slow, "，fast =", fast)
print("重复的数字是：", slow)
````

## 题目 2 做题前先补：二分查找的另一半模板

Day 09 学过“查找某个值”的二分。这一题是**找最小值**，写法略有不同：

```python
left = 0
right = len(nums) - 1
while left < right:            # 注意这里没有等号
    mid = (left + right) // 2
    if nums[mid] > nums[right]:
        left = mid + 1
    else:
        right = mid
return nums[left]
```

两个关键点：

1. **循环条件是 `left < right`**，因为最后我们要的是“唯一剩下的那个位置”，不能用 `left <= right`（那样会多算一轮）。
2. **比较对象是 `nums[right]`，不是 `nums[left]`。**

为什么和右边比？因为数组是“旋转过的升序数组”，比如 `[4,5,6,7,0,1,2]`：

- 如果 `nums[mid] > nums[right]`，说明中点在“较大的一段”里，最小值一定在 `mid` 的右边（不含 mid）；
- 否则说明中点在“较小的一段”里，最小值在 `mid` 或它左边。

**记忆口诀：和右边比，大于就往右走，否则往左收。**


````python
# 手工走一遍 [4, 5, 6, 7, 0, 1, 2]
nums = [4, 5, 6, 7, 0, 1, 2]
left = 0
right = len(nums) - 1

while left < right:
    mid = (left + right) // 2
    if nums[mid] > nums[right]:
        print("nums[" + str(mid) + "] =", nums[mid], "比右边大，最小值在右边，left =", mid + 1)
        left = mid + 1
    else:
        print("nums[" + str(mid) + "] =", nums[mid], "不大于右边，最小值在左边或就是它，right =", mid)
        right = mid

print("最小值是：", nums[left])
````

## 参考答案：寻找旋转排序数组中的最小值

```python
def find_min_answer(nums):
    left = 0
    right = len(nums) - 1

    while left < right:
        mid = (left + right) // 2
        if nums[mid] > nums[right]:

```

面试表达：

我用二分查找。旋转后的数组可以看成两段升序，最小值就是第二段的开头，它一定比最后一段的末尾小。所以每次拿中间元素和右边界比较：如果中间比右边大，说明中间还在第一段（较大的一段），最小值一定在它右边，把左边界移到 mid 加一；否则中间已经在第二段，最小值在 mid 或者它左边，把右边界收到 mid。因为循环条件是 left 小于 right，结束时 left 和 right 重合在最小值上。时间 O(log n)，空间 O(1)。


````python


def find_min_answer(nums):
    left = 0
    right = len(nums) - 1

    while left < right:
        mid = (left + right) // 2
        if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid

    return nums[left]


print(find_min_answer([3, 4, 5, 1, 2]))        # 1
print(find_min_answer([4, 5, 6, 7, 0, 1, 2]))  # 0
print(find_min_answer([11, 13, 15, 17]))       # 11
print(find_min_answer([2, 1]))                 # 1
````

## 题目 3 做题前先补：顺时针旋转 90 度等于“转置 + 每行反转”

先看一个 3×3 的例子，原矩阵：

```text
1 2 3
4 5 6
7 8 9
```

顺时针旋转 90 度后应该是：

```text
7 4 1
8 5 2
9 6 3
```

拆成两步就很好做：

**第一步：转置**（把行和列互换，也就是沿主对角线对称）：

```text
1 4 7
2 5 8
3 6 9
```

**第二步：每一行反转**：

```text
7 4 1
8 5 2
9 6 3
```

正好就是答案。

**转置怎么写？** 只用处理主对角线**右上方**的元素，把它和对称位置交换：`matrix[i][j]` 和 `matrix[j][i]`，其中 `j` 从 `i+1` 开始。这样每个交换只做一次，不会换回去。

**原地交换怎么写？** Python 可以直接写 `a, b = b, a`。


````python
# 手工走一遍 3x3
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
n = 3

# 第一步：转置
for i in range(n):
    for j in range(i + 1, n):
        matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
print("转置后：")
for line in matrix:
    print("   ", line)

# 第二步：每行反转
for i in range(n):
    left = 0
    right = n - 1
    while left < right:
        matrix[i][left], matrix[i][right] = matrix[i][right], matrix[i][left]
        left = left + 1
        right = right - 1
print("每行反转后：")
for line in matrix:
    print("   ", line)
````

## 题目 4 做题前先补：从右上角出发

矩阵的行从左到右递增，列从上到下递增。要在里面找目标值。

**从右上角出发**是最妙的起点，因为：

- 右上角这一列中，它是**最小的**（列往下递增）；
- 右上角这一行中，它是**最大的**（行往右递增）。

所以每次比较只有三种结果，而且每次都能砍掉一整行或一整列：

- 等于目标：找到了；
- **大于目标**：这一列下面只会更大，所以整列排除，往左走一格；
- **小于目标**：这一行左边只会更小，所以整行排除，往下走一格。

一直走到越界，说明不存在。

**为什么不能从左上角出发？** 左上角既是行最小又是列最小，不管比目标大还是小，都没法确定该往哪边走。


````python
# 手工走一遍查找 5
matrix = [
    [1, 4, 7, 11, 15],
    [2, 5, 8, 12, 19],
    [3, 6, 9, 16, 22],
    [10, 13, 14, 17, 24],
]
target = 5
row = 0
col = len(matrix[0]) - 1

while row < len(matrix) and col >= 0:
    value = matrix[row][col]
    print("看位置", row, col, "，值是", value)
    if value == target:
        print("    找到了")
        break
    elif value > target:
        col = col - 1
        print("    比目标大，往左走")
    else:
        row = row + 1
        print("    比目标小，往下走")
````
