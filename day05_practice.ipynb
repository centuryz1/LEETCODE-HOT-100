# 题目 2：反转链表

LeetCode 206. Reverse Linked List

## 题目描述（改写版）

给你一个单链表的头节点 `head`，把所有节点的连接方向反过来，并返回反转后链表的新头节点。

### 输入、输出与限制

- `head` 可能是 `None`，也可能只有一个节点。
- 需要返回新的头节点。
- 节点本身可以复用，只改变 `next` 指向，不需要创建一整条新链表。

### 示例

输入 `1 -> 2 -> 3 -> 4 -> 5`，输出 `5 -> 4 -> 3 -> 2 -> 1`。

```python
# 我的答案：反转链表

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next

    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head

    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next

    return values


def reverse_list(head):
    previous = None
    current = head

    while current is not None:
        next_node = current.next
        current.next = previous
        previous = current
        current = next_node

    return previous


def main():
    result = reverse_list(build_list([1, 2, 3, 4, 6]))
    print("反转后的结果为", list_values(result))


if __name__ == "__main__":
    main()
```

# 题目 3：环形链表

LeetCode 141. Linked List Cycle

## 题目描述（改写版）

给你一个单链表头节点 `head`，判断链表中是否存在环。如果沿着 `next` 不断前进，最终会再次到达之前访问过的节点，就说明有环。

### 输入、输出与限制

- 输入可能是空链表。
- 返回布尔值 `True` 或 `False`。
- 节点值可能重复，因此不能根据 `val` 判断是否回到了旧节点。
- 不需要找环的入口，本题只判断有没有环。

### 示例

`3 -> 2 -> 0 -> -4`，若 `-4` 的 `next` 指回节点 2，则返回 `True`。

```python
# 题目：环形链表
# 解法：Floyd 快慢指针判环（Floyd's Cycle Detection）
# 输入：可能为空的单链表头节点 head。
# 目标：判断沿 next 前进时是否会回到旧节点。
# 输出：存在环返回 True，否则返回 False。
# 注意：比较节点身份；访问 fast.next.next 前先检查 fast 和 fast.next。

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build_cycle_list(values, pos=-1):
    if values is None:
        return None
    
    dummy = ListNode()
    tail = dummy
    cycle_entry = None

    for index, value in enumerate(values):
        tail.next = ListNode(value)
        tail = tail.next

        if index == pos:
            cycle_entry = tail

    if pos >=0:
        tail.next = cycle_entry

    return dummy.next

def has_cycle(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False

head = build_cycle_list([1,3,4,5,7,8],3)
print(has_cycle(head))
    
  










```

# 题目 4：回文链表

LeetCode 234. Palindrome Linked List

这是今天的加练题。

## 题目描述（改写版）

给你一个单链表头节点 `head`。如果从前往后读取的节点值与从后往前读取完全相同，返回 `True`；否则返回 `False`。

### 输入、输出与限制

- 链表至少有一个节点。
- 节点值可以重复。
- 目标进阶要求是时间 `O(n)`、额外空间 `O(1)`。

### 示例

`1 -> 2 -> 2 -> 1` 返回 `True`；`1 -> 2` 返回 `False`。

```python
# 题目：回文链表
# 解法：快慢指针 + 后半链表反转（Fast-Slow Pointers + Half Reversal）
# 输入：单链表头节点 head。
# 目标：判断节点值正读和反读是否相同。
# 输出：是回文返回 True，否则返回 False。
# 注意：先找中点，再反转后半段；比较的是节点值。

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build_list(values):
    if values is None:
        return None
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next

def is_palindrome(head):
    if head is None:
        return True

    slow = head
    fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    reverse_slow = None
    current = slow
    while current is not None:
        next_node = current.next
        current.next = reverse_slow
        reverse_slow = current
        current = next_node

    while reverse_slow is not None:
        if reverse_slow.val != head.val:
            return False
        reverse_slow = reverse_slow.next
        head = head.next

    return True

def main():
    values = list(map(int, input().split()))
    head = build_list(values)
    print(is_palindrome(head))

if __name__ == '__main__':
    main()


```
