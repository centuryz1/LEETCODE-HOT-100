# 题目 1：相交链表

LeetCode 160. Intersection of Two Linked Lists

## 题目描述（改写版）

给你两个单链表的头节点 `head_a` 和 `head_b`。如果两个链表从某个节点开始共用同一段节点，请返回第一个公共节点；如果完全不相交，返回 `None`。

### 输入、输出与限制

- 输入是两个链表头节点，任意一个都可能是 `None`。
- 返回的是公共节点对象，不是节点的值。
- 相交后，后面的节点必须全部共用；题目保证链表中没有环。
- 不能为了得到答案而破坏原链表。

### 示例

链表 A 为 `4 -> 1 -> 8 -> 4 -> 5`，链表 B 为 `5 -> 6 -> 1 -> 8 -> 4 -> 5`。如果两个值为 8 的位置其实是同一个节点对象，就返回这个节点。

```python
# 题目：相交链表
# 解法：双指针路径拼接（Two Pointers with Path Switching）
# 输入：两个可能为空的链表头节点 head_a、head_b。
# 目标：找到两个链表开始共用的第一个节点。
# 输出：返回公共节点对象；不相交时返回 None。
# 注意：比较节点身份要用 is，不能只比较 val；不要修改链表。
class listnode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = listnode()
    tail = dummy

    for value in values:
        tail.next = listnode(value)
        tail = tail.next

    return dummy.next


def list_value(head):
    values = []
    current = head

    while current is not None:
        values.append(current.val)
        current = current.next

    return values


def attach(prefix_value,shared_head):
    head = build_list(prefix_value)
    if head is None:
        return shared_head

    tail = head
    while tail.next is not None:
        tail = tail.next
    tail.next = shared_head

    return head


def build_intersecting_lists(a,b,c):
    shared_head = build_list(c)
    head_a = attach(a,shared_head)
    head_b = attach(b,shared_head)

    return head_a,head_b,shared_head
    


def get_answer(head_a,head_b):
    p = head_a
    q = head_b

    while p is not q:
        if p is None:
            p = head_b
        p = p.next

        if q is None:
            q = head_a
        q = q.next

    return p



# 从这里开始创建本题输入链表；然后调用你写的函数。
head_a, head_b, shared_head = build_intersecting_lists([4, 1], [5, 6, 1], [8, 4, 5])
print(get_answer(head_a, head_b).val)
```

```python
## ACM格式
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next

    return dummy.next


def attach(prefix_values, shared_head):
    head = build_list(prefix_values)

    if head is None:
        return shared_head

    tail = head

    while tail.next is not None:
        tail = tail.next

    tail.next = shared_head

    return head


def build_intersecting_lists(a, b, shared):
    shared_head = build_list(shared)

    head_a = attach(a, shared_head)
    head_b = attach(b, shared_head)

    return head_a, head_b


def get_intersection_node(head_a, head_b):
    p = head_a
    q = head_b

    while p is not q:
        if p is None:
            p = head_b
        else:
            p = p.next

        if q is None:
            q = head_a
        else:
            q = q.next

    return p


if __name__ == "__main__":
    a = list(map(int, input().split()))
    b = list(map(int, input().split()))
    shared = list(map(int, input().split()))

    head_a, head_b = build_intersecting_lists(a, b, shared)

    answer = get_intersection_node(head_a, head_b)

    if answer is None:
        print("None")
    else:
        print(answer.val)
```
