## 题目 1 做题前先补：树节点、空树和栈

二叉树节点有两个孩子引用：`left` 和 `right`。不存在的孩子写成 `None`。

中序遍历顺序是：**左子树 -> 当前节点 -> 右子树**。

迭代写法使用列表模拟栈：

- `stack.append(node)`：把节点压栈，稍后再处理。
- `stack.pop()`：取出最后压入的节点。
- `while current is not None or stack`：当前还能向左走，或者栈里还有待处理节点时继续。

````python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(values):
    if not values or values[0] is None:
        return None
    root = TreeNode(values[0])
    queue = deque([root])
    index = 1
    while queue and index < len(values):
        node = queue.popleft()
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)
        index += 1
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)
        index += 1
    return root

root = build_tree([1, None, 2, 3])
print(root.val)             # 1
print(root.right.val)       # 2
print(root.right.left.val)  # 3
````

# 题目 1：二叉树的中序遍历

LeetCode 94. Binary Tree Inorder Traversal

## 题目描述（改写版）

给你一棵二叉树的根节点 `root`，按照左子树、当前节点、右子树的顺序访问所有节点，并返回节点值列表。

### 输入、输出与限制

- 输入可能是空树，此时返回空列表。
- 每个节点恰好访问一次。
- 返回的是值列表，不需要修改树。

### 示例

层序结构 `[1, None, 2, 3]` 表示 1 的右孩子是 2，2 的左孩子是 3；中序结果为 `[1, 3, 2]`。

## 解法名称

主学：**迭代中序遍历（Iterative Inorder Traversal）**。递归版本叫 **Recursive Inorder Traversal**。

## 递归思路

递归访问左子树、记录当前值、递归访问右子树，代码短，但递归栈最坏为 `O(n)`。

## 迭代思路

不断把当前节点及其左链压栈；走到 `None` 后弹出一个节点记录，再转向它的右子树。

时间复杂度 `O(n)`，空间复杂度 `O(h)`，`h` 是树高。

````python
root = build_tree([1, None, 2, 3])
stack, answer = [], []
current = root
while current is not None or stack:
    while current is not None:
        print("压栈：", current.val)
        stack.append(current)
        current = current.left
    current = stack.pop()
    answer.append(current.val)
    print("访问：", current.val, "当前答案：", answer)
    current = current.right
````

## 题目 2 做题前先补：递归函数的三件事

写树递归时依次回答：

1. **函数含义**：`max_depth(node)` 返回以 `node` 为根的树的最大深度。
2. **终止条件**：空树深度为 0。
3. **递推关系**：当前深度等于左右子树较大深度加 1。

```python
return max(left_depth, right_depth) + 1
```

这里的 `+1` 代表当前节点这一层。

````python
def tiny_depth(node):
    if node is None:
        return 0
    left_depth = tiny_depth(node.left)
    right_depth = tiny_depth(node.right)
    return max(left_depth, right_depth) + 1

print(tiny_depth(build_tree([3,9,20,None,None,15,7])))  # 3
````

# 题目 2：二叉树的最大深度

LeetCode 104. Maximum Depth of Binary Tree

## 题目描述（改写版）

给你一棵二叉树，返回从根节点到最远叶子节点所经过的节点数量。空树深度为 0。

### 输入、输出与限制

- `root` 可能是 `None`。
- 深度按节点数计算，只有根节点时深度是 1。
- 不需要修改树。

### 示例

树 `[3,9,20,None,None,15,7]` 的最深路径可以是 `3 -> 20 -> 15`，因此返回 3。

## 解法名称

专业名称：**递归深度优先搜索（Recursive Depth-First Search, DFS）**。

## 层序遍历思路

也可以一层层 BFS，每处理完一层把深度加 1，需要队列。

## 递归思路

分别求左右子树最大深度，取较大值再加当前节点这一层。递归代码直接对应树的定义。

时间复杂度 `O(n)`，递归栈空间复杂度 `O(h)`。

````python
def depth_with_trace(node, indent=0):
    if node is None:
        print("  " * indent + "空节点 -> 0")
        return 0
    print("  " * indent + f"进入节点 {node.val}")
    left = depth_with_trace(node.left, indent + 1)
    right = depth_with_trace(node.right, indent + 1)
    result = max(left, right) + 1
    print("  " * indent + f"节点 {node.val} -> {result}")
    return result

depth_with_trace(build_tree([3,9,20,None,None,15,7]))
````

### 输入、输出与限制

- 输入可以是空树。
- 直接复用并修改原树即可。
- 每个节点都要执行相同的左右交换。

### 示例

原树根节点 4 的左右孩子为 2 和 7；翻转后左右孩子变为 7 和 2，它们各自的孩子也同样交换。

## 解法名称

专业名称：**递归交换子树（Recursive Subtree Swapping）**。

## 思路

空节点无需处理。对当前节点交换左右孩子，然后递归翻转交换后的左右子树。也可以先递归再交换，结果相同。

每个节点只处理一次，时间复杂度 `O(n)`；递归栈空间复杂度 `O(h)`。

````python
def invert_with_trace(node):
    if node is None:
        return None
    print("交换节点：", node.val)
    node.left, node.right = node.right, node.left
    invert_with_trace(node.left)
    invert_with_trace(node.right)
    return node

demo = invert_with_trace(build_tree([4,2,7,1,3,6,9]))
print("翻转后中序：", inorder_traversal_answer(demo))
````

## 题目描述（改写版）

给你一棵二叉树，判断它是否关于根节点所在的竖直中线左右镜像对称。

### 输入、输出与限制

- 返回布尔值。
- 空树可以看作对称。
- 不只比较节点值，缺失节点的位置也必须镜像对应。

### 示例

`[1,2,2,3,4,4,3]` 左右结构与值互为镜像，返回 `True`；`[1,2,2,None,3,None,3]` 返回 `False`。

## 解法名称

专业名称：**镜像递归（Mirror Recursion）**。

## 思路

定义 `mirror(left, right)` 判断两棵子树是否互为镜像。先处理空节点和值不同的情况，再交叉递归比较外侧和内侧。

时间复杂度 `O(n)`，递归栈空间复杂度 `O(h)`。也可以使用队列迭代实现。

````python
def mirror_trace(left, right, depth=0):
    print("  " * depth, None if left is None else left.val, None if right is None else right.val)
    if left is None or right is None:
        return left is right
    if left.val != right.val:
        return False
    return mirror_trace(left.left, right.right, depth + 1) and mirror_trace(left.right, right.left, depth + 1)

sample = build_tree([1,2,2,3,4,4,3])
print(mirror_trace(sample.left, sample.right))
````

## 参考答案：对称二叉树

````python
def is_symmetric_answer(root):
    def mirror(left, right):
        if left is None and right is None:
            return True
        if left is None or right is None:
            return False
        if left.val != right.val:
            return False
        return mirror(left.left, right.right) and mirror(left.right, right.left)

    return True if root is None else mirror(root.left, root.right)
````
