## 今天最重要的事：DP 四步法

以后每道 DP 题都按这四步想，不要跳步：

**第一步：定义 `dp[i]` 是什么。**
这是最关键的一步。比如爬楼梯里 `dp[i]` 表示“爬到第 i 阶有几种方法”。定义写不清楚，后面一定乱。

**第二步：写出状态转移方程。**
也就是“`dp[i]` 由哪些更小的状态推出来”。比如爬楼梯是 `dp[i] = dp[i-1] + dp[i-2]`。

**第三步：确定初始值。**
最小的问题答案是什么？比如 `dp[0] = 1`、`dp[1] = 1`。初始值错了，整个表都错。

**第四步：确定遍历顺序。**
因为要用到更小的状态，所以一般是从小到大、从左到右。

再记一个通用套路：**DP 的每个状态，都是在做“选择”并取最优。** 打家劫舍是“偷或不偷”，零钱兑换是“用哪枚硬币”，最长递增子序列是“接在谁后面”。


````python
# 用一个最简单的例子体会“存中间结果”的好处
# 求 1 到 5 的累加和：如果每次都重新加一遍，就要重复算很多次

n = 5
dp = [0] * (n + 1)
for i in range(1, n + 1):
    dp[i] = dp[i - 1] + i
    print("dp[" + str(i) + "] =", dp[i], "（用了已经算好的 dp[" + str(i - 1) + "]）")
print("最终结果：", dp[5])
````

## 题目 1 做题前先补：爬到第 i 阶，只可能来自哪一步

爬楼梯：每次可以爬 1 阶或 2 阶，问爬到第 n 阶有多少种方法。

关键观察：**要站上第 i 阶，最后一步只有两种可能：**

- 从第 i-1 阶爬 1 阶上来；
- 从第 i-2 阶爬 2 阶上来。

所以：`dp[i] = dp[i-1] + dp[i-2]`。

初始值：`dp[0] = 1`（不爬也是一种“方法”，方便递推），`dp[1] = 1`。

再检查 `n = 3`：`dp[3] = dp[2] + dp[1] = 2 + 1 = 3` ✓（1+1+1、1+2、2+1 三种）。

这个过程和斐波那契数列是同一个递推式，所以也能用两个变量滚动，省掉整个数组。


````python
n = 6
dp = [0] * (n + 1)
dp[0] = 1
dp[1] = 1
print("dp[0] =", dp[0], "，dp[1] =", dp[1])
for i in range(2, n + 1):
    dp[i] = dp[i - 1] + dp[i - 2]
    print("dp[" + str(i) + "] = dp[" + str(i - 1) + "] + dp[" + str(i - 2) + "] =", dp[i])
print("组合后的 dp 数组：", dp)
````

面试表达：

这题的状态定义是：`dp[i]` 表示爬到第 i 阶的方法数。因为每次只能爬 1 阶或 2 阶，所以到达第 i 阶的最后一步要么来自第 i-1 阶，要么来自第 i-2 阶，所以 `dp[i] = dp[i-1] + dp[i-2]`。初始值 `dp[0]` 和 `dp[1]` 都是 1。从小到大填表就行，时间 O(n)。

因为每个状态只依赖前两个状态，我还可以把数组优化成两个变量滚动，把空间降到 O(1)。


````python


def climb_stairs_answer(n):
    if n <= 2:
        return n
    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]


def climb_stairs_second(n):
    """两个变量滚动版，空间 O(1)。"""
    if n <= 2:
        return n
    first = 1
    second = 2
    for i in range(3, n + 1):
        current = first + second
        first = second
        second = current
    return second


print(climb_stairs_answer(3), climb_stairs_second(3))     # 3 3
print(climb_stairs_answer(5), climb_stairs_second(5))     # 8 8
print(climb_stairs_answer(44))                            # 1134903170
print(climb_stairs_second(44))                            # 1134903170
````

## 题目 2 做题前先补：杨辉三角的规律

杨辉三角是这样的：

```text
1
1 1
1 2 1
1 3 3 1
1 4 6 4 1
```

两条规律：

1. 每行的第一个和最后一个是 1；
2. 中间的元素等于**上一行相邻两个元素之和**。

用公式写：第 `i` 行第 `j` 列（`j` 从 1 开始数，不是边界）= 上一行的第 `j-1` 个加第 `j` 个。

这正是二维 DP：每一行都由上一行推出来。

注意下标：`row[0]` 和 `row[i]` 永远是 1，只有 `j` 在 1 到 `i-1` 之间才需要计算。


````python
# 手工造前 4 行
num_rows = 4
result = []
for i in range(num_rows):
    row = [1] * (i + 1)
    for j in range(1, i):
        row[j] = result[i - 1][j - 1] + result[i - 1][j]
    result.append(row)
    print("第", i, "行：", row)
````

## 题目 3 做题前先补：每一步都在做“选还是不选”

打家劫舍的规则：不能偷相邻的两家，求能偷到的最大金额。

走到第 i 家时，只有两个选择：

- **偷**：那么第 i-1 家不能偷，最大金额 = `dp[i-2] + nums[i]`；
- **不偷**：最大金额就是 `dp[i-1]`。

在这两个选择里取大的，就是 `dp[i]`。

`dp[i]` 的定义：**考虑前 i 家（下标 0 到 i），能偷到的最大金额。**

所以转移方程：

```text
dp[i] = max(dp[i-1], dp[i-2] + nums[i])
```

初始值：

- `dp[0] = nums[0]`（只有一家，当然偷它）；
- `dp[1] = max(nums[0], nums[1])`（两家只能偷一家，偷大的那家）。

这就是 DP 的核心套路：**每个状态都是一次选择，答案就是所有选择里最优的那个。**


````python
nums = [2, 7, 9, 3, 1]
dp = [0] * len(nums)
dp[0] = nums[0]
if nums[1] > nums[0]:
    dp[1] = nums[1]
else:
    dp[1] = nums[0]
print("dp[0] =", dp[0], "，dp[1] =", dp[1])

for i in range(2, len(nums)):
    take = dp[i - 2] + nums[i]
    skip = dp[i - 1]
    if take > skip:
        dp[i] = take
    else:
        dp[i] = skip
    print("第", i, "家：偷了得", take, "，不偷得", skip, "，取", dp[i])

print("最大金额：", dp[-1])
````

## 题目 4 做题前先补：从多个选项里挑最小的

这题要问：把 n 拆成若干个**完全平方数**之和，最少需要几个？

`dp[i]` 的定义：**凑出 i 需要的最少平方数个数。**

转移：最后一个平方数可以是 1、4、9、16……只要它不超过 i：

```text
dp[i] = min(dp[i - 1], dp[i - 4], dp[i - 9], ...) + 1
```

也就是：**枚举最后一个平方数是谁，取“剩下部分最少需要几个”再加一。**

初始化：`dp[0] = 0`（凑 0 不需要任何数）。

循环写法：

```python
j = 1
while j * j <= i:
    ... 用 j * j 去更新 dp[i] ...
    j = j + 1
```

用 `while` 是因为不知道 `i` 的平方根是多少，直接用 `j * j <= i` 当条件最直观（也可以写 `j * j <= i`，等价于 `j <= int(i ** 0.5)`）。


````python
n = 12
dp = [0] * (n + 1)
for i in range(1, n + 1):
    best = n + 1
    j = 1
    while j * j <= i:
        candidate = dp[i - j * j] + 1
        if candidate < best:
            best = candidate
        j = j + 1
    dp[i] = best
    print("dp[" + str(i) + "] =", dp[i])
print("凑出", n, "最少需要", dp[n], "个完全平方数")
````

## 易漏细节

- `dp[0] = 0` 是初始值，凑 0 不需要任何数。
- 每层要枚举所有不超过 i 的平方数，取最小值加一。
- 初始值设成 `n + 1`（一个肯定取不到的大数）比较方便。


## 解法名称

**完全背包型动态规划（Unbounded Knapsack DP）**。

## 暴力思路

递归枚举用哪些平方数，重复子问题极多。

## 优化思路

`dp[i]` = 凑出 i 的最少个数。

- 枚举最后一个用的平方数 `j * j`（要求 `j * j <= i`）；
- 候选值 = `dp[i - j * j] + 1`；
- 取所有候选里最小的。

时间 O(n × √n)，空间 O(n)。

**另一种思路（BFS）**：把 0 当起点，每次加一个平方数，看几步能到 n。层数就是答案，效果和 DP 一样，面试时也可以提。

面试表达：

我用一维动态规划，`dp[i]` 表示凑出 i 最少需要几个完全平方数。对每个 i，我枚举最后一个使用的平方数 `j * j`，那么剩下的 `i - j * j` 最少需要 `dp[i - j * j]` 个，再加当前这一个，就是候选答案；把所有候选取最小值。`dp[0]` 是 0，因为凑 0 不需要任何数。时间 O(n 乘根号 n)，空间 O(n)。这题也可以用 BFS 做：从 0 出发，每次加一个平方数，看最少几步到达 n。


````python


def num_squares_answer(n):
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        best = n + 1
        j = 1
        while j * j <= i:
            candidate = dp[i - j * j] + 1
            if candidate < best:
                best = candidate
            j = j + 1
        dp[i] = best
    return dp[n]


print(num_squares_answer(12))     # 3
print(num_squares_answer(13))     # 2
print(num_squares_answer(1))      # 1
print(num_squares_answer(99))     # 3
print(num_squares_answer(100))    # 1
````

# 今日小结

今天的核心是 **DP 四步法**：

1. 定义 `dp[i]` 是什么；
2. 写状态转移方程；
3. 定初始值；
4. 定遍历顺序。

四道题的状态定义：

1. **爬楼梯**：`dp[i]` = 爬到第 i 阶的方法数，来自前两阶。
2. **杨辉三角**：每行由上一行的相邻两数推出。
3. **打家劫舍**：`dp[i]` = 前 i 家的最大金额，选择是“偷或不偷”。
4. **完全平方数**：`dp[i]` = 凑出 i 的最少个数，枚举最后一个平方数。

一个通用经验：**如果状态只依赖前一个或前两个，就可以用变量滚动省掉数组。**
