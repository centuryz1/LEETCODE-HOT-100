# 0. 先建立总地图

## 0.1 数据结构与算法方法不是一回事

### 数据结构：数据放在哪里、如何连接

- 数组、字符串
- 哈希表、集合
- 栈、队列
- 链表
- 树、堆、前缀树
- 图、并查集

### 算法方法：你怎样使用或处理数据

- 双指针、滑动窗口
- 二分查找
- 深度优先搜索 DFS、广度优先搜索 BFS
- 回溯、贪心、动态规划
- 排序、前缀和

例如“无重复字符的最长子串”使用：

```text
字符串（数据结构） + 哈希表（数据结构） + 滑动窗口（算法方法）
```

面试时这样拆解，思路会比只背“这是一道滑动窗口题”更完整。

## 0.2 线性结构与非线性结构

### 线性结构

元素大致排成一条线：

```text
数组：  [A][B][C][D]
链表：  A -> B -> C -> D
栈：    只能重点操作一端
队列：  一端进入，另一端出去
```

### 非线性结构

一个元素可能连接多个元素：

```text
树：        A
          /            B     C

图：      A --- B
          |   / |
          C --- D
```

树是一种没有环、通常有层级关系的特殊图；普通图可以有环，也可以有多条路径。

## 0.3 抽象数据类型与 Python 容器

同一种 Python 容器可以承担不同角色：

| 抽象结构 | Python 常用实现 |
|---|---|
| 动态数组 | `list` |
| 栈 | `list` |
| 队列 / 双端队列 | `collections.deque` |
| 哈希表 | `dict` |
| 哈希集合 | `set` |
| 最小堆 | `heapq` 操作的 `list` |
| 链表 | 自定义 `ListNode` |
| 二叉树 | 自定义 `TreeNode` |
| 图 | 邻接表，通常是 `list[list]` 或 `dict[list]` |

所以看到 `list` 不代表题目一定是“数组题”。关键是它承担什么职责。

````python
# 同一个 list 可以承担两个不同角色
array = [10, 20, 30]  # 动态数组：经常按下标访问

stack = []            # 栈：只在末尾压入和弹出
stack.append('A')
stack.append('B')

print('数组下标访问：', array[1])
print('栈顶弹出：', stack.pop())
````

## 0.4 选择数据结构，其实是在选择“快在哪里”

不存在所有操作都最快的数据结构。

- 数组按下标访问快，但中间插入慢。
- 链表已知节点时修改连接快，但找第 `k` 个慢。
- 哈希表查找快，但不表达大小顺序。
- 堆能快速拿最小值，但不能快速查任意元素。
- 栈只能方便地处理最近加入的元素。

刷题时先问：

> 这道题最频繁、最昂贵的操作是什么？

数据结构的选择，就是让这个核心操作变快。

## 0.5 Hot 100 数据结构识别速览

| 题目出现的信号 | 优先想到 |
|---|---|
| 按下标、连续区间、原地修改 | 数组 / 字符串 |
| 快速判断是否出现、次数、配对 | 哈希表 / 集合 |
| 最近一个尚未完成、括号、右侧第一个更大 | 栈 / 单调栈 |
| 一层层扩散、最少步数、先进先出 | 队列 / BFS |
| 节点通过 `next` 相连、倒数位置、环 | 链表 |
| 父子层级、子树、路径、祖先 | 树 / DFS / BFS |
| 反复取最小或最大、前 K 个 | 堆 |
| 岛屿、课程依赖、任意连接网络 | 图 |
| 字符串前缀、单词搜索 | Trie |
| 动态合并集合、判断是否连通 | 并查集 |

这只是第一候选。真正的题目经常还要组合双指针、滑动窗口、递归等方法。

# 1. 数组与字符串

## 1.1 核心模型

数组可以想成一排连续编号的格子：

```text
下标：  0    1    2    3
元素： [10] [20] [30] [40]
```

Python 的 `list` 是动态数组：长度可以变化，但依然擅长按下标访问。

字符串也可以按下标访问和切片，但字符串是不可变对象，不能直接修改某个字符。

### 数组最擅长

- 按下标读取和修改；
- 表达连续区间；
- 从左到右或从右到左扫描；
- 原地交换、排序、二分查找。

## 1.2 常用操作与复杂度

| 操作 | Python | 时间复杂度 |
|---|---|---:|
| 按下标访问 | `nums[i]` | `O(1)` |
| 修改下标 | `nums[i] = x` | `O(1)` |
| 尾部追加 | `nums.append(x)` | 均摊 `O(1)` |
| 尾部删除 | `nums.pop()` | `O(1)` |
| 头部/中间插入删除 | `insert`、`pop(i)` | `O(n)` |
| 查找某个值 | `x in nums` | `O(n)` |
| 长度为 `k` 的切片 | `nums[a:b]` | `O(k)` |
| 排序 | `nums.sort()` | `O(n log n)` |

“均摊 `O(1)`”表示偶尔扩容较慢，但大量追加操作平均下来仍是常数时间。

````python
nums = [10, 20, 30]
nums.append(40)
print('追加后：', nums)

print('下标 2：', nums[2])
print('切片 [1:3]：', nums[1:3])

last = nums.pop()
print('弹出末尾：', last, '剩余：', nums)
````

## 1.3 Hot 100 常见用法

### 双指针

适合两端夹逼、原地移动、合并有序数据：

- 移动零
- 盛最多水的容器
- 三数之和
- 接雨水

### 滑动窗口

适合连续子串或连续子数组：

- 无重复字符的最长子串
- 找到字符串中所有字母异位词

### 前缀信息

提前保存“从开头到当前位置”的结果：

- 和为 K 的子数组
- 除自身以外数组的乘积

### 排序与区间

- 合并区间

### 矩阵

二维数组本质上仍然是数组：

- 螺旋矩阵
- 矩阵置零
- 搜索二维矩阵

## 1.4 看到什么信号想到数组

- 输入本来就是列表或字符串；
- 题目强调连续子数组、连续子串；
- 需要返回下标；
- 需要原地修改；
- 数据已经有序，可以二分或双指针；
- 需要维护左右边界。

### 常见误区

- 把 `list.pop(0)` 当作 `O(1)` 队列操作，它实际上要移动后面的元素；
- 以为切片只是“查看”，其实切片会创建新对象；
- 混淆元素数量 `n` 和最后下标 `n-1`；
- 字符串不可直接执行 `s[i] = char`。

## 1.5 小练习：数组还是其他结构

判断下面需求首先想到什么：

1. 在有序数组中查找一个目标，要求 `O(log n)`。
2. 把数组中的 0 移到末尾，要求原地修改。
3. 频繁从最前面删除元素。
4. 判断一个值是否出现过，并执行很多次查询。

答案：

1. 数组 + 二分查找。
2. 数组 + 双指针。
3. 不应使用普通列表头删，优先队列 `deque`。
4. 哈希集合 `set`。

# 2. 哈希表：字典与集合

## 2.1 核心模型

哈希表通过一个键快速找到对应位置。

Python 中：

- `dict` 保存 `key -> value`。
- `set` 只保存不重复的 key。

```python
counts['a'] = 3       # 字符 a 出现 3 次
seen.add(42)          # 记录数字 42 出现过
```

哈希表不要求逐个扫描全部数据，平均情况下查询、插入、删除都是 `O(1)`。

## 2.2 `dict` 与 `set` 的职责

### 使用 `set`：只关心“有没有”

- 去重；
- 快速成员判断；
- 标记访问过的节点或状态。

### 使用 `dict`：还需要记录额外信息

- 值 -> 下标；
- 元素 -> 出现次数；
- 前缀和 -> 出现次数；
- 节点 -> 克隆节点；
- key -> 缓存节点。

选择问题：

> 只需要知道它存在吗？用 set。还要记住与它关联的信息吗？用 dict。

````python
nums = [2, 7, 2, 9]

seen = set()
counts = {}

for value in nums:
    seen.add(value)
    counts[value] = counts.get(value, 0) + 1

print('出现过哪些值：', seen)
print('每个值出现几次：', counts)
````

## 2.3 常用操作与复杂度

| 操作 | `dict` / `set` | 平均时间 |
|---|---|---:|
| 查询 | `key in table` | `O(1)` |
| 读取 | `table[key]` | `O(1)` |
| 写入 | `table[key] = value` | `O(1)` |
| 删除 | `del table[key]` / `discard` | `O(1)` |
| 遍历全部 | `for key in table` | `O(n)` |

面试中通常说“平均 `O(1)`”；极端哈希冲突下可能退化，但普通算法题按平均复杂度分析。

键必须可哈希：整数、字符串、元组通常可以；列表不能直接作为字典键。

## 2.4 Hot 100 中字典的四种角色

### 角色 1：查补数

两数之和中，遍历 `x` 时查询 `target-x` 是否出现。

### 角色 2：计数

字母异位词分组、找到所有字母异位词会统计字符频次。

### 角色 3：保存下标或位置

两数之和保存 `数值 -> 下标`。

### 角色 4：保存历史状态

和为 K 的子数组保存 `前缀和 -> 出现次数`。

这四种角色比背 API 更重要。看到题目时，先问字典里的 key 和 value 分别应该是什么。

````python
# 两数之和：字典保存 数值 -> 下标
def two_sum_demo(nums, target):
    index_by_value = {}
    for index, value in enumerate(nums):
        needed = target - value
        if needed in index_by_value:
            return [index_by_value[needed], index]
        index_by_value[value] = index
    return []


print(two_sum_demo([2, 7, 11, 15], 9))
````

## 2.5 Hot 100 中集合的常见角色

- **最长连续序列**：把所有数字放进集合，`O(1)` 判断相邻数字是否存在；
- **环或访问状态**：记录节点是否访问过；
- **矩阵置零**：记录哪些行、列需要清零；
- **图搜索**：`visited` 防止重复访问；
- **去重**：结果或候选项不能重复。

### 常见误区

- 需要次数却只用 `set`，信息不够；
- 一边遍历字典一边改变其大小；
- 使用 `table[key]` 读取不存在的键导致 `KeyError`；
- 字典查找虽然快，但创建字典仍需要遍历输入，整体通常是 `O(n)`。

# 3. 栈

## 3.1 核心模型：后进先出 LIFO

```text
压入 A： [A]
压入 B： [A, B]
压入 C： [A, B, C]
弹出：             C
```

栈只能重点操作最近加入的一端。

Python 通常用列表：

```python
stack.append(value)  # 入栈
stack.pop()          # 出栈
stack[-1]            # 查看栈顶
```

这三个操作在列表末尾平均都是 `O(1)`。

````python
stack = []
stack.append('A')
stack.append('B')
stack.append('C')

print('栈顶：', stack[-1])
print('弹出：', stack.pop())
print('剩余：', stack)
````

## 3.2 什么时候想到栈

### 最近一个尚未完成的任务

- 左括号等待右括号；
- 字符串编码等待闭括号；
- 函数调用等待返回。

### 撤销与回退

- 编辑器撤销；
- 浏览路径；
- DFS 的递归调用栈。

### 右侧或左侧第一个更大/更小

使用单调栈：

- 每日温度；
- 柱状图中最大的矩形；
- 接雨水的单调栈解法。

## 3.3 普通栈与单调栈

### 普通栈

只要求后进先出，例如有效的括号。

### 单调栈

仍然是列表，但人为维持从栈底到栈顶递增或递减。

每日温度中，栈保存还没找到更高温度的下标。当当前温度更高时：

```python
while stack and temperatures[stack[-1]] < current_temp:
    old_index = stack.pop()
```

这里使用 `while`，因为一个新温度可能连续解决多个旧日期。每个下标最多入栈、出栈一次，总时间 `O(n)`。

````python
temperatures = [73, 74, 75, 71, 72]
stack = []
answer = [0] * len(temperatures)

for current, temp in enumerate(temperatures):
    while stack and temperatures[stack[-1]] < temp:
        old = stack.pop()
        answer[old] = current - old
    stack.append(current)

print(answer)
````

## 3.4 栈的易错点

- 空栈执行 `pop()` 或访问 `stack[-1]` 会报错；
- 括号题遍历结束后还要检查栈是否为空；
- 单调栈经常保存下标，不只是保存值；
- “单调递增/递减”要说清楚是从栈底到栈顶；
- 递归本质上也使用调用栈，深度过大可能发生递归深度问题。

### 面试表达

不要只说“我用了栈”。可以说：

> 栈用于保存尚未匹配的左括号，栈顶代表最近一个等待匹配的括号，因此满足后进先出的嵌套顺序。

# 4. 队列与双端队列

## 4.1 核心模型：先进先出 FIFO

```text
入队方向 -> [A, B, C] -> 出队方向
最先进入 A，最先离开 A
```

Python 不建议使用 `list.pop(0)`，因为删除开头后要移动所有剩余元素，时间 `O(n)`。

应使用：

```python
from collections import deque

queue.append(value)   # 队尾入队
queue.popleft()       # 队首出队
```

两端操作都是 `O(1)`。

````python
from collections import deque

queue = deque()
queue.append('A')
queue.append('B')
queue.append('C')

print('出队：', queue.popleft())
print('剩余：', queue)
````

## 4.2 什么时候想到队列

### 按到达顺序处理

- 任务排队；
- 数据流；
- 模拟轮流处理。

### 一层层向外扩散：BFS

- 二叉树层序遍历；
- 网格中的最短步数；
- 腐烂的橘子；
- 从起点逐层访问图。

BFS 中队列保存“已经发现、但还没处理”的节点。

### 固定窗口或单调队列

- 滑动窗口最大值使用 `deque` 保存有希望成为最大值的下标。

## 4.3 普通队列与单调队列

普通队列只维持先进先出。

单调队列还会主动删除没有价值的元素。滑动窗口最大值中：

1. 队首下标如果已经离开窗口，执行 `popleft()`；
2. 队尾对应值不大于当前值，执行 `pop()`；
3. 当前下标执行 `append()`；
4. 队首就是当前窗口最大值下标。

它不是保存窗口中的全部元素，而是保存“仍可能成为最大值”的候选下标。

````python
from collections import deque

nums = [1, 3, -1, -3, 5]
k = 3
candidates = deque()
maximums = []

for right, value in enumerate(nums):
    while candidates and candidates[0] <= right - k:
        candidates.popleft()
    while candidates and nums[candidates[-1]] <= value:
        candidates.pop()
    candidates.append(right)
    if right >= k - 1:
        maximums.append(nums[candidates[0]])

print(maximums)  # [3, 3, 5]
````

## 4.4 队列的易错点

- 用 `list.pop(0)` 导致整体退化；
- BFS 忘记在合适时机标记访问，节点重复入队；
- 二叉树分层时没有先记录 `level_size`；
- 单调队列混淆队首过期和队尾淘汰；
- 队列里保存节点、值还是下标，要根据答案需求决定。

### 栈和队列怎么选

- 最近加入的优先处理：栈。
- 最早加入的优先处理：队列。
- 按深度一路走到底：DFS，常由递归栈实现。
- 按距离一层层扩展：BFS，常用队列。

# 5. 链表

## 5.1 核心模型：节点通过引用连接

```text
[1 | next] -> [2 | next] -> [3 | None]
```

每个节点至少包含：

- `val`：节点保存的值；
- `next`：下一个节点的引用。

链表没有像数组那样的连续下标。要找第 `k` 个节点，必须从头沿 `next` 走过去。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


head = ListNode(1, ListNode(2, ListNode(3)))
current = head
while current is not None:
    print(current.val)
    current = current.next
````

## 5.2 数组与链表对比

| 操作 | 数组 | 单链表 |
|---|---:|---:|
| 按位置访问 | `O(1)` | `O(n)` |
| 查找某个值 | `O(n)` | `O(n)` |
| 尾部追加 | 均摊 `O(1)` | 有尾指针时 `O(1)` |
| 已知节点后插入 | 中间可能 `O(n)` | `O(1)` |
| 已知前驱后删除 | 中间可能 `O(n)` | `O(1)` |

链表“插入删除快”有前提：**已经拿到正确节点或前驱节点**。如果还要从头寻找，整体仍可能是 `O(n)`。

## 5.3 Hot 100 链表四大模式

### 模式 1：虚拟头节点 dummy

统一处理删除头节点、合并结果头节点不确定等情况。

### 模式 2：反转链表

使用 `previous`、`current`、`next_node` 改变箭头方向。

### 模式 3：快慢指针

- 找中点；
- 判断环；
- 找环入口。

### 模式 4：固定间距双指针

让快指针先走 `n` 步，用于删除倒数第 N 个节点。

相关 Hot 100：相交链表、反转链表、回文链表、环形链表、合并两个有序链表、两数相加。

````python
def reverse_list(head):
    previous = None
    current = head

    while current is not None:
        next_node = current.next   # 先保存后路
        current.next = previous    # 改变箭头
        previous = current
        current = next_node

    return previous


reversed_head = reverse_list(head)
print(reversed_head.val)  # 3
````

## 5.4 链表的易错点

- 修改 `current.next` 前没有保存原来的后继，导致后半链表丢失；
- 把值相等当作节点相同，判断相交或环时应比较节点对象；
- 返回 `dummy` 而不是真正头节点 `dummy.next`；
- 指针移动顺序错误；
- 打印有环链表时无限循环；
- 不画箭头，只靠脑中想象复杂连接。

建议每次链表修改都画三样东西：当前节点、原后继、新指向。

# 6. 树与二叉搜索树

## 6.1 树的基本术语

```text
        1          根节点
       /       2   3        1 的孩子
     /     4   5          叶子节点：3、4、5
```

- 根节点 root：最上面的起点；
- 父节点 / 子节点：上下连接关系；
- 叶子节点：没有孩子；
- 子树：某节点及其所有后代；
- 深度：根到节点的距离或节点数，定义要看题目；
- 高度：节点到最深叶子的距离或层数。

二叉树每个节点最多有左右两个孩子。

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


root = TreeNode(1)
root.left = TreeNode(2, TreeNode(4), TreeNode(5))
root.right = TreeNode(3)

print(root.val, root.left.val, root.right.val)
````

## 6.2 三种 DFS 遍历

名称取决于“根节点什么时候处理”：

| 遍历 | 顺序 | 常见用途 |
|---|---|---|
| 前序 Preorder | 根、左、右 | 复制、序列化、先处理当前状态 |
| 中序 Inorder | 左、根、右 | BST 得到有序序列 |
| 后序 Postorder | 左、右、根 | 汇总子树信息、删除、树形 DP |

递归题先定义函数含义，再写：

1. 空节点返回什么；
2. 左右子树分别返回什么；
3. 当前节点如何合并结果。

````python
def preorder(node):
    if node is None:
        return []
    return [node.val] + preorder(node.left) + preorder(node.right)


def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.val] + inorder(node.right)


def postorder(node):
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.val]


print('前序：', preorder(root))
print('中序：', inorder(root))
print('后序：', postorder(root))
````

## 6.3 BFS 层序遍历

BFS 使用队列一层层处理：

```text
第 1 层：1
第 2 层：2, 3
第 3 层：4, 5
```

每轮开始记录队列长度，就能把同一层分组。

Hot 100 常见：

- 二叉树的层序遍历；
- 二叉树的右视图；
- 把树按层处理的题目。

````python
from collections import deque

def level_order(root):
    if root is None:
        return []

    answer = []
    queue = deque([root])
    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.val)
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        answer.append(level)
    return answer


print(level_order(root))
````

## 6.4 二叉搜索树 BST

BST 满足：

- 左子树所有值严格小于当前节点；
- 右子树所有值严格大于当前节点；
- 左右子树也都是 BST。

核心性质：**中序遍历得到递增序列**。

Hot 100 常见用法：

- 验证二叉搜索树：递归传递上下界；
- BST 第 K 小元素：中序遍历访问第 K 个；
- 搜索或插入：根据大小决定向左还是向右。

平衡情况下操作约 `O(log n)`，退化成单链后可能是 `O(n)`。

## 6.5 Hot 100 树题分类

| 问题类型 | 常见方法 |
|---|---|
| 最大深度、路径信息 | 递归 DFS |
| 翻转、对称 | 同时处理左右子树 |
| 层序、右视图 | BFS 队列 |
| 验证 BST、第 K 小 | BST 性质 + 中序 |
| 最近公共祖先 | 后序递归 |
| 从遍历序列构造树 | 递归 + 哈希表定位 |
| 树的直径 | 后序 DFS，返回高度并更新答案 |

树题最重要的能力是：清楚递归函数“对一棵子树返回什么”。

## 6.6 树的易错点

- 忘记空节点终止条件；
- 混淆节点深度和树高；
- 只比较 BST 的父子节点，漏掉祖先范围；
- 递归函数既修改全局答案又返回值，却没说明两者职责；
- BFS 把新加入的下一层节点也在本轮处理；
- 认为二叉树一定平衡，错误地把所有操作都说成 `O(log n)`。

复杂度通常：每个节点访问一次是 `O(n)`，递归栈为 `O(h)`，其中 `h` 是树高。

# 7. 堆与优先队列

## 7.1 核心模型

堆是一棵满足特殊大小关系的完全二叉树。

最小堆保证：

- 堆顶是当前最小元素；
- 不保证其余元素整体有序。

Python 的 `heapq` 使用普通列表存储最小堆：

```python
heapq.heappush(heap, value)
heapq.heappop(heap)
heap[0]  # 查看最小值
```

压入和弹出是 `O(log n)`，查看堆顶是 `O(1)`。

````python
import heapq

heap = []
for value in [5, 2, 8, 1, 3]:
    heapq.heappush(heap, value)

print('堆的内部列表：', heap)
print('当前最小值：', heap[0])
print('依次弹出：', [heapq.heappop(heap) for _ in range(len(heap))])
````

## 7.2 什么时候想到堆

- 数据不断到来，需要反复取最小或最大；
- 只关心前 K 大或前 K 小，不需要完整排序；
- 合并多条有序序列；
- 多个候选任务按优先级处理；
- 数据流中位数使用两个堆。

Hot 100 常见：

- 数组中的第 K 个最大元素；
- 前 K 个高频元素；
- 合并 K 个升序链表；
- 数据流的中位数。

如果只需要最后找一次最小值，普通扫描可能更简单；堆的优势在于**动态、反复取优先项**。

## 7.3 用最小堆维护前 K 大

保持一个大小最多为 `k` 的最小堆：

1. 新元素入堆；
2. 堆大小超过 `k`，弹出最小值；
3. 最终堆中留下前 `k` 大元素；
4. 堆顶是这 `k` 个元素中最小的，也就是整体第 `k` 大。

时间复杂度 `O(n log k)`，空间复杂度 `O(k)`。

````python
import heapq

def kth_largest_with_heap(nums, k):
    heap = []
    for value in nums:
        heapq.heappush(heap, value)
        if len(heap) > k:
            heapq.heappop(heap)
    return heap[0]


print(kth_largest_with_heap([3,2,1,5,6,4], 2))  # 5
````

## 7.4 最大堆与易错点

Python 标准刷题写法通常用最小堆。需要最大堆时，常压入负数：

```python
heappush(heap, -value)
largest = -heappop(heap)
```

易错点：

- 认为堆列表从左到右完全有序；
- 忘记取负数后弹出时再转回正数；
- 已经可以一次线性扫描，却无必要地使用堆；
- 前 K 大使用大小为 K 的最小堆时，误弹出最大值；
- 忘记元组比较会继续比较后面的字段。

# 8. 图与网格

## 8.1 核心模型

图由节点（vertex）和边（edge）组成。

- 无向图：连接可以双向走；
- 有向图：边有方向，例如课程依赖；
- 有权图：边上还有距离或成本；
- 无权图：只关心是否连接或经过几条边。

树也是图，但树通常连通、无环，并有清晰层级。普通图可能有环，所以遍历时通常必须使用 `visited`。

## 8.2 邻接表

Hot 100 中常用邻接表表示图：

```python
graph[node] = [它能到达的邻居]
```

与邻接矩阵相比，邻接表更适合边较少的图，空间约 `O(V+E)`。

构建无向图时，一条边 `[a,b]` 要添加两次：

```python
graph[a].append(b)
graph[b].append(a)
```

构建有向图时只按方向添加。

````python
from collections import defaultdict

edges = [(0, 1), (0, 2), (1, 3), (2, 3)]
graph = defaultdict(list)

for start, end in edges:
    graph[start].append(end)
    graph[end].append(start)  # 无向图

print(dict(graph))
````

## 8.3 图的 DFS 与 BFS

### DFS：一路走深，再回退

适合：

- 判断一个连通块；
- 枚举路径；
- 岛屿数量；
- 克隆图。

### BFS：按距离一层层扩展

适合：

- 无权图最短步数；
- 多源扩散；
- 腐烂的橘子。

两者时间通常都是 `O(V+E)`。差别主要是访问顺序和能解决的问题。

````python
def graph_dfs(graph, start):
    visited = set()
    order = []

    def dfs(node):
        if node in visited:
            return
        visited.add(node)
        order.append(node)
        for neighbor in graph[node]:
            dfs(neighbor)

    dfs(start)
    return order


print(graph_dfs(graph, 0))
````

## 8.4 二维网格也是隐式图

在岛屿数量中，每个格子可以看作节点，上下左右相邻格子之间有边。

不需要真的创建邻接表，只需要方向数组：

```python
directions = [(1,0), (-1,0), (0,1), (0,-1)]
```

网格 DFS/BFS 的固定检查：

1. 新坐标是否越界；
2. 是否满足可访问条件；
3. 是否已经访问过；
4. 什么时候标记访问。

常见题：岛屿数量、腐烂的橘子、单词搜索。

````python
grid = [
    ['1', '1', '0'],
    ['0', '1', '0'],
    ['1', '0', '1'],
]

rows, cols = len(grid), len(grid[0])
directions = [(1,0), (-1,0), (0,1), (0,-1)]

row, col = 0, 0
neighbors = []
for dr, dc in directions:
    nr, nc = row + dr, col + dc
    if 0 <= nr < rows and 0 <= nc < cols:
        neighbors.append((nr, nc))

print('合法相邻格子：', neighbors)
````

## 8.5 拓扑排序：处理依赖关系

课程表问题中，先修课形成有向图。

拓扑排序常使用：

- 邻接表：一门课完成后会解锁哪些课程；
- 入度数组：每门课还有几个前置条件；
- 队列：当前所有入度为 0、可以开始的课程。

每弹出一门课程，就把它指向课程的入度减 1。若最终处理数量等于课程总数，则不存在阻塞循环。

这是“图 + 数组 + 队列”的组合，而不是一种单独的数据结构。

````python
from collections import deque

def can_finish_courses(num_courses, prerequisites):
    graph = [[] for _ in range(num_courses)]
    indegree = [0] * num_courses

    for course, prerequisite in prerequisites:
        graph[prerequisite].append(course)
        indegree[course] += 1

    queue = deque(i for i in range(num_courses) if indegree[i] == 0)
    completed = 0

    while queue:
        course = queue.popleft()
        completed += 1
        for next_course in graph[course]:
            indegree[next_course] -= 1
            if indegree[next_course] == 0:
                queue.append(next_course)

    return completed == num_courses


print(can_finish_courses(2, [(1, 0)]))
print(can_finish_courses(2, [(1, 0), (0, 1)]))
````

## 8.6 图题易错点

- 无向边只添加了一个方向；
- 忘记 `visited`，有环时无限搜索；
- 标记访问太晚，导致同一节点重复入队；
- 混淆 DFS 能否直接得到最短路径；
- 网格坐标越界；
- 课程依赖边方向写反；
- 把树题的“只有一个父节点”假设带到普通图。

面试表达要明确：节点是什么、边是什么、图是有向还是无向、搜索状态如何去重。

# 9. 前缀树 Trie

## 9.1 核心模型

Trie 专门按字符前缀组织字符串：

```text
root
 ├─ c ─ a ─ t*
 │       └─ r*
 └─ d ─ o ─ g*
```

公共前缀共享路径。`*` 表示某个单词在这里结束。

它适合：

- 插入单词；
- 查询完整单词；
- 查询是否存在某个前缀。

操作时间与单词长度 `L` 有关，通常为 `O(L)`。

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_word = False


class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_word = True

    def search(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_word

    def starts_with(self, prefix):
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False
            node = node.children[char]
        return True


trie = Trie()
trie.insert('apple')
print(trie.search('apple'))
print(trie.search('app'))
print(trie.starts_with('app'))
````

## 9.2 Hot 100 常见用法与误区

常见题：

- 实现 Trie；
- 单词搜索 II：Trie + 网格回溯；
- 大量字符串前缀查询。

Trie 节点的 `children` 本身通常是哈希表，所以 Trie 是“树结构 + 哈希映射”的组合。

易错点：

- 只检查路径存在，没有检查 `is_word`，把前缀误认为完整单词；
- 每次查询都从错误节点开始，应从根开始；
- 插入完成后忘记标记单词结尾；
- 字符集很大时，每个节点使用固定大数组可能浪费空间。

# 10. 并查集 Union-Find

## 10.1 它解决什么问题

并查集维护若干互不相交的集合，支持：

- `find(x)`：找到 `x` 所属集合的代表；
- `union(a, b)`：合并 `a`、`b` 所在集合；
- 判断两点是否连通：`find(a) == find(b)`。

适合“边不断加入，动态判断连通性”的问题。

它不擅长直接给出具体路径，也不适合频繁删除连接。

````python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.size = [1] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, a, b):
        root_a = self.find(a)
        root_b = self.find(b)
        if root_a == root_b:
            return False

        if self.size[root_a] < self.size[root_b]:
            root_a, root_b = root_b, root_a
        self.parent[root_b] = root_a
        self.size[root_a] += self.size[root_b]
        return True

    def connected(self, a, b):
        return self.find(a) == self.find(b)


uf = UnionFind(5)
uf.union(0, 1)
uf.union(1, 2)
print(uf.connected(0, 2))
print(uf.connected(0, 4))
````

## 10.2 路径压缩与按大小合并

### 路径压缩

`find` 时让沿途节点直接指向代表节点，之后查找更快。

### 按大小或秩合并

让小树挂到大树下面，避免结构变成很长的链。

两种优化一起使用后，单次操作的均摊复杂度非常接近 `O(1)`，严谨写法是 `O(α(n))`，其中 `α` 是增长极慢的反阿克曼函数。

Hot 100 中图连通题可以用 DFS/BFS，也有一些可以用并查集。面试时应说明为何选择它：是否需要反复合并并查询连通性。

## 10.3 常见使用信号与误区

使用信号：

- 有若干组关系，需要不断合并；
- 判断两个元素是否属于同一组；
- 加入一条边后是否形成环；
- 统计连通分量数量。

常见题型：冗余连接、账户合并、岛屿连接问题。

易错点：

- `union` 直接连接原节点而不是连接两个根；
- 忘记路径压缩；
- 连通分量计数在重复合并时也错误减少；
- 以为并查集可以方便地删除一条边；
- 需要最短路径时误用并查集。

# 11. 复合数据结构：真正的面试题如何组合

## 11.1 前缀和 + 哈希表

代表题：和为 K 的子数组。

- 前缀和负责把一段区间和转换成两个历史状态之差；
- 哈希表负责快速查询某个历史前缀和出现多少次。

```text
算法思想：前缀和
数据结构：哈希表
```

这种拆分能帮助你迁移到“连续子数组计数”问题。

````python
def subarray_sum(nums, k):
    counts = {0: 1}
    prefix = 0
    answer = 0

    for value in nums:
        prefix += value
        answer += counts.get(prefix - k, 0)
        counts[prefix] = counts.get(prefix, 0) + 1

    return answer


print(subarray_sum([1, 1, 1], 2))  # 2
````

## 11.2 哈希表 + 双向链表：LRU 缓存

LRU 需要同时做到：

- `O(1)` 根据 key 找缓存：哈希表；
- `O(1)` 删除任意已知节点：双向链表；
- `O(1)` 找最久未使用项：链表尾部；
- `O(1)` 把最近使用项移到头部。

单独使用字典不能快速维护使用顺序；单独使用链表不能快速按 key 查找。

这类设计题的核心不是记代码，而是把每项复杂度要求分配给合适结构。

## 11.3 单调结构

### 单调栈

栈 + 主动维护单调性：

- 每日温度；
- 柱状图最大矩形；
- 接雨水。

### 单调队列

双端队列 + 主动维护单调性：

- 滑动窗口最大值。

“单调”不是 Python 提供的新容器，而是通过弹出无用候选，使结构中只保留未来仍有价值的元素。

## 11.4 图 + 入度 + 队列

拓扑排序包含：

- 邻接表表达依赖边；
- 入度数组统计未完成前置条件；
- 队列保存当前可执行节点。

### 15.1 参考答案

只需要判断是否出现过，使用集合。

````python
def first_duplicate_answer(nums):
    seen = set()
    for value in nums:
        if value in seen:
            return value
        seen.add(value)
    return None


assert first_duplicate_answer([3,1,4,2,1,3]) == 1
assert first_duplicate_answer([1,2,3]) is None
````

### 15.2 参考答案

这里可以使用栈保存左括号下标；遇到右括号时从栈顶匹配。

````python
def first_unmatched_right_answer(s):
    stack = []
    for index, char in enumerate(s):
        if char == '(':
            stack.append(index)
        elif stack:
            stack.pop()
        else:
            return index
    return -1


assert first_unmatched_right_answer('(())') == -1
assert first_unmatched_right_answer('())') == 2
````

### 15.3 参考答案

无向边加入两个方向；节点入队时立即标记，避免重复入队。

````python
def valid_path_answer(n, edges, source, destination):
    graph = [[] for _ in range(n)]
    for a, b in edges:
        graph[a].append(b)
        graph[b].append(a)

    queue = deque([source])
    visited = {source}

    while queue:
        node = queue.popleft()
        if node == destination:
            return True
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return False


assert valid_path_answer(3, [(0,1),(1,2)], 0, 2) is True
assert valid_path_answer(4, [(0,1),(2,3)], 0, 3) is False
````
