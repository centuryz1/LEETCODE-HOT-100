### 输入、输出与限制

- 两个头节点都可能是 `None`。
- 可以复用原链表节点，不必复制节点值。
- 输入中允许出现重复值，输出仍保持非递减顺序。

专业名称：**双指针归并（Two-Pointer Merge）**。

时间复杂度 `O(m+n)`，额外空间复杂度 `O(1)`。

## 题目 2 做题前先补：`divmod` 和进位

两个个位数加上旧进位得到 `total` 后：

```python

```

`divmod(a, b)` 同时返回 `(a // b, a % b)`：

- `carry` 是下一位要使用的进位。
- `digit` 是当前结果节点保存的个位数。

循环条件要写成 `while l1 or l2 or carry`，因为两个链表都结束后，仍可能剩下最后一个进位。

````python
# 完整参考答案：环形链表 II
# 本格可以单独运行，不依赖前面的任何代码格。

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def build_cycle_list(values, pos=-1):
    """创建链表；pos >= 0 时让尾节点指向下标 pos，形成环。"""
    if not values:
        return None, None

    nodes = [ListNode(value) for value in values]
    for index in range(len(nodes) - 1):
        nodes[index].next = nodes[index + 1]

    entry = None
    if pos >= 0:
        nodes[-1].next = nodes[pos]
        entry = nodes[pos]
    return nodes[0], entry


def detect_cycle_answer(head):
    slow = fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            finder = head
            while finder is not slow:
                finder = finder.next
                slow = slow.next
            return finder
    return None


cycle_head, expected_entry = build_cycle_list([3, 2, 0, -4], pos=1)
normal_head, _ = build_cycle_list([1, 2, 3], pos=-1)
assert detect_cycle_answer(cycle_head) is expected_entry
assert detect_cycle_answer(normal_head) is None
print('环入口：', detect_cycle_answer(cycle_head).val)
````

我先用 Floyd 快慢指针找到环内相遇点，再让一个指针回到头节点，与相遇点指针同速前进，两者相遇处就是环入口。时间 O(n)，空间 O(1)。
