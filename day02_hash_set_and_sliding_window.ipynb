```python
# 昨日复习 1：两数之和
# 解法：一次遍历哈希表（One-pass Hash Table）
# 输入：整数列表 nums 和整数 target。
# 输出：两个不同元素的下标。

def review_two_sum(nums, target):
    seen = {}

    for i,x in enumerate(nums):
        need = target - x
        if need in seen:
            return[seen[need],i]
        else:
            seen[x] = i
        
    return []

print(review_two_sum([2, 7, 11, 15], 9))  # 期望 [0, 1]
```

```python
# 昨日复习 2：移动零
# 解法：快慢指针 / 原地交换（Fast-Slow Pointers / In-place Swap）
# 输入：整数列表 nums。
# 输出：不需要返回，原地修改 nums。

def review_move_zeroes(nums):
    slow = 0

    for fast in range(0,len(nums)):
        if nums[fast] != 0:
            nums[slow],nums[fast] = nums[fast],nums[slow]
            slow +=1

nums = [0, 1, 0, 3, 12]
review_move_zeroes(nums)
print(nums)  # 期望 [1, 3, 12, 0, 0]
```

# 题目 1：最长连续序列

LeetCode 128. Longest Consecutive Sequence

## 题目描述（改写版）

给你一个没有排序的整数列表 `nums`，请你找出其中最长的连续数字序列的长度。

连续数字序列指的是数字一个接一个，比如 `1, 2, 3, 4`。这些数字在原列表里不一定挨在一起，也不要求原列表本身有序。

需要注意：

- 返回的是长度，不是序列本身。
- 原数组没有排序。
- 数字可以是负数。
- 可能有重复数字，重复数字不应该影响连续长度。

示例：`nums = [100, 4, 200, 1, 3, 2]`，最长连续序列是 `[1, 2, 3, 4]`，返回 `4`。


## 解法名称

本题常用专业说法：**哈希集合 + 序列起点判断**，英文可以说 **Hash Set with Sequence Start Detection**。

面试里你可以说：我先把所有数字放进哈希集合，然后只从连续序列的起点开始向后扩展。


## 思路拆解

暴力做法是对每个数字都往后查 `x + 1`、`x + 2` 是否存在，但这样会重复扩展。

优化关键：只有当 `x - 1` 不存在时，`x` 才是一个连续序列的起点。


```python
nums = [100, 4, 200, 1, 3, 2]
num_set = set(nums)

for x in num_set:
    if x - 1 not in num_set:
        print(x, '是起点')
    else:
        print(x, '不是起点')
```

## 你来写：最长连续序列

动手前再确认一遍：

- 输入：整数列表 `nums`。
- 目标：找到最长连续数字序列的长度。
- 输出：一个整数。
- 注意：原数组无序，重复数字不影响答案。


```python
# 题目：最长连续序列
# 解法：哈希集合 + 序列起点判断（Hash Set with Sequence Start Detection）
# 输入：整数列表 nums。
# 目标：找到最长连续数字序列的长度。
# 输出：最长长度，是一个整数。
# 注意：原数组无序，重复数字不影响答案。

def longest_consecutive(nums):
    set_nums = set(nums)
    ans = 0

    for i in set_nums:
        if i-1 not in set_nums:
            cur = i
            length = 1
            while cur+1 in set_nums:
                cur += 1
                length += 1
            ans = max(ans,length)

    return ans


```

```python
print(longest_consecutive([100, 4, 200, 1, 3, 2]))  # 期望 4
print(longest_consecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))  # 期望 9
print(longest_consecutive([]))  # 期望 0
```

## 参考答案：最长连续序列


```python
def longest_consecutive_answer(nums):
    num_set = set(nums)
    ans = 0

    for x in num_set:
        if x - 1 not in num_set:
            cur = x
            length = 1
            while cur + 1 in num_set:
                cur += 1
                length += 1
            ans = max(ans, length)

    return ans

print(longest_consecutive_answer([100, 4, 200, 1, 3, 2]))
print(longest_consecutive_answer([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))
print(longest_consecutive_answer([]))
```

面试表达：

这题我使用哈希集合加序列起点判断。先把所有数字放进集合，方便 O(1) 查询。然后只从 `x - 1` 不存在的数字开始扩展连续序列，避免重复扩展。时间复杂度 O(n)，空间复杂度 O(n)。


# 题目 2：三数之和

LeetCode 15. 3Sum

## 题目描述（改写版）

给你一个整数列表 `nums`，请你找出所有不重复的三元组 `[a, b, c]`，使得 `a + b + c == 0`。

需要注意：

- 返回的是数字三元组，不是下标。
- 一个三元组里必须使用三个不同位置的元素。
- 结果里不能有重复三元组。
- 三元组内部顺序不重要，最终答案顺序也不重要。

示例：`nums = [-1, 0, 1, 2, -1, -4]`，一种返回结果是 `[[-1, -1, 2], [-1, 0, 1]]`。


## 解法名称

本题常用专业说法：**排序 + 双指针 + 去重**，英文可以说 **Sorting + Two Pointers with Deduplication**。


## 思路拆解

三层循环是 O(n^3)，太慢。更好的方式是先排序，固定第一个数，再在右侧用 `left` 和 `right` 找另外两个数。

- 和小了，`left += 1`
- 和大了，`right -= 1`
- 找到答案后，跳过重复值


```python
nums = [-1, 0, 1, 2, -1, -4]
nums.sort()
print(nums)

i = 0
left, right = i + 1, len(nums) - 1
while left < right:
    total = nums[i] + nums[left] + nums[right]
    print(nums[i], nums[left], nums[right], total)
    if total < 0:
        left += 1
    elif total > 0:
        right -= 1
    else:
        break
```

## 参考答案：三数之和


```python
def three_sum_answer(nums):
    nums.sort()
    ans = []
    n = len(nums)

    for i in range(n):
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left, right = i + 1, n - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                ans.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1
                while left < right and nums[left] == nums[left - 1]:
                    left += 1
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1

    return ans

print(three_sum_answer([-1, 0, 1, 2, -1, -4]))
print(three_sum_answer([0, 1, 1]))
print(three_sum_answer([0, 0, 0]))
```

面试表达：

# 题目 3：无重复字符的最长子串

LeetCode 3. Longest Substring Without Repeating Characters

## 题目描述（改写版）

给你一个字符串 `s`，请你找出其中不包含重复字符的最长连续子串长度。

需要注意：

- 子串必须是连续的。
- 返回的是长度，不是子串本身。
- 字符可以是字母、数字、符号、空格等。
- 空字符串的答案是 0。

示例：`s = 'abcabcbb'`，最长不重复子串可以是 `'abc'`，返回 `3`。


## 解法名称

本题常用专业说法：**滑动窗口 + 哈希集合**，英文可以说 **Sliding Window + Hash Set**。


## 思路拆解

窗口 `[left, right]` 表示当前正在看的连续子串。

1. `right` 向右扩展。
2. 如果新字符没出现过，就加入窗口。
3. 如果新字符已经在窗口里，就移动 `left`，直到窗口里没有重复。
4. 每次窗口合法时，更新最大长度。


```python
s = 'abcabcbb'
seen = set()
left = 0

for right, ch in enumerate(s):
    while ch in seen:
        print('遇到重复:', ch, '移出:', s[left])
        seen.remove(s[left])
        left += 1
    seen.add(ch)
    print('窗口:', s[left:right + 1])
```

# 题目 4：找到字符串中所有字母异位词

## 题目描述（改写版）

给你两个字符串 `s` 和 `p`，请你在 `s` 中找到所有子串，使这些子串是 `p` 的字母异位词，并返回这些子串的起始下标。

需要注意：

- 子串必须是连续的。
- 子串长度必须和 `p` 一样。
- 字母异位词表示字母种类和出现次数相同，但顺序可以不同。
- 返回所有符合条件的起始下标。

示例：`s = 'cbaebabacd'`, `p = 'abc'`，返回 `[0, 6]`。


## 解法名称

本题常用专业说法：**固定长度滑动窗口 + 字符频次计数**，英文可以说 **Fixed-size Sliding Window + Frequency Count**。


## 思路拆解

目标子串长度必须等于 `len(p)`，所以窗口大小固定。每次右边加入新字符，如果窗口超过长度，就移出左边字符。只要窗口频次和 `p` 的频次相同，就记录窗口起点。


```python
s = 'cbaebabacd'
p = 'abc'
k = len(p)

for right, ch in enumerate(s):
    left = right - k + 1
    if left >= 0:
        print('窗口:', s[left:right + 1])
```

## 参考答案：找到字符串中所有字母异位词


```python
from collections import Counter

def find_anagrams_answer(s, p):
    need = Counter(p)
    window = Counter()
    ans = []
    k = len(p)

    for right, ch in enumerate(s):
        window[ch] += 1
        if right >= k:
            left_ch = s[right - k]
            window[left_ch] -= 1
            if window[left_ch] == 0:
                del window[left_ch]
        if window == need:
            ans.append(right - k + 1)

    return ans

print(find_anagrams_answer('cbaebabacd', 'abc'))
print(find_anagrams_answer('abab', 'ab'))
```
