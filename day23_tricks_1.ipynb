## 题目 1 做题前先补：异或（XOR）的三个性质

异或写作 `^`，规则是“相同为 0，不同为 1”。它有三个特别好用的性质：

1. `a ^ a = 0`：任何数和自己异或都等于 0；
2. `a ^ 0 = a`：任何数和 0 异或还是自己；
3. **满足交换律和结合律**：`a ^ b ^ c` 的顺序可以随便调。

把数组里所有数全部异或起来会发生什么？成对出现的数字两两抵消成 0，最后只剩下那个只出现一次的数。

这就是本题的全部思路：

```python
result = 0
for num in nums:
    result = result ^ num
return result
```

**注意：这是整个 Hot 100 里唯一建议用位运算的题目**，面试时可以直接用，但要能解释清楚为什么。


````python
# 手工验证异或的性质
print("5 ^ 5 =", 5 ^ 5)
print("5 ^ 0 =", 5 ^ 0)
print("顺序无关：", (1 ^ 2 ^ 3) == (3 ^ 1 ^ 2))

nums = [4, 1, 2, 1, 2]
result = 0
for num in nums:
    result = result ^ num
    print("异或到", num, "，结果变成", result)
print("只出现一次的数字：", result)
````

## 题目 2 做题前先补：多数元素一定“抵消不完”

多数元素的定义：出现次数**大于 n/2** 的元素。题目保证它一定存在。

假设把数组里的数字两两配对，配到的两个不同就互相“抵消”掉。因为多数元素出现次数超过一半，**就算它每一次都被拿来抵消，最后也一定会剩下它**。

这就是**摩尔投票法**：

- 维护一个候选人 `candidate` 和一个计数器 `count`；
- 遇到和候选人相同的数，`count` 加一（增加支持）；
- 遇到不同的数，`count` 减一（抵消一个）；
- 当 `count` 变成 0 时，说明前面这段被抵消完了，换当前数字当新候选人。

走完全程后，剩下的候选人就是多数元素。

**注意：这个方法成立的前提是“多数元素一定存在”。** 如果不存在，还需要再扫一遍验证。


````python
# 手工走一遍 [2, 2, 1, 1, 1, 2, 2]
nums = [2, 2, 1, 1, 1, 2, 2]
candidate = nums[0]
count = 0

for num in nums:
    if count == 0:
        candidate = num
    if num == candidate:
        count = count + 1
    else:
        count = count - 1
    print("处理", num, "：候选人", candidate, "，计数", count)

print("多数元素：", candidate)
````

## 题目 3 做题前先补：三个指针怎么分工

颜色分类：数组里只有 0、1、2 三种值，要求原地排成 0、0、1、1、2、2。

用三个指针一趟扫描：

- `low`：下一个 0 应该放的位置（它左边全是 0）；
- `mid`：当前正在检查的位置；
- `high`：下一个 2 应该放的位置（它右边全是 2）。

分三种情况处理 `nums[mid]`：

- 是 0：和 `nums[low]` 交换，`low` 和 `mid` 都前进一格（换过来的数一定是 1 或 0，已经处理过了）；
- 是 1：本来就是正确区间，`mid` 前进一格；
- 是 2：和 `nums[high]` 交换，`high` 后退一格，**`mid` 不动**（因为换回来的数还没检查过）。

**最容易错的是第三种情况：`mid` 不能动。** 在纸上画一遍就清楚了。


````python
# 手工走一遍 [2, 0, 2, 1, 1, 0]
nums = [2, 0, 2, 1, 1, 0]
low = 0
mid = 0
high = len(nums) - 1

while mid <= high:
    if nums[mid] == 0:
        nums[low], nums[mid] = nums[mid], nums[low]
        low = low + 1
        mid = mid + 1
        print("遇到 0：和 low 交换，", nums, "，low =", low, "，mid =", mid)
    elif nums[mid] == 1:
        mid = mid + 1
        print("遇到 1：mid 前进，", nums, "，mid =", mid)
    else:
        nums[mid], nums[high] = nums[high], nums[mid]
        high = high - 1
        print("遇到 2：和 high 交换，", nums, "，high =", high, "，mid 不动 =", mid)

print("最终结果：", nums)
````

## 题目 4 做题前先补：什么叫“下一个排列”

把数组看成一个数字的各位，下一个排列就是**字典序里刚好比它大的下一个**。

比如 `[1,2,3]` 的下一个是 `[1,3,2]`；`[3,2,1]` 已经是最大的，下一个就回到最小的 `[1,2,3]`。

找下一个排列分三步，在纸上画一遍最好懂：

**第一步：从右往左找第一个“左边比右边小”的位置 `i`。**
说明 `nums[i]` 还有变大的空间（`i` 右边是降序的，已经是这几个数字能组成的最小排列）。

**第二步：从右往左找第一个比 `nums[i]` 大的位置 `j`，交换 `i` 和 `j`。**

**第三步：把 `i` 右边的部分反转**，让它们从降序变成升序，也就是让右边尽可能小。

如果第一步找不到这样的 `i`（整个数组是降序的），说明已经是最大排列，直接反转整个数组即可。


````python
# 手工走一遍 [1, 3, 5, 4, 2]
nums = [1, 3, 5, 4, 2]

i = len(nums) - 2
while i >= 0 and nums[i] >= nums[i + 1]:
    i = i - 1
print("第一个左边比右边小的位置：", i, "，值是", nums[i] if i >= 0 else None)

if i >= 0:
    j = len(nums) - 1
    while nums[j] <= nums[i]:
        j = j - 1
    print("从右往左第一个比它大的位置：", j, "，值是", nums[j])
    nums[i], nums[j] = nums[j], nums[i]
    print("交换后：", nums)

left = i + 1
right = len(nums) - 1
while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left = left + 1
    right = right - 1
print("反转后半段后：", nums)
````

- 原地修改，不返回新数组。
- 要允许重复元素（所以判断用 `>=`）。
- 第三步的反转不能忘，否则结果不是最小的那个更大排列。


## 解法名称

**三步法（Next Permutation in Place）**。

## 暴力思路

枚举所有排列并排序，找到当前排列的下一个，指数级。

## 优化思路

三步：

1. 从右往左找第一个 `nums[i] < nums[i+1]` 的 `i`；
2. 如果找到了，再从右往左找第一个 `nums[j] > nums[i]` 的 `j`，交换；
3. 把 `i+1` 到末尾这一段反转（原来是降序，反转后变升序）。

如果第一步没找到，说明整个数组是降序，直接反转整个数组。

时间 O(n)，空间 O(1)。

def next_permutation_answer(nums):
    i = len(nums) - 2
    while i >= 0 and nums[i] >= nums[i + 1]:
        i = i - 1

    if i >= 0:

面试表达：

要得到下一个排列，我需要找到一个“还能变大的位置”。从右往左找第一个满足 `nums[i] < nums[i+1]` 的下标 i，因为 i 右边是降序的，说明右边已经是这段数字能组成的最小排列，只有 i 这个位置还能换更大的数上来。然后从右往左找第一个大于 `nums[i]` 的数，和它交换；交换之后，i 右边仍然是降序的，而我们要让整体尽可能小，所以把 i 右边反转成升序。如果一开始就没找到这样的 i，说明整个数组是降序，也就是最大排列，直接反转整个数组变成最小排列。时间 O(n)，空间 O(1)。


````python


def next_permutation_answer(nums):
    i = len(nums) - 2
    while i >= 0 and nums[i] >= nums[i + 1]:
        i = i - 1

    if i >= 0:
        j = len(nums) - 1
        while nums[j] <= nums[i]:
            j = j - 1
        nums[i], nums[j] = nums[j], nums[i]

    left = i + 1
    right = len(nums) - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left = left + 1
        right = right - 1


nums1 = [1, 2, 3]
next_permutation_answer(nums1)
print(nums1)      # [1, 3, 2]
nums2 = [3, 2, 1]
next_permutation_answer(nums2)
print(nums2)      # [1, 2, 3]
nums4 = [1, 3, 5, 4, 2]
next_permutation_answer(nums4)
print(nums4)      # [1, 4, 2, 3, 5]
````
