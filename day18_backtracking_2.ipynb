## 题目 1 做题前先补：边构造边剪枝

生成括号不是“先拼出所有字符串再检查合法性”，而是**一边拼一边保证不违规**。

合法的括号串有两个铁律：

1. 任意前缀里，**左括号数量不能少于右括号数量**（否则出现 `)(` 这种非法情况）；
2. 最终左括号和右括号各有 n 个。

所以做选择时只需要两个条件：

- 左括号还没放够 n 个：可以放 `(`；
- 右括号数量少于左括号数量：可以放 `)`。

再看 `n = 2` 的选择树：

```text
                 ""
          (              ← 只能放左括号
         "("
       (      )          ← 左括号还能放；右括号少于左括号也能放
     "(("    "()"
      )        (         ← "((" 只能放右括号；"()" 只能放左括号
    "(()"    "()("
      )        )
    "(())"   "()()"   ✓
```

在树上走的时候，只要不满足条件的分支直接不生成，这就是**剪枝**。


````python
# 手工看一次选择过程：n = 2
n = 2
path = []


def demo(open_count, close_count):
    print("path =", "".join(path), "，左括号", open_count, "个，右括号", close_count, "个")
    if len(path) == 2 * n:
        print("    得到一个合法结果")
        return
    if open_count < n:
        path.append("(")
        demo(open_count + 1, close_count)
        path.pop()
    if close_count < open_count:
        path.append(")")
        demo(open_count, close_count + 1)
        path.pop()


demo(0, 0)
````

## 题目 2 做题前先补：回溯和普通 DFS 的区别

Day 16 的岛屿数量也用 DFS，但那题**不需要恢复现场**，因为陆地淹掉就永远淹掉了。

单词搜索不一样：我们要找的是“某条路径能不能拼出单词”。假设从某个格子出发走错了，回到上一层时，**这个格子必须恢复原样**，因为另一条路径可能还要用同一个格子。

这就是回溯和普通 DFS 的关键区别：

```python
board[row][col] = "#"      # 标记：本次路径用过这个格子了
...递归四个方向...
board[row][col] = 原来字符   # 恢复：这一步不能忘
```

另外还要注意：格子可以被同一条路径以外的情况重复使用，所以标记只用临时的 `#`，递归返回前一定要还原成原来的字母。


````python
# 手工看一次“标记 + 恢复”
board = [
    ["A", "B"],
    ["C", "D"],
]


def demo(row, col, depth):
    if depth > 2:
        return
    temp = board[row][col]
    board[row][col] = "#"
    print("    进入", row, col, "，它是", temp, "，标记后 board =", board)
    if row + 1 < len(board):
        demo(row + 1, col, depth + 1)
    board[row][col] = temp
    print("    离开", row, col, "，恢复成", temp, "，board =", board)


demo(0, 0, 0)
````

## 输入

- `board`：1 到 6 行，1 到 6 列，只含大小写英文字母。
- `word`：长度 1 到 15。

## 输出

存在返回 `True`，否则返回 `False`。

## 示例

示例 1：

```text

```

`word = "ABCCED"` 返回 `True`；`word = "SEE"` 返回 `True`；`word = "ABCB"` 返回 `False`（B 会被重复使用）。

## 易漏细节

- 每个格子只能用一次，所以要标记；但**递归返回前必须恢复**，否则别的起点就没法用了。
- 字符不匹配要立刻返回。
- 匹配到单词末尾（`index == len(word)`）就返回 `True`。


## 解法名称

**网格回溯（Backtracking on Grid）**。

## 暴力思路

枚举所有起点和所有路径。回溯本身就是这里最标准的做法。

## 优化思路

双循环找起点，每个起点做一次 DFS：

1. `index` 表示当前要匹配 `word` 的第几个字符；
2. 如果 `index == len(word)`，说明全匹配上了，返回 `True`；
3. 越界或者当前格子字符不对，返回 `False`；
4. 标记当前格子（改成 `#`），递归四个方向；
5. **无论成功与否，返回前都要还原格子**；
6. 四个方向只要有一个成功，整体就成功。

时间 O(行 × 列 × 3 的 L 次方)，L 是单词长度（每次最多往 3 个方向走，因为来路已经标记过）；空间是递归深度 O(L)。

    def dfs(row, col, index):

## 题目 3 做题前先补：把“切分”看成一棵选择树

分割回文串其实是在决定：**第一刀切在哪、第二刀切在哪……**

以 `s = "aab"` 为例：

```text
从下标 0 开始切：
├─ 切出 "a"（回文）→ 剩下的 "ab" 继续
│   ├─ 切出 "a"（回文）→ 剩下 "b"
│   │   └─ 切出 "b"（回文）→ 切完了 ✓ ["a","a","b"]
│   └─ 切出 "ab"（不是回文）✗ 丢弃
└─ 切出 "aa"（回文）→ 剩下的 "b"
    └─ 切出 "b"（回文）→ 切完了 ✓ ["aa","b"]
```

所以：

- 循环变量 `end` 从 `start` 到字符串末尾，表示“这一刀切在哪里”；
- 切出来的片段 `s[start:end+1]` 是回文才继续；
- 递归的下一层从 `end + 1` 开始；
- 当 `start == len(s)` 时说明切完了，收集答案。

判断回文用双指针，不用切片倒序比较（那样每次都要额外空间）。


````python
# 手工看一次切片和回文判断
s = "aab"


def is_palindrome(left, right):
    while left < right:
        if s[left] != s[right]:
            return False
        left = left + 1
        right = right - 1
    return True


for end in range(len(s)):
    part = s[0:end + 1]
    print("切出", part, "，是不是回文：", is_palindrome(0, end))
````

## 参考答案：分割回文串

```python
def partition_answer(s):
    result = []
    path = []

    def is_palindrome(left, right):
        while left < right:
            if s[left] != s[right]:
                return False
            left = left + 1
            right = right - 1
        return True

    def backtrack(start):
        if start == len(s):
            result.append(path[:])
            return
        for end in range(start, len(s)):
            if not is_palindrome(start, end):
                continue
            path.append(s[start:end + 1])
            backtrack(end + 1)
            path.pop()

    backtrack(0)
    return result
```

面试表达：

我用回溯，参数 `start` 表示当前从哪里开始切。函数进来先判断是不是已经切到字符串末尾，是的话说明这一整条分割路径都合法，把 `path` 复制一份收集起来。然后循环尝试把 `start` 到 `end` 这一段切出来，如果它是回文，就加入 `path`，递归从 `end + 1` 继续切剩下的部分，返回后再把它弹出来。判断回文我用双指针，从两头往中间比较。时间最坏是指数级，空间 O(n)。


````python


def partition_answer(s):
    result = []
    path = []

    def is_palindrome(left, right):
        while left < right:
            if s[left] != s[right]:
                return False
            left = left + 1
            right = right - 1
        return True

    def backtrack(start):
        if start == len(s):
            result.append(path[:])
            return
        for end in range(start, len(s)):
            if not is_palindrome(start, end):
                continue
            path.append(s[start:end + 1])
            backtrack(end + 1)
            path.pop()

    backtrack(0)
    return result


print(sorted(partition_answer("aab")))       # [['a', 'a', 'b'], ['aa', 'b']]
print(partition_answer("a"))                 # [['a']]
print(len(partition_answer("abc")))          # 1
print(len(partition_answer("aaaa")))         # 8
````

## 题目 4 做题前先补：N 皇后的规则和判断

N 皇后要在 N × N 的棋盘上放 N 个皇后，满足：

- 每一行只能有一个；
- 每一列只能有一个；
- 每条斜线上只能有一个。

**按行放**是最省事的做法：一行一行往下放，每行只放一个，所以“行不冲突”自动满足。放之前只要检查三件事：

1. 这一列有没有被占；
2. 左上斜线有没有被占；
3. 右上斜线有没有被占。

棋盘用二维列表表示，`"."` 是空格，`"Q"` 是皇后。判断能不能放的代码单独写成一个函数，逻辑就清爽了。


````python
# 看一次斜线检查：在 4x4 棋盘上，第一行放 (0,1)，第二行 (1,3) 能不能放？
n = 4
board = []
for i in range(n):
    board.append(["."] * n)
board[0][1] = "Q"
row = 1
col = 3

# 检查列
col_ok = True
for i in range(row):
    if board[i][col] == "Q":
        col_ok = False
print("列 ", col, " 可以放吗：", col_ok)

# 检查左上斜线
i = row - 1
j = col - 1
left_ok = True
while i >= 0 and j >= 0:
    if board[i][j] == "Q":
        left_ok = False
    i = i - 1
    j = j - 1
print("左上斜线可以放吗：", left_ok)
````

面试表达：

我按行来放皇后，这样行冲突自动就避免了。递归函数参数是当前要放的行号；如果行号等于 n，说明所有皇后都放好了，就把棋盘每一行拼成字符串收集起来。否则在当行尝试每一列，用一个合法性函数检查这一列、左上斜线、右上斜线有没有冲突——因为只有上面几行放了皇后，所以斜线只需要往上检查。合法就放上皇后、递归下一行、再撤销。这是一个典型回溯，时间是 O(n!) 级别，但有剪枝会快很多。


````python


def solve_n_queens_answer(n):
    result = []
    board = []
    for i in range(n):
        board.append(["."] * n)

    def is_valid(row, col):
        for i in range(row):
            if board[i][col] == "Q":
                return False

        i = row - 1
        j = col - 1
        while i >= 0 and j >= 0:
            if board[i][j] == "Q":
                return False
            i = i - 1
            j = j - 1

        i = row - 1
        j = col + 1
        while i >= 0 and j < n:
            if board[i][j] == "Q":
                return False
            i = i - 1
            j = j + 1

        return True

    def backtrack(row):
        if row == n:
            one = []
            for line in board:
                one.append("".join(line))
            result.append(one)
            return
        for col in range(n):
            if not is_valid(row, col):
                continue
            board[row][col] = "Q"
            backtrack(row + 1)
            board[row][col] = "."

    backtrack(0)
    return result


print(len(solve_n_queens_answer(4)))     # 2
print(len(solve_n_queens_answer(1)))     # 1
print(len(solve_n_queens_answer(5)))     # 10
print(len(solve_n_queens_answer(6)))     # 4
for one in solve_n_queens_answer(4):
    print(one)
````
