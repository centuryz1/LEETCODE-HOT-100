# 题目 1：和为 K 的子数组

LeetCode 560. Subarray Sum Equals K

## 题目描述（改写版）

给你一个整数数组 `nums` 和一个整数 `k`，请统计数组中有多少个**连续且非空**的子数组，其元素之和恰好等于 `k`。

### 输入是什么

- `nums`：整数列表，里面可以有正数、负数和 `0`。
- `k`：目标和，是一个整数。

### 输出是什么

返回满足条件的连续子数组数量，是一个整数。

### 关键限制和易漏细节

- 子数组必须连续，不能跳着选元素。
- 求的是“有多少个”，不是找最长长度，也不是返回子数组内容。
- 不同起点或不同终点的子数组要分别计数。
- 数组中可能有负数，因此不能简单套用“窗口和大了就缩小”的普通滑动窗口。
- `1 <= len(nums) <= 20000`。

### 示例 1

输入：`nums = [1, 1, 1]`，`k = 2`

输出：`2`

解释：满足条件的是下标范围 `[0, 1]` 和 `[1, 2]`，对应两个 `[1, 1]`。

### 示例 2

输入：`nums = [1, 2, 3]`，`k = 3`

输出：`2`

解释：子数组 `[1, 2]` 与 `[3]` 的和都等于 `3`。

## 解法名称

专业名称：**前缀和 + 哈希表**，英文是 **Prefix Sum + Hash Map**。

核心公式：如果当前位置的前缀和是 `prefix`，以前出现过前缀和 `prefix - k`，那么两者之间那一段子数组的和就是 `k`。

## 暴力思路

枚举每一个起点，再向右累加，遇到和等于 `k` 就把答案加一。

一共有 `O(n²)` 个连续子数组，因此暴力法时间复杂度是 `O(n²)`，数据较大时会超时。

## 优化思路

用字典 `count` 记录“某个前缀和以前出现过多少次”。遍历到当前元素时：

1. 把当前元素加到 `prefix`。
2. 查找 `prefix - k` 出现过多少次。
3. 这些出现次数，就是以当前位置结尾、和为 `k` 的子数组数量。
4. 再把当前 `prefix` 记录进字典。

一开始必须放入 `count = {0: 1}`。它代表“还没选择任何元素时，前缀和为 0 出现过一次”，这样从数组下标 `0` 开始的合法子数组也能被统计。

复杂度：时间 `O(n)`，空间 `O(n)`。

```python
nums = [1, 2, 1, 2]
k = 3
prefix = 0
count = {0: 1}
answer = 0

for index, num in enumerate(nums):
    prefix += num
    need = prefix - k
    found = count.get(need, 0)
    answer += found

    print(f'下标={index}, 当前数={num}, 前缀和={prefix}, 需要找={need}')
    print(f'以前出现 {found} 次，本轮后答案={answer}')

    count[prefix] = count.get(prefix, 0) + 1
    print('前缀和计数字典：', count)
    print('---')
```

# 题目 2：最大子数组和

LeetCode 53. Maximum Subarray

## 题目描述（改写版）

给你一个非空整数数组 `nums`，请找出其中元素和最大的连续非空子数组，并返回这个最大和。

### 输入是什么

- 输入是一个非空整数列表 `nums`。
- 数组中可能全是负数，也可能包含正数和 `0`。

### 输出是什么

返回所有连续非空子数组中最大的元素和，是一个整数。

### 关键限制和易漏细节

- 子数组必须连续且至少包含一个元素。
- 只返回最大和，不要求返回子数组本身。
- 如果所有元素都是负数，答案是其中最大的那个负数，不能默认答案为 `0`。
- `1 <= len(nums) <= 100000`。

### 示例 1

输入：`nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]`

输出：`6`

解释：连续子数组 `[4, -1, 2, 1]` 的和最大，等于 `6`。

### 示例 2

输入：`nums = [-3, -2, -5]`

输出：`-2`

解释：数组全为负数时，选择单独的 `-2` 最优。

## 解法名称

专业名称：**动态规划 / Kadane 算法**，英文是 **Dynamic Programming / Kadane's Algorithm**。

`current` 表示“必须以当前位置结尾的最大子数组和”，`best` 表示遍历到目前为止见过的全局最大和。

## 暴力思路

枚举每个起点和终点，计算每个连续子数组的和。即使一边枚举一边累加，时间复杂度仍是 `O(n²)`。

## 优化思路

来到当前数字 `num` 时，只需要做一个选择：

- 把 `num` 接在前面的连续子数组后面，得到 `current + num`；
- 放弃前面的累积，从 `num` 重新开始。

所以状态转移是：`current = max(num, current + num)`。

每一步再更新：`best = max(best, current)`。

复杂度：时间 `O(n)`，空间 `O(1)`。

```python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
current = nums[0]
best = nums[0]
print(f'下标=0, current={current}, best={best}')

for index in range(1, len(nums)):
    num = nums[index]
    extend = current + num
    restart = num
    current = max(extend, restart)
    best = max(best, current)
    print(f'下标={index}, 当前数={num}, 接上去={extend}, 重新开始={restart}, current={current}, best={best}')
```

# 题目 3：除自身以外数组的乘积

LeetCode 238. Product of Array Except Self

## 题目描述（改写版）

给你一个整数数组 `nums`，请构造一个同样长度的数组 `answer`。其中 `answer[i]` 等于 `nums` 中除了 `nums[i]` 以外，其余所有元素的乘积。

要求不能使用除法，并在 `O(n)` 时间内完成。

### 输入是什么

输入是一个长度至少为 `2` 的整数列表 `nums`，其中可能包含 `0` 和负数。

### 输出是什么

返回新列表 `answer`，输入列表不需要修改。

### 关键限制和易漏细节

- 计算 `answer[i]` 时不能乘 `nums[i]` 自己。
- 不能使用总乘积再除以当前数的方法。
- 数组可能包含一个或多个 `0`，所以除法思路本来也容易出错。
- 输出顺序必须和原数组下标一一对应。
- `2 <= len(nums) <= 100000`。

### 示例 1

输入：`nums = [1, 2, 3, 4]`

输出：`[24, 12, 8, 6]`

解释：下标 `0` 的答案是 `2 * 3 * 4 = 24`；下标 `1` 的答案是 `1 * 3 * 4 = 12`。

### 示例 2

输入：`nums = [-1, 1, 0, -3, 3]`

输出：`[0, 0, 9, 0, 0]`

## 解法名称

专业名称：**前缀积 + 后缀积**，英文是 **Prefix Products + Suffix Products**。

对位置 `i` 来说，除了自己以外的乘积可以拆成：

`左边所有元素的乘积 × 右边所有元素的乘积`

## 暴力思路

对每个下标，都重新遍历整个数组并跳过自己。每个位置花 `O(n)`，总时间复杂度为 `O(n²)`。

## 优化思路

第一遍从左向右：

- 让 `answer[i]` 先保存 `i` 左侧所有元素的乘积。
- 最左边没有元素，所以左侧乘积初始化为 `1`。

第二遍从右向左：

- 用变量 `right_product` 保存当前位置右侧所有元素的乘积。
- 把 `answer[i]` 乘上 `right_product`，就得到最终结果。

复杂度：时间 `O(n)`；除输出数组外，额外空间 `O(1)`。

```python
nums = [1, 2, 3, 4]
n = len(nums)
answer = [1] * n
left_product = 1

print('第一遍：记录左侧乘积')
for i in range(n):
    answer[i] = left_product
    left_product *= nums[i]
    print(f'i={i}, answer={answer}, 下一个左侧乘积={left_product}')

print('\n第二遍：乘上右侧乘积')
right_product = 1
for i in range(n - 1, -1, -1):
    answer[i] *= right_product
    right_product *= nums[i]
    print(f'i={i}, answer={answer}, 下一个右侧乘积={right_product}')
```

# 题目 4：滑动窗口最大值

LeetCode 239. Sliding Window Maximum

这是今天的加练困难题。

## 题目描述（改写版）

给你一个整数数组 `nums` 和窗口大小 `k`。窗口最开始覆盖数组最左边的 `k` 个元素，之后每次向右移动一格。请返回每个窗口中的最大值。

### 输入是什么

- `nums`：非空整数列表。
- `k`：窗口长度，并且 `1 <= k <= len(nums)`。

### 输出是什么

返回一个整数列表。第 `i` 个结果是窗口 `nums[i:i+k]` 中的最大值。

### 关键限制和易漏细节

- 窗口长度始终固定为 `k`。
- 一共有 `len(nums) - k + 1` 个窗口，因此结果长度也应如此。
- 数组中可以有负数和重复值。
- 队列中通常存下标而不是只存数值，才能判断元素是否已经离开窗口。

### 示例 1

输入：`nums = [1, 3, -1, -3, 5, 3, 6, 7]`，`k = 3`

输出：`[3, 3, 5, 5, 6, 7]`

解释：六个窗口依次是 `[1,3,-1]`、`[3,-1,-3]`、`[-1,-3,5]`、`[-3,5,3]`、`[5,3,6]`、`[3,6,7]`。

### 示例 2

输入：`nums = [1]`，`k = 1`

输出：`[1]`

## 解法名称

专业名称：**单调队列 + 双端队列**，英文是 **Monotonic Queue + Deque**。

队列中保存“仍可能成为最大值”的元素下标，并保证对应数值从队首到队尾单调不增。因此队首下标对应的元素始终是当前窗口最大值。

## 暴力思路

对每个长度为 `k` 的窗口调用一次 `max()`。窗口数量约为 `n`，每次找最大值需要 `O(k)`，总时间复杂度是 `O(nk)`。

## 优化思路

遍历每个下标 `right` 时，按顺序做三件事：

1. 如果队首下标已经离开当前窗口，就从左边删除。
2. 当队尾对应的值小于等于当前值时，从右边删除。因为当前元素更靠右且不更小，旧元素以后不可能成为最大值。
3. 把当前下标加入队尾。

当窗口形成，也就是 `right >= k - 1` 时，把 `nums[queue[0]]` 加入答案。

每个下标最多入队一次、出队一次，因此时间复杂度是 `O(n)`，队列空间复杂度是 `O(k)`。

```python
from collections import deque

nums = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
queue = deque()
answer = []

for right, num in enumerate(nums):
    print(f'\n处理下标 {right}，值 {num}')

    while queue and queue[0] <= right - k:
        removed = queue.popleft()
        print(f'  下标 {removed} 已离开窗口，从队首删除')

    while queue and nums[queue[-1]] <= num:
        removed = queue.pop()
        print(f'  {nums[removed]} 不可能再成为最大值，从队尾删除')

    queue.append(right)
    print('  队列下标：', list(queue), '对应数值：', [nums[i] for i in queue])

    if right >= k - 1:
        answer.append(nums[queue[0]])
        print('  当前窗口最大值：', nums[queue[0]])

print('\n最终答案：', answer)
```

## 参考答案：滑动窗口最大值

```python
from collections import deque

def max_sliding_window_answer(nums, k):
    queue = deque()
    answer = []

    for right, num in enumerate(nums):
        while queue and queue[0] <= right - k:
            queue.popleft()

        while queue and nums[queue[-1]] <= num:
            queue.pop()

        queue.append(right)

        if right >= k - 1:
            answer.append(nums[queue[0]])

    return answer

print(max_sliding_window_answer([1, 3, -1, -3, 5, 3, 6, 7], 3))
print(max_sliding_window_answer([1], 1))
print(max_sliding_window_answer([1, -1], 1))
print(max_sliding_window_answer([9, 8, 7, 6], 2))
```

面试表达：

这题我使用单调队列。双端队列中存数组下标，并让下标对应的数值保持单调不增。处理新元素时，先删除已经离开窗口的队首下标，再删除队尾所有小于等于当前值的下标，最后加入当前下标。这样队首始终对应当前窗口最大值。每个下标最多入队和出队一次，所以时间复杂度是 `O(n)`，空间复杂度是 `O(k)`。
