网格题几乎都要用这两样东西，先写熟：

**一、方向数组**：表示上、下、左、右四个方向。

```python
directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]
```

每一项是 `(行的变化, 列的变化)`。遍历它，就能得到四个邻居的坐标：`(row + drow, col + dcol)`。

**二、越界判断**：访问邻居之前必须检查坐标是否还在网格里。

```python
if 0 <= new_row < rows and 0 <= new_col < cols:
```

这两段代码今天每道网格题都会用，和链表题的 `dummy` 一样属于固定套路。


````python
grid = [
    ["1", "1", "0"],
    ["0", "1", "0"],
    ["0", "0", "1"],
]
rows = len(grid)
cols = len(grid[0])
directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]

row = 0
col = 1
print("当前位置：", row, col, "，值是：", grid[row][col])
for drow, dcol in directions:
    new_row = row + drow
    new_col = col + dcol
    if 0 <= new_row < rows and 0 <= new_col < cols:
        print("    邻居", new_row, new_col, "，值是：", grid[new_row][new_col])
    else:
        print("    方向", drow, dcol, "越界了，跳过")
````

## 题目 1 做题前先补：把网格看成一幅图

网格其实就是一个图：

- 每个格子是一个节点；
- 相邻（上下左右）的两个格子如果有边连起来，就是“连通”。

“岛屿”就是一片连通的 `"1"`。数岛屿数量 = 数连通块的个数。

**最经典的数法：淹岛。**

- 从左到右、从上到下扫描网格；
- 遇到一个 `"1"`，说明发现了一座新岛，计数器加一；
- 然后立刻从这个格子出发，用 DFS 把整座岛“淹掉”——把所有相连的 `"1"` 改成 `"0"`。

这样后面再遇到它们，就不会重复计数了。**这个“改成 0”就是访问标记，比额外建一个 visited 数组更省空间。**


````python
# 手工演示一次“淹岛”
grid = [
    ["1", "1", "0"],
    ["0", "1", "0"],
    ["0", "0", "1"],
]


def sink(row, col):
    if row < 0 or row >= len(grid) or col < 0 or col >= len(grid[0]):
        return
    if grid[row][col] != "1":
        return
    grid[row][col] = "0"
    print("    淹掉：", row, col)
    sink(row + 1, col)
    sink(row - 1, col)
    sink(row, col + 1)
    sink(row, col - 1)


print("从 (0,0) 开始淹：")
sink(0, 0)
print("淹完之后网格变成：")
for line in grid:
    print("   ", line)
````

## 题目 2 做题前先补：多源 BFS 和“扩散圈数”

这道题和岛屿数量的最大区别是：**腐烂是会向外扩散的，而且每次扩散一圈（一分钟）**。

这类题的标准做法叫**多源 BFS**：一开始把所有腐烂的橘子都放进队列当起点，而不是只放一个。

为什么要多源？因为所有腐烂的橘子是**同时**向外传染的，它们是同一个“波”的起点。

层数怎么统计？用固定层大小的写法：

```python
while queue and fresh > 0:
    size = len(queue)          # 这一圈有多少个腐烂橘子
    for i in range(size):      # 只处理这一圈
        ...                    # 把邻居变腐烂并加入队列，它们属于下一圈
    minutes = minutes + 1      # 一圈处理完，时间加一
```

注意 `minutes` 是在 `for` 循环**外面**加一的，因为它表示“扩散了多少圈”。

`fresh > 0` 这个条件加在 `while` 上是小优化：新鲜橘子都烂完了就提前结束，不用再跑最后一圈。


````python
from collections import deque

grid = [
    [2, 1, 1],
    [1, 1, 0],
    [0, 1, 1],
]
rows = len(grid)
cols = len(grid[0])
directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]

queue = deque()
fresh = 0
for i in range(rows):
    for j in range(cols):
        if grid[i][j] == 2:
            queue.append((i, j))
        elif grid[i][j] == 1:
            fresh = fresh + 1
print("初始腐烂的橘子：", list(queue), "，新鲜橘子数：", fresh)

minutes = 0
while queue and fresh > 0:
    size = len(queue)
    print("第", minutes + 1, "分钟，这一圈有", size, "个腐烂橘子")
    for k in range(size):
        row, col = queue.popleft()
        for drow, dcol in directions:
            new_row = row + drow
            new_col = col + dcol
            if 0 <= new_row < rows and 0 <= new_col < cols and grid[new_row][new_col] == 1:
                grid[new_row][new_col] = 2
                fresh = fresh - 1
                queue.append((new_row, new_col))
    minutes = minutes + 1

print("总分钟数：", minutes, "，剩余新鲜橘子：", fresh)
````

给你一个网格 `grid`，每个格子是 0、1 或 2：

- `0` 表示空格子；

- `2` 表示一个腐烂的橘子。

## 输入

- `grid`：1 到 10 行，1 到 10 列，只含 0、1、2。

## 输出

返回分钟数；如果永远不可能全部腐烂，返回 -1；一开始就没有新鲜橘子，返回 0。

## 示例

示例 1：`[[2,1,1],[1,1,0],[0,1,1]]`，答案是 4。

示例 2：`[[2,1,1],[0,1,1],[1,0,1]]`，答案是 -1（左下角那个橘子永远接触不到）。

示例 3：`[[0,2]]`，答案是 0（没有新鲜橘子）。

## 易漏细节

- 一开始没有新鲜橘子时，答案必须是 0，不是 1。
- 所有腐烂橘子是同时扩散的，所以要把它们全部入队（多源 BFS）。
- 结束时如果还有新鲜橘子，返回 -1。


## 解法名称

**多源广度优先搜索（Multi-Source BFS）**。

## 暴力思路

模拟每一分钟：扫描整个网格，找出所有会被感染的新鲜橘子再统一感染。每次扫描 O(行 × 列)，最坏分钟数也是 O(行 × 列)，总时间 O((行 × 列)²)，太慢。

## 优化思路

多源 BFS：

1. 先扫一遍网格：把所有腐烂橘子的坐标放进队列，同时数出新鲜橘子的个数 `fresh`。
2. 只要队列不空、且还有新鲜橘子，就处理当前这一圈：
   - `size = len(queue)` 固定这一圈的个数；
   - 逐个弹出，检查四个邻居，如果是新鲜橘子就变腐烂、`fresh` 减一、加入队列；
3. 每处理完一圈，分钟数加一。
4. 循环结束后，如果 `fresh` 还大于 0，返回 -1；否则返回分钟数。

时间 O(行 × 列)，每个格子最多入队一次；空间 O(行 × 列)。

## 题目 3 做题前先补：什么是拓扑排序

这题问的是“能不能把所有课都修完”，本质是问：**这些先修关系里有没有循环依赖**？

比如 A 要先修 B，B 要先修 A，那就永远修不完。

把课程看成节点、先修关系看成有向边（B -> A 表示“修完 B 才能修 A”），问题就变成：**这个有向图里有没有环**。

判断有没有环的标准方法叫**拓扑排序**，最常用的写法是“BFS + 入度”：

- **入度**：一个节点有几条边指向它。在课程表里，就是“这门课还有几门先修课没修”。
- 先把所有入度为 0 的课程放进队列（这些课可以直接修）。
- 每次从队列取出一个节点，算它修完了；把它的所有后继节点的入度减一。
- 如果某个后继节点的入度减到 0，说明它的先修课都修完了，可以放进队列。

最后统计修完的课程数：**等于总数说明没有环，小于总数说明剩下的课互相锁死（有环）。**


````python
from collections import deque

# 手工模拟一次：4 门课，先修关系 [1,0] 表示修 1 之前要先修 0
num_courses = 4
prerequisites = [[1, 0], [2, 1], [3, 2]]

graph = {}
for i in range(num_courses):
    graph[i] = []
indegree = [0] * num_courses

for pair in prerequisites:
    course = pair[0]
    pre = pair[1]
    graph[pre].append(course)
    indegree[course] = indegree[course] + 1

print("每门课的入度：", indegree)
print("先修关系表：", graph)

queue = deque()
for i in range(num_courses):
    if indegree[i] == 0:
        queue.append(i)
print("可以直接修的课：", list(queue))

order = []
while queue:
    node = queue.popleft()
    order.append(node)
    for nxt in graph[node]:
        indegree[nxt] = indegree[nxt] - 1
        if indegree[nxt] == 0:
            queue.append(nxt)

print("修课顺序：", order)
print("能不能全部修完：", len(order) == num_courses)
````

## 题目 4 做题前先补：前缀树是什么

**前缀树（Trie）**是一种专门用来存字符串的树，每个节点代表一个字符，从根到某个节点的路径拼起来就是一个前缀。

比如依次插入 `"apple"` 和 `"app"`：

```text
root
 └─ a ─ p ─ p ─ l ─ e     （这里的 p 节点标记为“某个单词结尾”）
```

插入 `"app"` 时，前面 `a -> p -> p` 已经在树上了，只需要在最后那个 `p` 上标记“单词结尾”。

每个节点需要两样东西：

- `children`：字典，`字符 -> 子节点`；
- `is_end`：布尔值，标记“从根走到这里是不是一个完整的单词”。

用 `children` 字典而不是固定 26 个字母的数组，代码更短，也更通用。


````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False


root = TrieNode()
# 手工插入 "app"
node = root
for ch in "app":
    if ch not in node.children:
        node.children[ch] = TrieNode()
    node = node.children[ch]
node.is_end = True

print("根节点的孩子：", list(root.children.keys()))
print("走 a -> p -> p 之后，is_end =", node.is_end)
print("这条路走 a -> p 就停，is_end =", root.children["a"].children["p"].is_end)
````

面试表达：

我用前缀树。每个节点有一个字典存放子节点，还有一个布尔标记表示“到这里是不是一个完整单词的结尾”。插入时从根开始，遇到没有的字符就建新节点，最后把结尾标记设为真。查询时同样从根往下走，中途字符不存在就直接返回假；如果是查完整单词，走完后要检查结尾标记；如果只是查前缀，走完就可以返回真。插入和查询的时间都是 O(单词长度)。
