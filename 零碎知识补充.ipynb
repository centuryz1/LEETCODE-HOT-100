# 模块 A：复杂度基础

## A01. 什么是时间复杂度

时间复杂度用来粗略描述：当输入数据变多时，程序运行时间会怎么增长。

它不关心程序具体运行了 0.01 秒还是 1 秒，而是关心增长趋势。

常见写法：

- `O(1)`
- `O(n)`
- `O(n^2)`
- `O(log n)`
- `O(n log n)`

这里的 `n` 通常表示输入规模，比如数组长度、字符串长度、节点数量。

## A02. O(1)：常数时间

`O(1)` 表示：操作次数基本不随输入规模变化。

换句话说，数据有 10 个、1000 个、100000 个，这个操作的耗时大致是同一个级别。

注意：`O(1)` 不是说只执行 1 次，也不是说一定特别快。它强调的是：和 `n` 没有明显关系。

### O(1) 例子：按下标访问列表元素

如果知道元素的位置，就可以直接取出来。

````python
nums = [10, 20, 30, 40, 50]

print(nums[0])
print(nums[3])
````

### O(1) 例子：字典按 key 查询

Python 的 `dict` 查询，刷题时通常按平均 `O(1)` 理解。

````python
age = {
    'Alice': 18,
    'Bob': 20,
    'Cindy': 22
}

print(age['Bob'])
print('Alice' in age)
````

## A03. O(n)：线性时间

`O(n)` 表示：输入规模变大，操作次数大致也按比例变大。

如果数组长度是 `n`，程序大概要看一遍数组，这通常就是 `O(n)`。

````python
nums = [10, 20, 30, 40, 50]

for x in nums:
    print(x)
````

## A04. O(n^2)：平方时间

`O(n^2)` 常见于两层循环。

如果外层循环跑 `n` 次，内层循环也大约跑 `n` 次，总操作次数就接近 `n * n`。

````python
nums = [1, 2, 3, 4]

for i in range(len(nums)):
    for j in range(len(nums)):
        print(nums[i], nums[j])
````

## A05. 复杂度速记表

| 复杂度 | 粗略理解 | 常见场景 |
|---|---|---|
| O(1) | 直接拿、直接查 | 列表下标访问、字典查询 |
| O(n) | 看一遍 | 一层循环遍历数组 |
| O(n^2) | 两两配对 | 两层循环枚举所有组合 |
| O(log n) | 每次砍掉一半 | 二分查找 |
| O(n log n) | 每层处理 n 个，层数 log n | 常见排序算法 |

## B03-1. dict 常用操作

`dict` 是刷题里非常重要的工具，常用来做：

- 快速查询某个值是否出现过
- 记录某个值出现的位置
- 统计数量
- 建立映射关系

刷题时看到「快速判断是否存在」「记录出现次数」「根据某个 key 分组」，通常都可以先想到字典。

### 创建字典

字典的基本格式是 `{key: value}`。

- key：用来查询的东西
- value：key 对应的值

````python
student = {
    'name': 'Alice',
    'age': 18,
    'score': 95
}

print(student)
````

### 查询：用 key 找 value

如果 key 一定存在，可以用 `dict[key]`。

````python
student = {'name': 'Alice', 'age': 18, 'score': 95}

print(student['name'])
print(student['score'])
````

如果 key 不一定存在，建议先用 `in` 判断，或者用 `get`。

````python
student = {'name': 'Alice', 'age': 18}

print('score' in student)
print(student.get('score'))      # key 不存在，返回 None
print(student.get('score', 0))   # key 不存在，返回默认值 0
````

### 新增和修改

同一个写法既可以新增，也可以修改。

````python
student = {'name': 'Alice', 'age': 18}

student['score'] = 95   # 新增
student['age'] = 19     # 修改

print(student)
````

### 删除

常见删除方式是 `del` 和 `pop`。

````python
student = {'name': 'Alice', 'age': 18, 'score': 95}

del student['score']
print(student)

age = student.pop('age')
print(age)
print(student)
````

### 遍历字典

刷题里最常用的是 `.items()`，可以同时拿到 key 和 value。

````python
count = {'a': 2, 'b': 3, 'c': 1}

for key in count:
    print('key:', key)

for key, value in count.items():
    print('key:', key, 'value:', value)
````

### 计数写法

如果不用 `defaultdict`，可以用 `get` 来计数。

````python
nums = [1, 2, 1, 3, 2, 1]
count = {}

for x in nums:
    count[x] = count.get(x, 0) + 1

print(count)
````

### 常见误区

1. `dict[key]` 在 key 不存在时会报错。
2. `key in dict` 判断的是 key 是否存在，不是 value 是否存在。
3. 字典的 key 必须是可哈希的，常见可用 key 有数字、字符串、元组；列表不能直接当 key。
4. 刷题时 Python 字典查询通常按平均 `O(1)` 理解。

````python
d = {'a': 1, 'b': 2}

print('a' in d)  # True，因为 a 是 key
print(1 in d)    # False，因为 1 是 value，不是 key
````

## B03-2. defaultdict 常用操作

`defaultdict` 来自 `collections` 模块。

它和普通字典很像，区别是：当访问一个不存在的 key 时，它会自动创建一个默认值。

刷题中最常用的两种：

- `defaultdict(int)`：适合计数
- `defaultdict(list)`：适合分组

### defaultdict(int)：计数

普通字典计数时，需要先判断 key 是否存在。`defaultdict(int)` 会把不存在的 key 默认当成 `0`。

````python
from collections import defaultdict

nums = [1, 2, 1, 3, 2, 1]
count = defaultdict(int)

for x in nums:
    count[x] += 1

print(count)
print(dict(count))  # 转成普通 dict，显示更清楚
````

### defaultdict(list)：分组

`defaultdict(list)` 会把不存在的 key 默认对应到一个空列表，所以可以直接 `.append()`。

````python
from collections import defaultdict

words = ['apple', 'ant', 'banana', 'book']
groups = defaultdict(list)

for word in words:
    first = word[0]
    groups[first].append(word)

print(groups)
print(list(groups.values()))
````

### defaultdict 和普通 dict 的区别

| 场景 | 普通 dict | defaultdict |
|---|---|---|
| key 不存在时读取 | 可能报错 | 自动给默认值 |
| 计数 | `d[x] = d.get(x, 0) + 1` | `d[x] += 1` |
| 分组 | 需要先创建空列表 | 可以直接 `.append()` |

如果你只是普通查询，用 `dict` 就够了。  
如果你要不断计数或分组，`defaultdict` 会更顺手。

### defaultdict 常见误区

1. `defaultdict` 不是新数据结构，本质上还是字典。
2. `defaultdict(int)` 的默认值是 `0`。
3. `defaultdict(list)` 的默认值是空列表 `[]`。
4. 如果要把结果返回成普通列表，常见写法是 `list(groups.values())`。
5. 如果要显示成普通字典，可以用 `dict(d)`。

## B03-3. Counter 计数器

`Counter` 来自 Python 标准库的 `collections` 模块，专业名称是**计数器（Counter）**。

它专门用来统计“每个元素出现了多少次”。

```python
from collections import Counter

counts = Counter(["a", "b", "a", "c", "a"])
```

结果类似普通字典：

```python
Counter({"a": 3, "b": 1, "c": 1})
```

可以先把它理解成：**已经帮你写好计数循环的字典**。

### 为什么需要 Counter？

普通字典计数：

```python
counts = {}
for value in nums:
    counts[value] = counts.get(value, 0) + 1
```

使用 `defaultdict(int)`：

```python
counts = defaultdict(int)
for value in nums:
    counts[value] += 1
```

使用 `Counter`：

```python
counts = Counter(nums)
```

三种方式都正确。只需要统计频次时，`Counter` 最简洁；需要在循环中同时保存其他信息时，普通字典通常更灵活。

````python
from collections import Counter

nums = [1, 2, 1, 3, 2, 1]
counts = Counter(nums)

print(counts)
print(dict(counts))

assert counts[1] == 3
assert counts[2] == 2
assert counts[3] == 1
````

### 创建 Counter 的三种常见方式

#### 1. 从列表、字符串等可迭代对象创建

```python
Counter([1, 1, 2])
Counter("banana")
```

#### 2. 从普通字典创建

```python
Counter({"a": 3, "b": 2})
```

#### 3. 创建空计数器后逐步更新

```python
counts = Counter()
counts["a"] += 1
```

字符串也是字符序列，所以 `Counter("banana")` 会统计每个字符的数量。

````python
from collections import Counter

char_counts = Counter("banana")
dict_counts = Counter({"apple": 2, "banana": 1})
empty_counts = Counter()
empty_counts["python"] += 1

print(char_counts)
print(dict_counts)
print(empty_counts)

assert char_counts["a"] == 3
assert char_counts["n"] == 2
````

### 读取次数：不存在的元素返回 0

```python
counts = Counter("apple")

counts["p"]  # 2
counts["x"]  # 0
```

普通字典使用 `d["x"]` 时，键不存在会触发 `KeyError`；`Counter` 对不存在的元素返回 `0`。

仍然可以判断键是否真实存在：

```python
"x" in counts  # False
```

因此要区分：

- `counts[x] == 0`：当前计数为 0；
- `x in counts`：这个键是否实际保存在计数器中。

### `most_common()`：寻找出现次数最多的元素

```python
counts.most_common()
```

返回按次数从高到低排列的 `(元素, 次数)` 列表。

```python
counts.most_common(2)
```

只返回出现次数最多的前两个元素。

这在“出现频率最高的 K 个元素”一类问题中非常常见。

注意：返回值是二元组组成的列表，不是字典。

````python
from collections import Counter

counts = Counter("mississippi")
top_two = counts.most_common(2)

print(counts)
print(top_two)

assert top_two[0] == ("i", 4)
assert len(top_two) == 2
````

### `update()` 与 `subtract()`

#### 增加计数

```python
counts.update(["a", "a", "b"])
```

`update()` 会把新数据的次数累加进去，不是直接覆盖原计数。

#### 减少计数

```python
counts.subtract(["a", "b"])
```

`subtract()` 会做减法，并且可能保留计数为 `0` 或负数的键。

如果只想手动调整一个元素，也可以：

```python
counts["a"] += 1
counts["b"] -= 1
```

````python
from collections import Counter

counts = Counter({"a": 2, "b": 1})
counts.update(["a", "c"])
print("update 后：", counts)

counts.subtract(["a", "b", "b"])
print("subtract 后：", counts)

assert counts["a"] == 2
assert counts["b"] == -1
assert counts["c"] == 1
````

### Counter、dict、defaultdict 怎么选？

| 需求 | 推荐工具 |
|---|---|
| 只统计每个元素出现次数 | `Counter` |
| 一边遍历一边简单计数 | 三者都可以 |
| 按键把元素放进多个列表 | `defaultdict(list)` |
| 除了次数还要保存下标、状态等信息 | 普通 `dict` |
| 需要获取频次最高的若干元素 | `Counter.most_common(k)` |

不要为了显得专业而强行使用 `Counter`。选择能让代码最清楚的工具才是目标。

### 刷题例子：判断两个字符串是否为字母异位词

字母异位词要求两个字符串包含完全相同的字符及次数，只是排列顺序不同。

```python
from collections import Counter

def is_anagram(s, t):
    return Counter(s) == Counter(t)
```

专业解法名称：**频次统计（Frequency Counting）**。

- 创建两个计数器都需要遍历字符串；
- 设字符串总长度为 `n`，时间复杂度通常为 `O(n)`；
- 计数器保存字符频次，空间复杂度为 `O(k)`，`k` 是不同字符数量。

````python
from collections import Counter

def is_anagram(s, t):
    return Counter(s) == Counter(t)


assert is_anagram("anagram", "nagaram") is True
assert is_anagram("rat", "car") is False
assert is_anagram("aacc", "ccac") is False
print("字母异位词测试通过")
````

### Counter 常见误区

1. `Counter` 不是自动排序的列表，本质上是字典的子类。
2. `most_common(k)` 返回 `(元素, 次数)` 二元组组成的列表。
3. 访问不存在的元素返回 `0`，但不会因为读取就一定把键加入计数器。
4. `update()` 是累加，不是覆盖。
5. `subtract()` 可能留下 `0` 和负数计数。
6. `Counter(nums)` 需要 `nums` 中的元素可以作为字典键，例如整数、字符串、元组；列表本身不能作为键。
7. 只需要判断元素是否出现过时，使用 `set` 通常比 `Counter` 更直接。

## B06. if、elif 和 else

它们都用于条件分支，但组合方式不同。

### 单独的 if

`if` 表示：如果这个条件成立，就执行一次对应代码。

```python
if 条件:
    条件成立时执行
```

多个彼此独立的 `if` 会分别判断，因此可能有多个代码块同时执行。

### if / elif / else

它们组成一条完整的分支链：

```python
if 条件1:
    条件1成立时执行
elif 条件2:
    条件1不成立、条件2成立时执行
else:
    前面条件都不成立时执行
```

Python 从上到下判断，一旦找到第一个成立的分支并执行，就会跳过后面所有 `elif` 和 `else`。

一句话记忆：

- 多个独立 `if`：每个都要判断，可能执行多个。
- `if / elif / else`：只选择一条路，最多执行一个分支。

````python
score = 95

print('两个独立 if：')
if score >= 60:
    print('及格')
if score >= 90:
    print('优秀')

# 两个条件都成立，所以会输出两句话。
````

````python
score = 95

print('if / elif / else 分支链：')
if score >= 90:
    print('优秀')
elif score >= 60:
    print('及格')
else:
    print('不及格')

# 第一个条件已经成立，后面的 elif 不再判断，只输出“优秀”。
````

### 什么时候使用独立 if？

当多个条件可以同时成立，而且每个成立条件都需要处理时，使用多个独立 `if`。

例如，一个数字既可能是正数，也可能是偶数，这两个特征可以同时存在。

### 什么时候使用 elif？

当多个结果互相排斥、最终只能选择一种时，使用 `if / elif / else`。

例如，成绩最终只能归入“优秀、及格、不及格”中的一类。

### 条件顺序很重要

`elif` 链会在第一个成立条件处停止，因此一般把更具体、更严格的条件放在前面。

错误顺序：

```python
if score >= 60:
    print('及格')
elif score >= 90:
    print('优秀')
```

当 `score = 95` 时，第一个条件已经成立，永远到不了“优秀”。

常见误区：

1. `elif` 不能单独出现，前面必须有 `if`。
2. `else` 没有条件，表示前面都不成立。
3. 不要为了少写几个 `if` 就强行使用 `elif`，先判断这些条件能否同时成立。

````python
def classify_score(score):
    if score >= 90:
        return '优秀'
    elif score >= 60:
        return '及格'
    else:
        return '不及格'


assert classify_score(95) == '优秀'
assert classify_score(70) == '及格'
assert classify_score(40) == '不及格'
print('if / elif / else 示例测试通过')
````

## B07. continue 的用法

`continue` 只能写在 `for` 或 `while` 循环内部。

它的作用是：

> 立即结束当前这一轮循环，跳过本轮剩余代码，直接进入下一轮。

注意，它不是结束整个循环。

```python
for 元素 in 数据:
    if 当前元素不需要处理:
        continue

    # 只有没有执行 continue 时，才会运行这里
```

常见用途：

- 跳过不符合条件的数据。
- 跳过重复元素。
- 跳过无效输入。
- 减少代码缩进，让主要逻辑更清楚。

````python
nums = [3, -1, 5, -2, 8]

for num in nums:
    if num < 0:
        print('跳过负数：', num)
        continue

    print('处理非负数：', num)

# 遇到负数时，continue 后面的“处理非负数”不会执行。
````

### continue 和 if / else 的关系

下面两种写法效果相近：

写法 1：使用 `continue`

```python
for num in nums:
    if num < 0:
        continue
    print(num)
```

写法 2：使用 `if`

```python
for num in nums:
    if num >= 0:
        print(num)
```

当主要处理逻辑很长时，`continue` 可以先排除不需要的数据，减少整体缩进。

### continue、break、pass 的区别

| 关键字 | 作用 |
|---|---|
| `continue` | 跳过当前这一轮，进入下一轮 |
| `break` | 直接结束整个循环 |
| `pass` | 什么都不做，只是占位，然后继续执行后面的代码 |

一句话记忆：

- `continue`：这一轮不做了。
- `break`：整个循环不做了。
- `pass`：这里暂时空着，程序照常往下走。

````python
print('continue：跳过 3，但循环继续')
for num in [1, 2, 3, 4, 5]:
    if num == 3:
        continue
    print(num)

print('\nbreak：遇到 3，整个循环结束')
for num in [1, 2, 3, 4, 5]:
    if num == 3:
        break
    print(num)
````

### while 中使用 continue 要特别小心

`while` 依赖循环变量逐渐改变，最终让条件变成 False。

如果在更新变量之前执行 `continue`，可能永远停在同一个值，形成死循环。

危险写法：

```python
number = 0
while number < 5:
    if number == 2:
        continue       # number 一直是 2，永远无法结束
    number += 1
```

安全思路：先更新会影响循环条件的变量，再决定是否 `continue`。

````python
number = 0

while number < 5:
    number += 1

    if number == 3:
        print('跳过 3')
        continue

    print('处理：', number)

# number 在 continue 之前已经增加，所以循环能够正常结束。
````

### 小练习

观察下面函数：它需要计算列表中所有正数的和，忽略 0 和负数。

你可以使用 `continue` 先跳过不需要的数字。

````python
def sum_positive(nums):
    total = 0

    for num in nums:
        if num <= 0:
            continue
        total += num

    return total


assert sum_positive([3, -1, 0, 5, -2]) == 8
assert sum_positive([-1, -2]) == 0
print('continue 练习测试通过')
````

## D01. 双指针

双指针不是一种具体的数据结构，而是一种写代码的思路。

你可以先这样理解：

**用两个变量，分别表示两个位置，然后让这两个位置按规则移动。**

这两个变量通常叫：

- `left` 和 `right`
- `slow` 和 `fast`
- `i` 和 `j`

它们不是特殊语法，只是普通变量名。

### 为什么需要双指针

有些题如果用两层循环，会很慢。

双指针的目标通常是：

- 少走重复路
- 一边遍历，一边维护答案
- 把 O(n^2) 的枚举，尽量优化成 O(n)

可以粗略理解成：两个位置配合干活，而不是所有组合都试一遍。

### 类型 1：左右指针

左右指针通常从数组两端开始：

- `left = 0`
- `right = len(nums) - 1`

然后根据条件移动左边或右边。

````python
nums = [1, 2, 3, 4, 5]

left = 0
right = len(nums) - 1

while left < right:
    print('left 指向:', nums[left], 'right 指向:', nums[right])
    left += 1
    right -= 1
````

左右指针适合：

- 从两边往中间找
- 反转数组或字符串
- 有序数组中找两个数
- 盛最多水的容器这类左右边界问题

### 类型 2：快慢指针

快慢指针通常都从左边开始：

- `fast` 负责往前扫描
- `slow` 负责记录某个重要位置

它常用于原地修改数组。

````python
nums = [0, 1, 0, 3]

slow = 0

for fast in range(len(nums)):
    print('fast =', fast, 'slow =', slow, 'nums =', nums)
    if nums[fast] != 0:
        nums[slow], nums[fast] = nums[fast], nums[slow]
        slow += 1

print('结果:', nums)
````

快慢指针适合：

- 移动零
- 删除数组重复项
- 原地保留某些元素
- 链表中找中点或判断环，后面会学

### 类型 3：同向双指针

两个指针都往同一个方向走，也属于双指针。

比如 `left` 和 `right` 都从左边开始，`right` 负责扩展范围，`left` 负责缩小范围。

这类思想后面会发展成「滑动窗口」。

````python
nums = [1, 2, 3, 4]

left = 0

for right in range(len(nums)):
    print('当前范围:', left, '到', right, '里面是', nums[left:right + 1])
````

### 刷题时怎么识别双指针

看到这些信号，可以先想双指针：

1. 题目和数组、字符串、链表有关。
2. 要求原地修改。
3. 要找两个位置、两个数、左右边界。
4. 暴力法需要两层循环，但感觉有重复计算。
5. 题目要求保持相对顺序。

不是所有题都能用双指针，但这些信号很常见。

### 常见误区

1. 双指针不是一定有两个真实指针，只是两个变量记录位置。
2. `left`、`right`、`slow`、`fast` 都只是变量名，可以换名字。
3. 最重要的不是创建两个变量，而是想清楚每个变量代表什么。
4. 每次循环后，至少要有一个指针移动，否则可能死循环。
5. 写之前先问：这两个指针分别负责什么？什么时候移动？

### 一句话记忆

双指针就是：**用两个位置变量一起扫描数据，并根据规则移动它们。**

左右指针：从两边往中间走。  
快慢指针：一个负责扫描，一个负责记录位置。  
同向双指针：两个都往右走，常用于范围类问题。

## C01. 队列与双端队列 deque

队列通常遵循先进先出（FIFO）。Python 的 `collections.deque` 是双端队列，两端都能高效操作：`append` 和 `appendleft` 用于加入，`pop` 和 `popleft` 用于删除，`queue[0]` 与 `queue[-1]` 用于查看两端元素。

当算法频繁从左端删除元素时，优先使用 `deque.popleft()`，不要使用较慢的 `list.pop(0)`。

````python
from collections import deque

queue = deque()
queue.append(10)
queue.append(20)
queue.appendleft(5)
print(queue)           # deque([5, 10, 20])
print(queue.popleft()) # 5
print(queue.pop())     # 20
````

## D02. 滑动窗口

滑动窗口用于处理数组或字符串中的连续范围，通常用 `left` 和 `right` 表示边界。右边界负责扩大范围；窗口不满足条件时，再移动左边界缩小范围。

常见类型有固定长度窗口和可变长度窗口。识别信号包括“连续子数组”“连续子串”“长度为 k 的范围”等。

常见误区：如果数组含负数，窗口和不一定随着边界移动而单调变化，不能看到连续范围就机械套用滑动窗口。

````python
nums = [10, 20, 30, 40]
k = 2

for right in range(k - 1, len(nums)):
    left = right - k + 1
    print('固定窗口：', nums[left:right + 1])
````

## D03. 前缀和与前缀积

前缀表示从开头到某个位置之前的累计结果。常用定义中，`prefix[i]` 表示前 `i` 个元素之和：`prefix[0] = 0`，`prefix[i + 1] = prefix[i] + nums[i]`。

区间 `[left, right]` 的和可以写成 `prefix[right + 1] - prefix[left]`。前缀积思想相同，只是把加法换成乘法；右侧累计结果则可以从右向左维护。

常见误区：一定要先说明 `prefix[i]` 是否包含下标 `i`，公式必须和定义一致。

````python
nums = [2, 4, 1, 3]
prefix = [0]

for num in nums:
    prefix.append(prefix[-1] + num)

print(prefix)  # [0, 2, 6, 7, 10]
left, right = 1, 3
print(prefix[right + 1] - prefix[left])  # 4 + 1 + 3 = 8
````

## D04. 动态规划入门

动态规划，英文是 Dynamic Programming（DP）。核心是利用已经算出的较小问题结果推导当前结果。

做题时先回答四个问题：状态变量表示什么；状态如何转移；初始状态是什么；最终返回哪个状态。如果当前状态只依赖前一个或前两个状态，可以只保留几个变量。

常见误区：只背状态转移公式，却说不清变量含义。写代码前先用一句完整的话定义状态。

````python
# 示例：当前结果只依赖前两个结果。
def fibonacci(n):
    if n < 2:
        return n

    previous, current = 0, 1
    for _ in range(2, n + 1):
        previous, current = current, previous + current
    return current

print(fibonacci(6))  # 8
````

## D05. 单调队列

单调队列会让候选元素保持单调顺序。维护窗口最大值时，通常让对应数值从队首到队尾单调不增，因此队首就是当前最大值。

加入新元素前，从队尾删除所有小于等于新元素的旧元素。处理滑动窗口时通常保存下标，因为下标既能找到数值，也能判断元素是否已经离开窗口。

单调队列不是把全部元素排序，它保持原来的先后关系，只删除已经不可能成为答案的候选元素。

````python
from collections import deque

nums = [2, 1, 4, 3]
queue = deque()

for index, num in enumerate(nums):
    while queue and nums[queue[-1]] <= num:
        queue.pop()
    queue.append(index)
    print('下标：', list(queue), '数值：', [nums[i] for i in queue])
````

## C02. 二维列表与矩阵

Python 中的矩阵通常使用“列表里面放列表”表示：

```python
matrix = [[1, 2, 3],
          [4, 5, 6]]
```

- `len(matrix)` 是行数。
- `len(matrix[0])` 是列数。
- `matrix[row][col]` 先找第 `row` 行，再找这一行的第 `col` 个元素。
- `row` 和 `col` 都从 0 开始。

遍历矩阵通常使用两层 `for`：外层遍历行，内层遍历列。

常见误区：把行号和列号写反；只适用于正方形矩阵；修改一行时误用共享列表，例如 `[[0] * cols] * rows` 会让多行指向同一个列表。

````python
matrix = [[1, 2, 3],
          [4, 5, 6]]

rows = len(matrix)
cols = len(matrix[0])

for row in range(rows):
    for col in range(cols):
        print(f'位置 ({row}, {col}) 的值是 {matrix[row][col]}')
````

## D06. 排序后线性扫描

有些问题原本元素关系很乱，排序后，相近或相关的元素会靠在一起。这样常常只需要从左向右扫描一次，不必让每个元素和所有其他元素比较。

常见步骤：

1. 选择排序关键字，例如按第一个元素排序。
2. 维护已经处理好的结果或当前状态。
3. 读取下一个元素，只和最近状态比较。
4. 根据条件更新或新建结果。

Python 可以使用 `items.sort(key=lambda item: item[0])` 按子列表第一个元素排序。

识别信号：题目涉及区间、时间段、相邻关系，或者排序后能够让冲突集中出现。

````python
items = [[5, 8], [1, 3], [2, 6]]
items.sort(key=lambda item: item[0])
print(items)  # [[1, 3], [2, 6], [5, 8]]
````

## D07. 模拟与边界

模拟算法就是按照题目规定的过程一步步执行。它通常不依赖复杂公式，但非常依赖清楚的变量含义和边界检查。

写模拟题前先明确：

1. 当前状态由哪些变量表示？
2. 每一步按照什么顺序执行？
3. 什么时候改变方向或收缩范围？
4. 循环在什么条件下停止？
5. 单行、单列、空范围等边界会不会重复处理？

矩阵边界常用 `top`、`bottom`、`left`、`right` 表示。只要范围仍然有效，就用 `while` 继续；某条边是否仍存在，通常用 `if` 判断一次。

常见误区：更新边界的时机不对；使用更新后的边界访问旧范围；没有在访问后半部分前重新检查边界。

````python
top, bottom = 0, 2
left, right = 0, 3

while top <= bottom and left <= right:
    print('当前有效边界：', top, bottom, left, right)
    top += 1
    bottom -= 1
    left += 1
    right -= 1
````
