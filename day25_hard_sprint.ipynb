## 题目 1 做题前先补：三种操作对应表格的三个方向

编辑距离：把 `word1` 变成 `word2`，允许插入、删除、替换一个字符，问最少需要几步。

`dp[i][j]` 的定义：**把 `word1` 的前 i 个字符变成 `word2` 的前 j 个字符，最少需要几步。**

转移分两种情况：

**情况一：两个当前字符相同**（`word1[i-1] == word2[j-1]`）。

那这一对字符不用管，直接继承左上角：`dp[i][j] = dp[i-1][j-1]`。

**情况二：不同**。三种操作对应表格的三个方向：

| 操作 | 含义 | 从哪来 |
| --- | --- | --- |
| 删除 | 删掉 `word1` 的第 i 个字符 | 上面 `dp[i-1][j]` |
| 插入 | 在 `word1` 里补一个 `word2[j-1]` | 左边 `dp[i][j-1]` |
| 替换 | 把 `word1[i-1]` 改成 `word2[j-1]` | 左上 `dp[i-1][j-1]` |

取三者最小再加一：`dp[i][j] = min(上面, 左边, 左上) + 1`。

边界：`dp[i][0] = i`（把 i 个字符全删掉），`dp[0][j] = j`（插入 j 个字符）。

**注意：这题和最长公共子序列是同一张表格，只是转移和边界不同。**


````python
word1 = "horse"
word2 = "ros"
m = len(word1)
n = len(word2)
dp = []
for i in range(m + 1):
    dp.append([0] * (n + 1))

for i in range(m + 1):
    dp[i][0] = i
for j in range(n + 1):
    dp[0][j] = j

for i in range(1, m + 1):
    for j in range(1, n + 1):
        if word1[i - 1] == word2[j - 1]:
            dp[i][j] = dp[i - 1][j - 1]
        else:
            best = dp[i - 1][j - 1]
            if dp[i - 1][j] < best:
                best = dp[i - 1][j]
            if dp[i][j - 1] < best:
                best = dp[i][j - 1]
            dp[i][j] = best + 1

print("编辑距离表格：")
for line in dp:
    print("   ", line)
print("答案：", dp[m][n])
````

## 题目 2 做题前先补：中位数怎么定义

中位数是“排好序以后正中间的那个数”：

- 总个数是奇数：正中间那个；
- 总个数是偶数：中间两个数的平均值。

本题两个数组都已经是升序的了，所以最直接的想法是**把它们合并成一个有序数组**，再取中间。

合并的方法是 Day 06 学过的“双指针归并”（合并两个有序链表的思想，换成数组）：

```python
while i < len(nums1) and j < len(nums2):
    if nums1[i] < nums2[j]:
        取 nums1[i]，i 前进
    else:
        取 nums2[j]，j 前进
```

然后把剩下的接完，最后按下标取中位数。

这个版本时间是 O(m+n)，题目要求 O(log(m+n))，所以面试时先说这个版本，再提二分优化。


````python
# 手工合并一次
nums1 = [1, 3]
nums2 = [2]
merged = []
i = 0
j = 0

while i < len(nums1) and j < len(nums2):
    if nums1[i] < nums2[j]:
        merged.append(nums1[i])
        i = i + 1
    else:
        merged.append(nums2[j])
        j = j + 1
while i < len(nums1):
    merged.append(nums1[i])
    i = i + 1
while j < len(nums2):
    merged.append(nums2[j])
    j = j + 1

print("合并结果：", merged)
total = len(merged)
if total % 2 == 1:
    print("奇数个，中位数是：", merged[total // 2])
else:
    print("偶数个，中位数是：", (merged[total // 2 - 1] + merged[total // 2]) / 2)
````

返回中位数（可能是小数）。

示例 2：`nums1 = [1,2]`，`nums2 = [3,4]`，合并是 `[1,2,3,4]`，中位数 2.5。

- 有一个数组可能为空。
- 偶数个时要取平均，注意结果是小数。
- 面试时先说 O(m+n) 的归并版本，再讲二分优化。


## 解法名称

**双指针归并（Merge and Pick）** O(m+n)；进阶是**二分分割（Binary Search on Partition）** O(log(min(m,n)))。

## 暴力思路

把两个数组拼起来排序，时间 O((m+n) log(m+n))，不符合要求。

## 优化思路（先掌握这个）

双指针归并：和合并两个有序链表一模一样，只是把节点换成了数组元素。合并完按下标取中位数即可。

时间 O(m+n)，空间 O(m+n)。

## 进阶思路（能说清就行）

不用真的合并。想象把两个数组一起分成左右两半，左边一共 `(m+n+1)//2` 个元素。我们在较短的数组上二分，找一个分割点 i，使得：

- 左边最大的元素 ≤ 右边最小的元素；
- 左边正好有 `(m+n+1)//2` 个元素。

找到之后，中位数只由“左边最大”和“右边最小”决定。时间 O(log(min(m, n)))，空间 O(1)。下面的“第二层答案”给了完整代码。

## 题目 3 做题前先补：0/1 背包和“倒序遍历”

问题转化：能不能把数组分成两部分，使两部分的和相等？

先算总和 `total`：

- 如果 `total` 是奇数，直接返回 `False`（没法平分）；
- 否则目标就是 `target = total // 2`：**能不能从数组里挑出一些数，让它们的和正好等于 target？**

这就是**0/1 背包**：每个数字只能用一次。

`dp[j]` 的定义：**能不能用数组里的数字凑出和 j**（布尔值）。

转移：对每个数字 `num`，`dp[j] = dp[j] or dp[j - num]`。

**关键：内层必须从大到小遍历 `j`。** 为什么？

因为 `dp[j]` 要用 `dp[j - num]`，如果从小到大遍历，`dp[j - num]` 可能已经被本轮更新过了，那就相当于同一个数字用了两次（变成完全背包）。倒着遍历时，`dp[j - num]` 一定还是上一轮的结果，保证每个数字只用一次。

**这是 0/1 背包最重要的一个细节，面试必问。**


````python
# 手工看一次 0/1 背包的倒序为什么重要
nums = [1, 5, 11, 5]
total = sum(nums)
target = total // 2
print("总和", total, "，目标", target)

dp = [False] * (target + 1)
dp[0] = True

for num in nums:
    j = target
    while j >= num:
        if dp[j - num]:
            dp[j] = True
        j = j - 1
    print("处理完数字", num, "，dp =", dp)

print("能不能凑出", target, "：", dp[target])
````

面试表达：

## 题目 4 做题前先补：栈里存下标，再放一个“哨兵”

这题要求最长的**有效括号子串**长度。

用栈存**下标**（不是字符），因为最后要算长度。

技巧：**一开始先在栈里放一个 -1 当哨兵**，它表示“最后一个没匹配上的位置”。

- 遇到 `(`：把它的下标压栈；
- 遇到 `)`：
  - 先弹栈（相当于和最近的 `(` 配对）；
  - 如果栈空了：说明这个 `)` 是多余的，把它的下标压栈，作为新的“断点”；
  - 如果栈不空：当前下标减去栈顶下标，就是一段有效括号的长度，更新答案。

为什么“栈顶下标”能算长度？因为栈顶就是上一个没被匹配的位置，两者之间的部分全是成对匹配的。

**这是栈题里很经典的一个技巧，和每日温度“栈里存下标”是同一个思路。**


````python
# 手工走一遍 ")()())"
s = ")()())"
stack = [-1]
best = 0

for i in range(len(s)):
    if s[i] == "(":
        stack.append(i)
        print("遇到 ( ：入栈", i, "，栈 =", stack)
    else:
        stack.pop()
        if len(stack) == 0:
            stack.append(i)
            print("遇到 ) ：栈空了，把", i, "当新断点，栈 =", stack)
        else:
            length = i - stack[-1]
            if length > best:
                best = length
            print("遇到 ) ：匹配成功，长度", length, "，当前最好", best)

print("最长有效括号长度：", best)
````
