## 1. 两数之和

LeetCode 1. Two Sum

给你一个整数列表 `nums`，再给你一个整数 `target`。

请在 `nums` 中找到两个不同位置的数，使它们相加等于 `target`，并返回这两个数的下标。

要求：

- 返回的是下标，不是数字本身。
- 同一个元素不能使用两次。
- 两个下标必须不同。
- 题目通常保证存在一个答案。

示例：

`nums = [2, 7, 11, 15]`, `target = 9`

返回：`[0, 1]`




```python
# 题目：两数之和
# 解法：一次遍历哈希表（One-pass Hash Table）
# 输入：整数列表 nums 和整数 target。
# 目标：找到两个不同下标 i、j，让 nums[i] + nums[j] == target。
# 输出：返回这两个下标组成的列表，例如 [0, 1]。
# 注意：不能重复使用同一个元素。
nums = [2, 7, 11, 15]
target = 9


def two_sum(nums, target):
    seen = {}

    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [i, seen[need]]
        else:
            seen[x] = i

    return []


print(two_sum(nums, target))
```

## 2. 字母异位词分组

LeetCode 49. Group Anagrams

给你一个字符串列表 `strs`，请把其中的字符串分组。

如果两个字符串包含的字母种类和每个字母出现次数都完全一样，只是排列顺序不同，那么它们属于同一组。

要求：

- 每个字符串都必须出现在某个分组里。
- 返回的是分组后的二维列表。
- 分组之间的顺序不重要。
- 同一组里字符串的顺序通常也不重要。
- 空字符串如果出现，也可以作为一组。

示例：

`strs = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']`

一种可能的返回结果：`[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]`

```python
# 题目：字母异位词分组
# 解法：排序 + 哈希分组（Sorting + Hash Map Grouping）
# 输入：字符串列表 strs。
# 目标：把字母组成相同、只是顺序不同的字符串放到同一组。
# 输出：二维列表，例如 [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]。
# 注意：分组顺序不重要，同组内顺序通常也不重要。
strs = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']


from collections import defaultdict

def group_anagrams(strs):
    groups = defaultdict(list)

    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)

    return list(groups.values())     # value取值  values取列表


print(group_anagrams(strs))
```

## 3. 移动零

LeetCode 283. Move Zeroes

给你一个整数列表 `nums`，请把列表里的所有 `0` 移动到末尾，同时保持所有非零元素的相对顺序不变。

要求：

- 必须原地修改 `nums`。
- 不需要返回新列表。
- 非零元素的相对顺序不能改变。

示例：

原列表：`[0, 1, 0, 3, 12]`

修改后：`[1, 3, 12, 0, 0]`

```python
# 题目：移动零
# 解法：快慢指针 / 原地交换（Fast-Slow Pointers / In-place Swap）
# 输入：整数列表 nums。
# 目标：把所有 0 移到列表末尾。
# 要求：非零元素的相对顺序不能变。
# 输出：不需要返回新列表，直接修改原来的 nums。
# 例子：[0, 1, 0, 3, 12] 改成 [1, 3, 12, 0, 0]。
nums = [0,1,0,3,12]

def move_zeroes(nums):
    slow = 0

    for fast in range(0,len(nums)):
        if nums[fast] != 0:
            nums[slow],nums[fast] = nums[fast],nums[slow]
            slow += 1


move_zeroes(nums)

print(nums)
```

## 4. 盛最多水的容器

LeetCode 11. Container With Most Water

给你一个非负整数列表 `height`，其中 `height[i]` 表示第 `i` 个位置有一条高度为 `height[i]` 的竖线。

请从这些竖线里选出两条，和 x 轴一起组成一个容器，返回这个容器最多可以装多少水。

要求：

- 只能选择两条竖线作为左右边界。
- 容器宽度是两个下标之间的距离。
- 容器高度由较短的那条竖线决定。
- 竖线不能倾斜。
- 返回最大面积。

示例：

`height = [1, 8, 6, 2, 5, 4, 8, 3, 7]`

返回：`49`

```python
# 题目：盛最多水的容器
# 解法：双指针 + 贪心（Two Pointers + Greedy）
# 输入：非负整数列表 height，每个数表示一条竖线的高度。
# 目标：选择两条竖线，让它们和 x 轴组成的容器面积最大。
# 面积：(right - left) * min(height[left], height[right])。
# 输出：最大面积，是一个整数。
# 注意：竖线不能倾斜，只能选择两条。
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]


def max_area(height):
    left,right = 0,len(height)-1
    ans = 0

    while left<right:

        width = right-left
        h = min(height[left],height[right])
        area = width*h
        ans = max(ans,area)

        if height[left]<height[right]:
            left +=1
        else:
            right -=1

    return ans

print('最大面积是',max_area(height))
    
```
