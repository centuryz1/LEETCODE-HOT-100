## Python 基础 2：字典 dict

`dict` 可以理解成一张查询表：通过 key 快速找到 value。

例如：记录某个数字出现在哪个下标。

```python
seen = {}

seen[2] = 0   # 数字 2 出现在下标 0
seen[7] = 1   # 数字 7 出现在下标 1
seen[9] = 2
print(seen)
print(2 in seen)
print(9 in seen)
print(seen[7])
```

为什么算法题喜欢用字典？

因为 `x in seen` 通常可以看作 O(1)，也就是查得很快。

暴力枚举两层循环通常是 O(n^2)。如果能用字典把「查找」变快，经常可以优化成 O(n)。

## Python 基础 3：defaultdict

有时候我们想按某个 key 分组。

比如：`eat`、`tea`、`ate` 排序后都是 `aet`，所以它们应该放到同一组。

```python
from collections import defaultdict

groups = defaultdict(list)

groups['aet'].append('eat')
groups['aet'].append('tea')
groups['aet'].append('ate')
groups['ant'].append('tan')

print(groups)
print(list(groups.values()))
```

# 题目 1：两数之和

LeetCode 1. Two Sum

## 题目描述（改写版）

给你一个整数列表 `nums`，再给你一个整数 `target`。

你需要在 `nums` 里找到两个不同位置的数，使这两个数相加正好等于 `target`，然后返回这两个数在列表里的下标。

需要注意：

- 返回的是下标，不是数字本身。
- 同一个元素不能使用两次，也就是两个下标必须不同。
- 题目通常保证一定存在一个答案。
- 返回下标的顺序一般不重要，但我们习惯返回先出现的下标在前。

输入：

- `nums`: 一个整数列表，例如 `[2, 7, 11, 15]`
- `target`: 一个整数，例如 `9`

输出：

- 一个长度为 2 的列表，表示两个数的下标，例如 `[0, 1]`

例子：

`nums = [2, 7, 11, 15]`, `target = 9`

因为 `nums[0] + nums[1] = 2 + 7 = 9`，所以返回 `[0, 1]`。

## 先想暴力法

最直接的方法：枚举所有两个数的组合。

也就是：

- 先拿第 0 个数，和后面的数挨个试。
- 再拿第 1 个数，和后面的数挨个试。
- 一直试到找到答案。

这个方法能做出来，但时间复杂度是 O(n^2)。

```python
def two_sum_bruteforce(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []


print(two_sum_bruteforce([2, 7, 11, 15], 9))
```

## 解法名称

本题常用专业说法：**一次遍历哈希表**，英文可以说 **One-pass Hash Table**。

如果先讲暴力法，可以说：**双重循环枚举**，英文可以说 **Brute Force Enumeration**。

面试里你可以这样开头：这题我会用一次遍历哈希表来做，把查找另一个数的过程从线性查找优化成哈希查询。

## 优化思路

对于当前数字 `x`，我们真正想找的是：

`target - x`

例如 `target = 9`：

- 看到 `2`，需要找 `7`。
- 看到 `7`，需要找 `2`。

如果我们用字典记录已经见过的数字，就可以很快查到它。

```python
nums = [2, 7, 11, 15]
target = 9
seen = {}

for i, x in enumerate(nums):
    need = target - x
    print('当前位置:', i, '当前数字:', x, '需要找:', need, '已经见过:', seen)

    if need in seen:
        print('找到答案:', [seen[need], i])
        break

    seen[x] = i
```

## 参考答案：两数之和

```python
def two_sum_answer(nums, target):
    seen = {}

    for i, x in enumerate(nums):
        need = target - x

        if need in seen:
            return [seen[need], i]

        seen[x] = i

    return []


print(two_sum_answer([2, 7, 11, 15], 9))
print(two_sum_answer([3, 2, 4], 6))
print(two_sum_answer([3, 3], 6))
```

面试表达：

# 题目 2：字母异位词分组

LeetCode 49. Group Anagrams

## 题目描述（改写版）

给你一个字符串列表 `strs`，请你把其中的字符串分组。

如果两个字符串包含的字母种类和每个字母出现次数都完全一样，只是排列顺序不同，那么它们属于同一组。

这种字符串叫「字母异位词」。

需要注意：

- 每个字符串都必须出现在某个分组里。
- 返回的是分组后的列表，也就是列表里面再套列表。
- 分组之间的顺序不重要。
- 同一组里字符串的顺序通常也不重要。
- 空字符串如果出现，也可以作为一组。

输入：

- `strs`: 字符串列表，例如 `['eat', 'tea', 'tan', 'ate', 'nat', 'bat']`

输出：

- 分组后的二维列表，例如 `[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]`

例子：

`eat`、`tea`、`ate` 都由 `a`、`e`、`t` 组成，所以它们是一组。  
`tan`、`nat` 都由 `a`、`n`、`t` 组成，所以它们是一组。  
`bat` 没有其他同组字符串，所以自己单独一组。

## 解法名称

本题常用专业说法：**排序 + 哈希分组**，英文可以说 **Sorting + Hash Map Grouping**。

还有一种进阶写法叫：**字符计数 + 哈希分组**，英文可以说 **Character Count + Hash Map Grouping**。今天先掌握排序写法。

面试里你可以这样开头：这题我会先把每个字符串转换成一个统一的 key，再用哈希表按 key 分组。

## 关键想法

我们需要给同一组字符串设计同一个 key。

最简单的方法：排序。

- `eat` 排序后是 `aet`
- `tea` 排序后是 `aet`
- `ate` 排序后是 `aet`

所以可以用排序后的字符串作为分组 key。

```python
test1 = 'waffkfawfwlf'
print(sorted(test1))
print(''.join(sorted(test1)))
```

## 你来写：字母异位词分组

动手前再确认一遍：

- 输入：字符串列表 `strs`。
- 目标：把字母组成相同、只是顺序不同的字符串放到同一组。
- 输出：二维列表，例如 `[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]`。
- 注意：分组顺序不重要，同组内顺序通常也不重要。

提示：

1. `groups = defaultdict(list)`。
2. 遍历每个字符串 `s`。
3. `key = ''.join(sorted(s))`。
4. `groups[key].append(s)`。
5. 最后返回 `list(groups.values())`。

```python
# 题目：字母异位词分组
# 解法：排序 + 哈希分组（Sorting + Hash Map Grouping）
# 输入：字符串列表 strs。
# 目标：把字母组成相同、只是顺序不同的字符串放到同一组。
# 输出：二维列表，例如 [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]。
# 注意：分组顺序不重要，同组内顺序通常也不重要。

def group_anagrams(strs):
    groups = defaultdict(list)

    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)

    return list(groups.values())

print(group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

## 参考答案：字母异位词分组

```python
def group_anagrams_answer(strs):
    groups = defaultdict(list)    # 键不存在时，应该调用哪个函数生成默认值。这里是list

    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)

    return list(groups.values())


print(group_anagrams_answer(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

解释：非零数字是 `1, 3, 12`，它们的相对顺序保持不变，所有 `0` 被放到了后面。

## 解法名称

本题常用专业说法：**快慢指针**，也属于 **双指针**。英文可以说 **Fast-Slow Pointers** 或 **Two Pointers**。

因为这题要求直接修改原数组，也可以补一句：这是一个 **原地算法**，英文是 **In-place Algorithm**。

面试里你可以这样开头：这题我会用快慢指针原地处理数组，`fast` 负责扫描，`slow` 负责记录下一个非零元素应该放的位置。

## 双指针怎么理解

我们用两个位置：

- `fast`：负责从左到右扫描每个元素。
- `slow`：表示下一个非零元素应该放到哪里。

看到非零数，就把它交换到 `slow` 的位置，然后 `slow` 往右走一步。

```python
nums = [0, 1, 0, 3, 12]
slow = 0

for fast in range(len(nums)):
    print('fast =', fast, 'slow =', slow, '当前数组 =', nums)
    if nums[fast] != 0:
        nums[slow], nums[fast] = nums[fast], nums[slow]
        slow += 1

print('最终结果:', nums)
```

# 题目 4：盛最多水的容器

LeetCode 11. Container With Most Water

这是今天的加练题。如果你刚开始学，今天先理解思路即可。

## 题目描述（改写版）

给你一个整数列表 `height`，其中 `height[i]` 表示第 `i` 个位置有一条高度为 `height[i]` 的竖线。

你需要从这些竖线里选出两条，和 x 轴一起组成一个容器。请返回这个容器最多可以装多少水。

需要注意：

- 只能选择两条竖线作为左右边界。
- 容器的宽度是两个下标之间的距离。
- 容器的高度由较短的那条竖线决定。
- 不能把竖线倾斜，也就是只能按原来的竖直高度计算。
- 返回的是最大面积，也就是最多能装的水量。

输入：

- `height`: 一个非负整数列表，例如 `[1, 8, 6, 2, 5, 4, 8, 3, 7]`

输出：

- 一个整数，表示最大可装水量，例如 `49`

例子：

对于 `height = [1, 8, 6, 2, 5, 4, 8, 3, 7]`，选择下标 `1` 和下标 `8` 的两条线：

- 宽度是 `8 - 1 = 7`
- 高度是 `min(8, 7) = 7`
- 面积是 `7 * 7 = 49`

所以答案是 `49`。

## 面积怎么算

如果左边下标是 `left`，右边下标是 `right`：

- 宽度：`right - left`
- 高度：两边较短的那根，`min(height[left], height[right])`
- 面积：`宽度 * 高度`

```python
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
left = 0
right = len(height) - 1

width = right - left
h = min(height[left], height[right])
area = width * h

print(width, h, area)
```

## 解法名称

本题常用专业说法：**双指针 + 贪心**，英文可以说 **Two Pointers + Greedy**。

这里的贪心体现在：每次移动较短的那条边，因为容器高度由短边决定，只有移动短边才可能找到更高的有效高度。

面试里你可以这样开头：这题我会用双指针加贪心，从左右两端开始，每次计算面积并移动较短的一边。

## 为什么移动短边

容器能装多少水，由短边决定。

如果移动长边，宽度变小，短边还没变高，面积很难变大。  
所以我们每次移动较短的那一边，去赌能遇到更高的短边。

```python
def max_area_trace(height):
    left, right = 0, len(height) - 1
    ans = 0

    while left < right:
        width = right - left
        h = min(height[left], height[right])
        area = width * h
        ans = max(ans, area)

        print('left:', left, 'right:', right, '面积:', area, '当前最大:', ans)

        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return ans


max_area_trace([1, 8, 6, 2, 5, 4, 8, 3, 7])
```

动手前再确认一遍：

- 输入：非负整数列表 `height`，每个数表示一条竖线的高度。
- 目标：选择两条竖线，让它们和 x 轴组成的容器面积最大。
- 面积：`(right - left) * min(height[left], height[right])`。

## 参考答案：盛最多水的容器

```python
def max_area_answer(height):
    left, right = 0, len(height) - 1
    ans = 0

    while left < right:
        width = right - left
        h = min(height[left], height[right])
        ans = max(ans, width * h)

        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return ans


print(max_area_answer([1, 8, 6, 2, 5, 4, 8, 3, 7]))
print(max_area_answer([1, 1]))
```
