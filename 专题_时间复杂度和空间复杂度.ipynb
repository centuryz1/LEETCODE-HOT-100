# 时间复杂度与空间复杂度：秋招面试专题

这份专题解决两个问题：

1. 一段代码运行得有多“费时间”？
2. 一段代码运行得有多“费内存”？

学习分三层：

- **必学**：会判断 `O(1)`、`O(n)`、`O(n²)`、`O(log n)`。
- **面试常用**：会分析排序、哈希表、双指针、二分查找。
- **进阶**：会分析递归栈、多个输入规模和均摊复杂度。

建议分两次学习，每次 30 分钟左右。

## 第一部分：复杂度到底在衡量什么？

### 1. 为什么不直接说“运行 2 秒”？

同一段代码的实际时间会受到很多因素影响：

- 电脑性能；
- Python 版本；
- 当时运行的其他程序；
- 输入数据大小；
- 输入数据的具体分布。

所以面试中通常不问“运行几秒”，而是问：

> 当输入规模变大时，运算次数和额外内存大致怎样增长？

复杂度关注的是**增长趋势**，不是精确秒数或精确字节数。

### 2. `n` 是什么？

`n` 通常代表输入规模，但具体含义由题目决定：

| 输入 | `n` 可能表示 |
|---|---|
| 列表 `nums` | `len(nums)` |
| 字符串 `s` | `len(s)` |
| 链表 | 节点数量 |
| 二叉树 | 节点数量 |
| 矩阵 | 可能需要行数 `m` 和列数 `n` |

分析复杂度前，第一句话应该是：

> 设输入数组长度为 `n`。

如果有两个独立输入，例如长度分别为 `m` 和 `n` 的两个数组，不要强行都写成 `n`。

### 3. 大 O 表示法是什么？

`O(...)` 读作“大 O”。它用来描述输入规模变大时，资源消耗的增长级别。

例如：

```text
输入有 n 个元素，代码大约检查 n 次       -> O(n)
每个元素都要和所有元素比较一次          -> O(n²)
每次把搜索范围缩小一半                  -> O(log n)
无论输入多大，只做固定几步              -> O(1)
```

面试中默认讨论的是渐进复杂度，也就是 `n` 很大时最主要的增长部分。

````python
sizes = [10, 100, 1000]

print(f"{'n':>6} {'O(1)':>8} {'O(log n)':>10} {'O(n)':>10} {'O(n^2)':>12}")
for n in sizes:
    import math
    print(f"{n:>6} {1:>8} {math.ceil(math.log2(n)):>10} {n:>10} {n*n:>12}")
````

## 第二部分：常见复杂度从快到慢

```text
O(1)
  ↓
O(log n)
  ↓
O(n)
  ↓
O(n log n)
  ↓
O(n²)
  ↓
O(2ⁿ)
  ↓
O(n!)
```

| 复杂度 | 常见场景 |
|---|---|
| `O(1)` | 访问列表某个下标、字典平均查询 |
| `O(log n)` | 二分查找 |
| `O(n)` | 遍历数组、滑动窗口 |
| `O(n log n)` | 高效排序 |
| `O(n²)` | 两层遍历、三数之和的排序双指针解法 |
| `O(2ⁿ)` | 枚举所有子集 |
| `O(n!)` | 枚举所有排列 |

这里的“快慢”是同等输入规模下的大致增长趋势，不代表每个 `O(n)` 程序一定比每个 `O(log n)` 程序实际更快。

## 第三部分：时间复杂度怎么算？

使用下面四步：

1. **确定输入规模**：`n`、`m` 分别代表什么？
2. **找重复执行的操作**：循环、递归、排序。
3. **写出大致执行次数**：`n`、`n²`、`log n` 等。
4. **只保留增长最快的部分**：去掉常数和低阶项。

例如：

```text
3n + 10     -> O(n)
2n² + n     -> O(n²)
5           -> O(1)
```

为什么能去掉常数？因为复杂度比较的是增长级别。`n` 足够大时，`n²` 的影响远大于 `n` 和固定常数。

### 4. 固定操作：`O(1)`

```python
def first_number(nums):
    return nums[0]
```

无论列表长度是 10 还是 100 万，都只访问一次下标，因此是 `O(1)`。

注意：`O(1)` 不代表“只执行一行”，而是执行次数不随输入规模增长。

下面虽然有三行计算，仍然是 `O(1)`：

```python
a = nums[0]
b = nums[-1]
return a + b
```

````python
def edge_sum(nums):
    first = nums[0]
    last = nums[-1]
    return first + last


assert edge_sum([1, 2, 3]) == 4
assert edge_sum(list(range(10000))) == 9999
print("列表变长了，但函数仍只访问首尾两个位置。")
````

### 5. 一层循环：通常是 `O(n)`

```python
total = 0
for number in nums:
    total += number
```

数组有 `n` 个元素，循环执行 `n` 次，每次做常数时间操作，所以：

```text
n × O(1) = O(n)
```

这叫**线性时间复杂度（Linear Time Complexity）**。

````python
def count_linear_work(n):
    operations = 0
    for _ in range(n):
        operations += 1
    return operations


for n in [10, 100, 1000]:
    print(f"n={n:4}, 操作次数={count_linear_work(n)}")
````

### 6. 两层循环：先看每层分别执行多少次

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

外层执行 `n` 次，每次内层又执行 `n` 次：

```text
n × n = n² -> O(n²)
```

这叫**平方时间复杂度（Quadratic Time Complexity）**。

但不要看到两个 `for` 就立刻说 `O(n²)`，还要判断它们是嵌套还是前后并列，以及每层范围是否真的是 `n`。

````python
def count_quadratic_work(n):
    operations = 0
    for _ in range(n):
        for _ in range(n):
            operations += 1
    return operations


for n in [10, 100, 1000]:
    print(f"n={n:4}, 操作次数={count_quadratic_work(n):,}")
````

### 7. 前后两个循环：相加后取增长最快的项

```python
for x in nums:       # n 次
    print(x)

for x in nums:       # n 次
    print(x)
```

总次数是 `n + n = 2n`，去掉常数后是 `O(n)`，不是 `O(n²)`。

再看：

```python
for x in nums:              # O(n)
    print(x)

for x in nums:              # O(n²)
    for y in nums:
        print(x, y)
```

总复杂度是 `O(n + n²)`，只保留增长最快的 `n²`，所以是 `O(n²)`。

口诀：

- 前后执行：复杂度相加，最后保留增长最快的项。
- 嵌套执行：复杂度通常相乘。

### 8. 条件分支：通常取最慢的分支

```python
if condition:
    do_constant_work()     # O(1)
else:
    for x in nums:         # O(n)
        process(x)
```

讨论最坏情况时，取较慢的分支，因此是 `O(n)`。

面试中如果没有特别说明，一般先回答**最坏时间复杂度（Worst-case Time Complexity）**。

有时面试官也会追问：

- 最好情况（Best Case）
- 平均情况（Average Case）
- 均摊复杂度（Amortized Complexity）

入门阶段先把最坏情况算准。

### 9. 每次减半：`O(log n)`

```python
i = n
while i > 1:
    i //= 2
```

变化过程类似：

```text
n -> n/2 -> n/4 -> n/8 -> ... -> 1
```

需要“除以 2”多少次才能到 1？大约是 `log₂n` 次，所以是 `O(log n)`。

二分查找也是同样思路：每次排除一半搜索范围。

看到“每次减半、翻倍”时，要想到 `log n`。

````python
def halving_steps(n):
    steps = 0
    while n > 1:
        n //= 2
        steps += 1
    return steps


for n in [8, 16, 1024, 1_000_000]:
    print(f"n={n:>8}, 减半次数={halving_steps(n)}")
````

### 10. 嵌套循环不一定都是 `O(n²)`

#### 情况 A：三角形循环

```python
for i in range(n):
    for j in range(i):
        ...
```

执行次数约为：

```text
0 + 1 + 2 + ... + (n-1) = n(n-1)/2
```

去掉常数和低阶项，仍然是 `O(n²)`。

#### 情况 B：两个不同输入

```python
for x in nums:       # n
    for y in words:  # m
        ...
```

复杂度是 `O(nm)`，不是随便写成 `O(n²)`。

#### 情况 C：内层每次减半

```python
for x in nums:       # n
    k = n
    while k > 1:     # log n
        k //= 2
```

复杂度是 `O(n log n)`。

## 第四部分：Python 常用操作复杂度

下面是面试中常用的平均情况或常见实现复杂度：

| Python 操作 | 时间复杂度 |
|---|---|
| `nums[i]` | `O(1)` |
| `nums.append(x)` | 均摊 `O(1)` |
| `nums.pop()` | `O(1)` |
| `nums.pop(0)` | `O(n)` |
| `x in list` | `O(n)` |
| `x in set` | 平均 `O(1)` |
| `dict[key]` | 平均 `O(1)` |
| `len(nums)` | `O(1)` |
| `nums[a:b]` | `O(k)`，`k` 为切片长度 |
| `sorted(nums)` | `O(n log n)` |
| `"".join(parts)` | 与总字符数成线性关系 |

重要提醒：字典和集合查询通常回答“平均 `O(1)`”；极端哈希冲突下最坏情况可能退化。

## 第五部分：空间复杂度怎么算？

### 11. 空间复杂度衡量什么？

空间复杂度描述：输入规模变大时，算法使用的内存怎样增长。

面试中经常讨论的是**额外空间复杂度（Auxiliary Space Complexity）**：

> 不把题目本来给你的输入算进去，只计算算法为了完成任务额外申请的空间。

例如函数接收一个长度为 `n` 的数组，这个输入数组通常不算额外空间；你又创建了一个长度为 `n` 的新数组，这个新数组算 `O(n)` 额外空间。

### 12. 固定数量变量：`O(1)` 空间

```python
left = 0
right = len(nums) - 1
answer = 0
```

无论 `nums` 多长，都只使用固定数量变量，因此额外空间是 `O(1)`。

双指针算法经常具有：

- 时间复杂度：`O(n)`
- 额外空间复杂度：`O(1)`

这也是双指针的优势之一。

### 13. 新建随输入增长的容器：`O(n)` 空间

```python
seen = set()
for number in nums:
    seen.add(number)
```

最坏情况下，集合保存 `n` 个不同数字，因此额外空间是 `O(n)`。

下面也一样：

```python
copy = nums[:]
```

复制了长度为 `n` 的列表，额外空间是 `O(n)`。

````python
def unique_numbers(nums):
    seen = set()
    for number in nums:
        seen.add(number)
    return seen


for n in [10, 100, 1000]:
    result = unique_numbers(range(n))
    print(f"输入规模={n:4}, 集合元素数量={len(result)}")
````

### 14. 递归调用栈也占空间

```python
def countdown(n):
    if n == 0:
        return
    countdown(n - 1)
```

调用 `countdown(n)` 时，在到达 0 之前会保留大约 `n` 层函数调用，因此：

- 时间复杂度：`O(n)`
- 递归栈空间：`O(n)`

很多初学者只看到了“没有新建列表”，就误以为空间是 `O(1)`。递归栈不能漏算。

````python
def countdown_depth(n):
    if n == 0:
        return 1
    return 1 + countdown_depth(n - 1)


for n in [3, 10, 50]:
    print(f"n={n:2}, 最大调用层数约为 {countdown_depth(n)}")
````

### 15. 输出结果算不算空间？

要看面试官采用的口径。

例如返回所有排列，输出本身就非常大。常见表达方式是分开说明：

> 不计算返回结果所占空间时，额外空间是……；如果把输出结果也计算在内，总空间是……

这样比直接争论“算还是不算”更专业。

同样，Python 的 `nums.sort()` 是原地排序，而 `sorted(nums)` 会创建新列表。讨论空间时应该说明语言实现和使用方式。

## 第六部分：秋招常见算法复杂度

| 算法 | 时间复杂度 | 额外空间复杂度 |
|---|---:|---:|
| 遍历数组 | `O(n)` | `O(1)` |
| 双指针遍历 | `O(n)` | `O(1)` |
| 二分查找（迭代） | `O(log n)` | `O(1)` |
| 哈希表查找两数之和 | `O(n)` | `O(n)` |
| 排序后双指针 | `O(n log n)` | 取决于排序实现 |
| 三数之和：排序 + 双指针 | `O(n²)` | 通常不计结果时接近 `O(1)`，还需说明排序开销 |
| 遍历 `m × n` 矩阵 | `O(mn)` | `O(1)` 或取决于辅助结构 |
| 枚举所有子集 | `O(2ⁿ)` | 与递归栈和输出有关 |

面试表达中，不要只报结论，还要说原因。

### 16. 例子：两数之和

哈希表解法：

```python
def two_sum(nums, target):
    seen = {}

    for i, number in enumerate(nums):
        need = target - number
        if need in seen:
            return [seen[need], i]
        seen[number] = i
```

分析：

- 数组最多遍历一次：`O(n)`。
- 字典查询和写入平均为 `O(1)`。
- 最坏情况下字典保存 `n` 个元素：`O(n)` 额外空间。

面试回答：

> 设数组长度为 `n`。算法只遍历一次数组，每次进行平均 `O(1)` 的哈希表查询和写入，因此平均时间复杂度是 `O(n)`；哈希表最多保存 `n` 个元素，因此空间复杂度是 `O(n)`。

````python
def two_sum(nums, target):
    seen = {}
    for i, number in enumerate(nums):
        need = target - number
        if need in seen:
            return [seen[need], i]
        seen[number] = i
    return None


assert two_sum([2, 7, 11, 15], 9) == [0, 1]
print(two_sum([3, 2, 4], 6))
````

### 17. 例子：三数之和

排序 + 双指针解法：

```text
排序：O(n log n)
固定第一个数：n 次
每次左右指针合计移动：O(n)
双指针主体：O(n²)
```

总时间：

```text
O(n log n + n²) = O(n²)
```

因为 `n²` 增长更快，所以最终保留 `O(n²)`。

空间复杂度要谨慎表达：不计算结果列表时，左右指针只占 `O(1)`；但 Python 排序实现可能使用额外空间，因此面试中可以明确说明“忽略排序实现的额外空间”或进一步讨论实现细节。

### 18. 例子：二分查找

```python
while left <= right:
    middle = (left + right) // 2
    if nums[middle] < target:
        left = middle + 1
    elif nums[middle] > target:
        right = middle - 1
    else:
        return middle
```

每轮把候选范围缩小一半，因此时间复杂度是 `O(log n)`。

只使用 `left`、`right`、`middle` 等固定数量变量，因此迭代写法的额外空间是 `O(1)`。

````python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1
    steps = 0

    while left <= right:
        steps += 1
        middle = (left + right) // 2
        if nums[middle] < target:
            left = middle + 1
        elif nums[middle] > target:
            right = middle - 1
        else:
            return middle, steps
    return -1, steps


for n in [16, 1024, 1_048_576]:
    index, steps = binary_search(list(range(n)), n - 1)
    print(f"n={n:>8}, 查找步数={steps}")
````

## 第七部分：容易答错的地方

1. **看到两段 `for` 就说 `O(n²)`**：前后并列通常相加，不是相乘。
2. **看到两层循环就机械说 `O(n²)`**：还要看范围，可能是 `O(nm)` 或 `O(n log n)`。
3. **忘记排序**：先排序再遍历通常至少有 `O(n log n)`。
4. **字典操作永远说 `O(1)`**：更严谨的是平均 `O(1)`。
5. **新建集合却说空间 `O(1)`**：集合大小会随输入增长。
6. **递归没建列表就说空间 `O(1)`**：递归调用栈也占空间。
7. **把实际运行时间当复杂度**：复杂度关注增长趋势。
8. **只报答案，不解释来源**：面试官更关心推导过程。

## 第八部分：面试回答模板

可以按四句话回答：

> 设输入规模为 `n`。算法的主要部分是……，它会执行……次，每次操作是……复杂度，因此总时间复杂度是……。算法额外使用了……，其规模最多为……，所以空间复杂度是……。

例如滑动窗口：

> 设字符串长度为 `n`。左右指针都只会从左向右移动，每个字符最多被加入和移出窗口各一次，因此时间复杂度是 `O(n)`。集合最多保存窗口内的 `n` 个字符，因此最坏空间复杂度是 `O(n)`。

这比只回答“时间 `O(n)`、空间 `O(n)`”完整得多。
