# 题目 1：有效的括号

LeetCode 20. Valid Parentheses

## 题目描述（改写版）

给你一个只包含圆括号 `()`、方括号 `[]` 和花括号 `{}` 的字符串 `s`。判断这些括号是否有效配对。

一个有效字符串必须同时满足：

1. 每个左括号都由相同类型的右括号关闭。
2. 括号关闭顺序正确，最后打开的括号必须最先关闭。
3. 每个右括号都有对应的左括号，最后也不能剩下未关闭的左括号。

### 输入和输出

- 输入字符串非空，只包含六种括号字符。
- 有效返回 `True`，否则返回 `False`。

### 示例

- `"()[]{}"` 返回 `True`。
- `"([{}])"` 返回 `True`。
- `"(]"` 返回 `False`，因为括号类型不匹配。
- `"([)]"` 返回 `False`，因为关闭顺序错误。

## 解法名称

专业名称：**栈匹配（Stack Matching）**。

## 不使用栈为什么麻烦

只统计左右括号数量不够。例如 `([)]` 的数量完全对应，但嵌套顺序错误。我们必须记住“最近一个还没关闭的左括号”。

## 栈思路

遇到左括号就压栈；遇到右括号时，它必须与当前栈顶左括号匹配。若栈为空或类型不匹配，立即返回 `False`。遍历结束后，栈也必须为空。

时间复杂度 `O(n)`，栈最多保存 `n` 个字符，空间复杂度 `O(n)`。

````python
s = "([{}])"
pairs = {")": "(", "]": "[", "}": "{"}
stack = []

for char in s:
    if char not in pairs:
        stack.append(char)
        print(f"读到左括号 {char}，压栈：{stack}")
    else:
        top = stack.pop()
        print(f"读到右括号 {char}，弹出 {top}，剩余：{stack}")
````

````python
def is_valid_parentheses_answer(s):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []

    for char in s:
        if char not in pairs:
            stack.append(char)
        else:
            if not stack or stack.pop() != pairs[char]:
                return False

    return not stack
````

# 题目 2：最小栈

LeetCode 155. Min Stack

## 题目描述（改写版）

请设计一个栈类 `MinStack`，支持以下操作，并且每个操作都要在 `O(1)` 时间内完成：

- `push(value)`：把整数压入栈顶。
- `pop()`：删除栈顶元素。
- `top()`：返回栈顶元素。
- `get_min()`：返回当前栈中的最小元素。

### 输入、输出与限制

- 测试程序会按顺序多次调用这些方法。
- 调用 `pop`、`top`、`get_min` 时保证栈非空。
- 最小值可能重复；弹出一个最小值后，旧的相同最小值仍要生效。

### 示例

依次压入 `-2、0、-3`，`get_min()` 返回 `-3`；弹出 `-3` 后，`top()` 返回 `0`，`get_min()` 返回 `-2`。

## 解法名称

专业名称：**辅助栈（Auxiliary Min Stack）**。

## 每次遍历求最小值

只维护普通栈时，`get_min()` 需要扫描全部元素，时间为 `O(n)`，不符合要求。

## 辅助栈思路

主栈保存所有值；辅助栈与主栈同步，每个位置保存“到当前位置为止的最小值”。压入新值时，压入 `min(value, 当前最小值)`；弹出时两个栈一起弹出。

四种操作时间复杂度都是 `O(1)`，额外空间复杂度 `O(n)`。

````python
values = []
minimums = []

for value in [-2, 0, -3]:
    values.append(value)
    current_min = value if not minimums else min(value, minimums[-1])
    minimums.append(current_min)
    print("主栈：", values, "最小值栈：", minimums)

values.pop()
minimums.pop()
print("同步弹出后，栈顶：", values[-1], "最小值：", minimums[-1])
````

````python
class MinStackAnswer:
    def __init__(self):
        self.values = []
        self.minimums = []

    def push(self, value):
        self.values.append(value)
        current_min = value if not self.minimums else min(value, self.minimums[-1])
        self.minimums.append(current_min)

    def pop(self):
        self.values.pop()
        self.minimums.pop()

    def top(self):
        return self.values[-1]

    def get_min(self):
        return self.minimums[-1]
````

## 题目 3 做题前先补：单调栈为什么保存下标

单调栈不是一种新的 Python 容器，仍然使用列表；“单调”描述的是我们主动维护的顺序。

本题栈里保存的是**还没有找到更高温度的日期下标**。对应温度保持从栈底到栈顶单调不增。

保存下标而不是只保存温度，是因为答案需要计算等待天数：

```python
answer[old_index] = current_index - old_index
```

当前温度更高时，要连续解决所有比它低的旧日期，所以这里必须使用 `while`，不能只用一次 `if`。

````python
temperatures = [73, 74, 75]
stack = []
answer = [0] * len(temperatures)

for current_index, current_temp in enumerate(temperatures):
    while stack and temperatures[stack[-1]] < current_temp:
        old_index = stack.pop()
        answer[old_index] = current_index - old_index
    stack.append(current_index)
    print("当前下标：", current_index, "栈：", stack, "答案：", answer)
````

# 题目 3：每日温度

LeetCode 739. Daily Temperatures

## 题目描述（改写版）

给你一个整数列表 `temperatures`，其中 `temperatures[i]` 是第 `i` 天的温度。请返回列表 `answer`：`answer[i]` 表示从第 `i` 天开始，还要等待多少天才会遇到更高温度。如果以后不会再升高，该位置填 0。

### 输入、输出与限制

- 输入非空，温度可以重复。
- 必须是严格更高，温度相同不能解决之前的等待。
- 输出长度与输入相同。

### 示例

输入 `[73,74,75,71,69,72,76,73]`，输出 `[1,1,4,2,1,1,0,0]`。例如第 2 天温度 75，要等 4 天才遇到 76。

## 解法名称

专业名称：**单调递减栈（Monotonic Decreasing Stack）**。

## 暴力思路

对每一天向右寻找第一个更高温度，最坏时间复杂度 `O(n^2)`。

## 单调栈思路

从左到右遍历。栈保存尚未解决的日期下标；当前温度高于栈顶日期温度时，当前日期就是栈顶日期遇到的第一个更高温度。持续弹栈并填写答案，最后把当前下标入栈。

每个下标最多入栈和出栈一次，因此时间复杂度 `O(n)`，空间复杂度 `O(n)`。

````python
temperatures = [73,74,75,71,69,72,76,73]
answer = [0] * len(temperatures)
stack = []

for current_index, current_temp in enumerate(temperatures):
    print(f"\n第 {current_index} 天，温度 {current_temp}")
    while stack and temperatures[stack[-1]] < current_temp:
        old_index = stack.pop()
        answer[old_index] = current_index - old_index
        print(f"  解决第 {old_index} 天，等待 {answer[old_index]} 天")
    stack.append(current_index)
    print("  未解决下标栈：", stack)

print("最终答案：", answer)
````

## 题目 4 做题前先补：一列能接多少水

先只看下标 `i` 这一列。它能接多少水，取决于左右两边最高墙中较矮的那一边：


原因是水面由短板决定。若左边最高 5、右边最高 3、当前柱高 1，那么水面最多到 3，这一列能接 `3 - 1 = 2`。

边界第一列和最后一列没有左右两堵墙，不能接水。题目保证柱高非负，所以正确计算后单列水量不会为负。

````python
height = [0, 2, 0, 3, 1, 0, 1, 2]
i = 5
left_max = max(height[:i + 1])
right_max = max(height[i:])
water_i = min(left_max, right_max) - height[i]

print("当前位置：", i)
print("当前柱高：", height[i])
print("左边最高：", left_max)
print("右边最高：", right_max)
print("这一列能接：", water_i)
````

### 输入、输出与限制

- 输入列表非空，柱子高度均为非负整数。
- 每根柱子宽度固定为 1。

### 示例 1

输入 `[0,1,0,2,1,0,1,3,2,1,2,1]`，总共可以接 6 单位雨水。

### 示例 2

输入 `[4,2,0,3,2,5]`，各低洼位置累计可以接 9 单位雨水。

## 四种解法的专业名称

1. **按列枚举（Brute-Force Column Scanning）**：每一列分别向左右寻找最高墙。
2. **前缀最大值 + 后缀最大值（Prefix and Suffix Maximum Arrays）**：提前保存每列左右最高墙。
3. **双指针水位法（Two-Pointer Waterline Method）**：使用 `left_max`、`right_max` 把额外空间优化为 `O(1)`，这是今天要求掌握的答案。
4. **单调栈分层结算（Monotonic Stack Basin Filling）**：按横向水层计算，作为进阶理解。

## 第一层：按列暴力

对每个位置向左、向右扫描最高墙，再用短板公式计算。容易理解，但时间复杂度 `O(n^2)`。

## 第二层：前后缀最大值

提前用两个数组保存每个位置的 `left_max` 和 `right_max`，之后每列 `O(1)` 计算。总时间 `O(n)`，额外空间 `O(n)`。

## 第三层：双指针优化

维护左右指针与两侧历史最高墙。哪一侧当前柱子更矮，就能确定哪一侧的水量，因为它的水位上限已经由较矮侧决定。处理该侧后向中间移动。

时间复杂度 `O(n)`，额外空间复杂度 `O(1)`。

````python
height = [4, 2, 0, 3, 2, 5]
left, right = 0, len(height) - 1
left_max = right_max = 0
water = 0

while left < right:
    print(f"\nleft={left}({height[left]}), right={right}({height[right]})")
    if height[left] <= height[right]:
        left_max = max(left_max, height[left])
        added = left_max - height[left]
        water += added
        print(f"处理左边：left_max={left_max}，新增水量={added}")
        left += 1
    else:
        right_max = max(right_max, height[right])
        added = right_max - height[right]
        water += added
        print(f"处理右边：right_max={right_max}，新增水量={added}")
        right -= 1
    print("累计水量：", water)

print("最终水量：", water)
````

## 接雨水的第二层答案：前缀最大值 + 后缀最大值

如果双指针暂时看不懂，先掌握这个版本。`left_max[i]` 和 `right_max[i]` 分别保存位置 `i` 左右两侧（包含自己）的最高柱子。它比暴力解法少了重复扫描。

````python
def trap_with_prefix_suffix(height):
    n = len(height)
    left_max = [0] * n
    right_max = [0] * n

    left_max[0] = height[0]
    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], height[i])

    right_max[-1] = height[-1]
    for i in range(n - 2, -1, -1):
        right_max[i] = max(right_max[i + 1], height[i])

    water = 0
    for i in range(n):
        water += min(left_max[i], right_max[i]) - height[i]
    return water


print(trap_with_prefix_suffix([4,2,0,3,2,5]))  # 9
````

## 接雨水的第四层答案：单调栈分层结算

这个版本把水看成横向的一层层。遇到更高柱子时，弹出的 `bottom` 是凹槽底部；新栈顶是左墙，当前柱子是右墙。水量为：

```text
宽度 × 有效高度
```

第一次学习只要求看懂三个角色，不要求立刻默写。

````python
def trap_with_monotonic_stack(height):
    stack = []
    water = 0

    for right in range(len(height)):
        while stack and height[right] > height[stack[-1]]:
            bottom = stack.pop()
            if not stack:
                break
            left = stack[-1]
            width = right - left - 1
            bounded_height = min(height[left], height[right]) - height[bottom]
            water += width * bounded_height
        stack.append(right)

    return water


print(trap_with_monotonic_stack([4,2,0,3,2,5]))  # 9
````

面试表达：

我使用双指针水位法，维护 left_max 和 right_max。每次处理当前柱子较矮的一侧，因为这一侧的可接水量已经能由自己的历史最高墙确定。时间 O(n)，额外空间 O(1)。
