## 题目 1 做题前先补：交换相邻节点要改哪三条指针

两两交换看起来简单，但第一次写几乎都会写错，因为交换两个相邻节点要动**三条指针**。

现在想把 1 和 2 换位置：

```text
dummy -> 1 -> 2 -> 3 -> 4
         ^    ^
       first second
```

交换之后希望变成：

```text
dummy -> 2 -> 1 -> 3 -> 4
```

做法只有三步，但顺序很重要：

1. `first.next = second.next`：让 1 先指向 3（也就是 2 后面的部分）。
2. `second.next = first`：让 2 指向 1。
3. `pre.next = second`：最后让前一个节点（这里是 dummy）指向 2。

为什么第 1 步用的是 `second.next`？因为这时 `second.next` 还指向 3，没有被改过。如果先写第 2 步，`second.next` 就变成 1 了，你就找不到 3 了。

**通用经验：链表题里，先把后面还要用到的节点用变量存下来，再改指针。**

下面的格子亲手跑一遍，看看每一步之后链表长什么样。


````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


head = build_list([1, 2, 3, 4])
print("交换前：", list_values(head))

# 先把前两个节点用变量存下来
first = head
second = head.next
print("first =", first.val, "，second =", second.val)

# 第 1 步：让 1 指向 3
first.next = second.next
print("第 1 步之后：", list_values(head))

# 第 2 步：让 2 指向 1
second.next = first
print("第 2 步之后（从 second 开始看）：", list_values(second))

# 第 3 步：让 second 成为新的头
head = second
print("第 3 步之后：", list_values(head))
````

# 题目 1：两两交换链表中的节点

LeetCode 24. Swap Nodes in Pairs

## 题目描述（改写版）

给你一条单链表的头节点 `head`。请你把链表中**相邻的两个节点两两交换**，然后返回交换后的头节点。

- 不能只交换节点里的值，要真正改动指针。
- 如果节点总数是奇数，最后一个节点保持不动。
- 如果链表为空，返回空。

## 输入

- `head`：单链表头节点，节点个数在 0 到 100 之间，节点值在 0 到 100 之间。

## 输出

返回交换后的新头节点。

## 示例

示例 1：`1 -> 2 -> 3 -> 4`，结果是 `2 -> 1 -> 4 -> 3`。

示例 2：`1 -> 2 -> 3`，结果是 `2 -> 1 -> 3`。最后的 3 没有搭档，保持不动。

示例 3：空链表，返回空。

## 易漏细节

- 头节点会被换掉，所以要么用虚拟头节点，要么单独处理头部。
- 循环条件是“后面还剩两个节点”：`pre.next` 和 `pre.next.next` 都不能为空。
- 交换完成后，`pre` 要移到 `first`（交换后的后一个节点），不是 `second`。


## 解法名称

**迭代 + 虚拟头节点（Iterative with Dummy Head）**。

## 暴力思路

把链表所有值读进列表，在列表里两两交换，再写回节点。

能过题，但改的是节点里的值，不是节点本身。面试官通常会追一句“不能改值，只能改指针”，所以这个思路只能用来理解题意。

## 优化思路

用虚拟头节点 `dummy` 解决“头节点会变”的问题，这个思路在 Day 06 删除倒数第 N 个节点时已经用过。

- `pre` 指向“已经处理完的部分的最后一个节点”，一开始是 `dummy`。
- 每一轮做四件事：

1. `first = pre.next`、`second = pre.next.next`，先把要交换的两个节点存下来。
2. 按前面学的三步改指针：`first.next = second.next`、`second.next = first`、`pre.next = second`。
3. `pre = first`，准备处理下一对。
4. 循环条件是 `while pre.next is not None and pre.next.next is not None`，也就是后面不足两个节点就停下。

时间 O(n)，只遍历一遍；空间 O(1)，只用了几个变量。

## 画图跟踪

以 `dummy -> 1 -> 2 -> 3 -> 4` 为例：

| 轮次 | pre 指向 | first | second | 交换后链表 |
| --- | --- | --- | --- | --- |
| 第 1 轮 | dummy | 1 | 2 | dummy -> 2 -> 1 -> 3 -> 4 |
| 第 2 轮 | 1 | 3 | 4 | dummy -> 2 -> 1 -> 4 -> 3 |
| 结束 | 3 | 后面不足两个节点，退出循环 | | |


````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def swap_pairs_trace(head):
    """带打印的版本，方便看清每一轮指针的变化。"""
    dummy = ListNode()
    dummy.next = head
    pre = dummy
    round_number = 1

    while pre.next is not None and pre.next.next is not None:
        first = pre.next
        second = pre.next.next
        print("第", round_number, "轮：准备交换", first.val, "和", second.val)

        first.next = second.next
        second.next = first
        pre.next = second

        print("    交换后整条链表：", list_values(dummy.next))
        pre = first
        round_number = round_number + 1

    return dummy.next


head = build_list([1, 2, 3, 4, 5])
result = swap_pairs_trace(head)
print("最终结果：", list_values(result))
````

## 参考答案：两两交换

先自己写完再看。重点对照三个地方：循环条件、三步改指针的顺序、`pre` 移动到哪里。


````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def swap_pairs_answer(head):
    dummy = ListNode()
    dummy.next = head
    pre = dummy

    while pre.next is not None and pre.next.next is not None:
        first = pre.next
        second = pre.next.next

        first.next = second.next
        second.next = first
        pre.next = second

        pre = first

    return dummy.next


print(list_values(swap_pairs_answer(build_list([1, 2, 3, 4]))))   # 期望 [2, 1, 4, 3]
print(list_values(swap_pairs_answer(build_list([1, 2, 3]))))      # 期望 [2, 1, 3]
print(list_values(swap_pairs_answer(build_list([]))))             # 期望 []
print(list_values(swap_pairs_answer(build_list([1]))))            # 期望 [1]
````

## 拓展答案：递归写法

面试里也可以先写递归版本，代码更短，但空间是 O(n) 的递归栈。

思路：把前两个节点换好之后，剩下的部分递归处理，再接回 `first.next` 上。

```python
def swap_pairs_recursive(head):
    if head is None or head.next is None:
        return head

    first = head
    second = head.next

    first.next = swap_pairs_recursive(second.next)
    second.next = first

    return second
```

如果面试官问“哪种更好”，回答：迭代版空间 O(1) 更优，递归版代码短但空间 O(n)，我会优先写迭代版。


面试表达：

我使用虚拟头节点加迭代。`pre` 指向已经处理完的部分的最后一个节点，每轮先把 `pre` 后面两个节点存到 `first` 和 `second`，然后按三步改指针：`first.next = second.next`、`second.next = first`、`pre.next = second`，最后把 `pre` 移到 `first`，直到后面不足两个节点为止。时间 O(n)，空间 O(1)。

## 题目 2 做题前先补：LRU 是什么

LRU 是 Least Recently Used 的缩写，意思是“最近最少使用”。它是一种缓存淘汰策略：

- 缓存容量有限，放不下新数据时，要删掉一些旧数据。
- 删谁？删**最久没有被访问过**的那个。

生活类比：手机后台最多同时保留 5 个应用，你又打开了第 6 个，系统会关掉“最久没用过”的那个。

规则只有两条：

1. 每次 `get` 或 `put` 都算“使用”了一次，这个 key 变成“最近使用”。
2. 容量满了还要插入新 key 时，淘汰“最久未使用”的那个。

先用列表把规则跑一遍，不管性能，只看逻辑。


````python
# 用列表模拟 LRU 的顺序变化：下标 0 是最久未使用，末尾是最近使用
order = []

# put(1, 1)
order.append(1)
print("put(1, 1) ->", order)

# put(2, 2)
order.append(2)
print("put(2, 2) ->", order)

# get(1)：命中，把 1 移到末尾，变成最近使用
order.remove(1)
order.append(1)
print("get(1)    ->", order)

# put(3, 3)：容量是 2，先淘汰最久未使用的 order[0]，再插入 3
order.pop(0)
order.append(3)
print("put(3, 3) ->", order)
````

## 为什么列表版本不够快

上面每一步都要在列表里查找或删除某个位置，最坏要看一遍整个列表，是 O(n)。题目要求 `get` 和 `put` 都是 O(1)，所以要换成两个工具配合：

- **哈希表 `dict`**：存放 `key -> 节点` 的映射，O(1) 找到节点。
- **双向链表**：维护使用顺序。头部是最近使用（最新），尾部是最久未使用（最旧）。

为什么必须是**双向**链表？因为淘汰时要删除尾部节点，而删除一个节点必须知道它的前一个节点。单链表找前驱要从头遍历，是 O(n)；双向链表自带 `prev`，是 O(1)。

再用一对虚拟头尾节点 `head` 和 `tail`，把空链表、只有一个节点这些边界情况统一掉，这和前面用 `dummy` 是同一个思路。

双向链表只有两个要记牢的动作：**插入头部**和**摘掉节点**。下面亲手做一遍。


````python
class DLinkedNode:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


# 建一对虚拟头尾节点
head = DLinkedNode()
tail = DLinkedNode()
head.next = tail
tail.prev = head
print("一开始是空链表，head.next 就是 tail：", head.next is tail)

# 动作一：把新节点插到头部（head 后面）
node = DLinkedNode(1, 100)
node.prev = head
node.next = head.next
head.next.prev = node
head.next = node
print("插入头部之后，head.next.value =", head.next.value)

# 动作二：把这个节点从链表里摘出来
node.prev.next = node.next
node.next.prev = node.prev
print("摘掉之后，head.next 又是 tail 了：", head.next is tail)
print("摘掉之后，tail.prev 又是 head 了：", tail.prev is head)
````

插节点的四条线，顺序要记清楚：

1. `node.prev = head`、`node.next = head.next`：先让新节点自己指好两边。
2. `head.next.prev = node`：再让原来的第一个节点回头看新节点。
3. `head.next = node`：最后让 head 指向新节点。

摘节点的两条线：

1. `node.prev.next = node.next`：我的前一个跳过我去指我的后一个。
2. `node.next.prev = node.prev`：我的后一个跳过我去指我的前一个。

**记忆方法：先接新节点自己，再改邻居；摘节点时，让前后邻居越过我互相认识。**


# 题目 2：LRU 缓存

LeetCode 146. LRU Cache

## 题目描述（改写版）

请你设计一个满足 LRU（最近最少使用）策略的缓存类 `LRUCache`：

- `LRUCache(capacity)`：用一个正整数 `capacity` 初始化缓存容量。
- `get(key)`：如果 `key` 存在，返回它的值，并且把它标记为“最近使用”；如果不存在，返回 `-1`。
- `put(key, value)`：如果 `key` 已存在，更新它的值，并标记为“最近使用”；如果不存在，插入这个键值对。如果插入后数量超过容量，淘汰**最久未使用**的那个 key。
- `get` 和 `put` 都必须在 `O(1)` 时间内完成。

## 输入

- 构造函数接收容量 `capacity`，1 到 3000 之间。
- `get`、`put` 的参数都是整数，`key` 在 0 到 10000 之间，`value` 在 0 到 100000 之间。
- 最多调用 200000 次 `get` 和 `put`。

## 输出

- `get` 命中时返回对应的值，未命中返回 `-1`。
- `put` 不需要返回值。

## 示例

容量为 2 时依次执行：

```text
put(1, 1)    缓存：{1=1}
put(2, 2)    缓存：{1=1, 2=2}
get(1)       返回 1，缓存顺序变成：2 最旧，1 最新
put(3, 3)    容量满，淘汰最久未使用的 2，插入 3
get(2)       返回 -1（已经被淘汰）
put(4, 4)    淘汰最久未使用的 1，插入 4
get(1)       返回 -1
get(3)       返回 3
get(4)       返回 4
```

## 易漏细节

- `get` 也算“使用”，命中后必须更新使用顺序。
- `put` 一个已存在的 key，是“更新”而不是“插入新节点”，容量不会变。
- 淘汰时要同时从哈希表和链表里删掉，只删一个会出错。


## 解法名称

**哈希表 + 双向链表（Hash Map + Doubly Linked List）**。

## 暴力思路

用一个列表按使用顺序保存 key：

- `get` 时先线性查找，是 O(n)。
- 命中后还要 `remove` 再 `append`，又是 O(n)。
- 淘汰时 `pop(0)`，还是 O(n)。

规则对，但性能不达标。

## 优化思路

职责拆开：

- `cache` 字典：`key -> 节点`，负责 O(1) 定位。
- 双向链表：负责“谁新谁旧”。**头部最前 = 最近使用，尾部最后 = 最久未使用**。

`get(key)`：

1. 不在 `cache` 里，返回 -1。
2. 在 `cache` 里，把节点从当前位置摘下来，插到头部，返回它的 `value`。

`put(key, value)`：

1. key 已存在：更新 `value`，把节点摘下来插到头部，容量不变。
2. key 不存在：新建节点，插到头部，写进 `cache`，数量加一。
3. 如果数量超过容量：找到尾部节点（`tail.prev`），把它摘掉，同时 `del cache[它的 key]`，数量减一。

注意第 3 步：删链表和删字典必须同时做，只做一个就会出现“链表里没有、字典里还有”的不一致。

每个操作都是常数次指针修改加一次字典查找，所以都是 O(1)。空间是 O(capacity)。


````python
class DLinkedNode:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


def list_order(head, tail):
    """从头部往尾部读，返回当前使用顺序：靠前的是最近使用。"""
    result = []
    current = head.next
    while current is not tail:
        result.append(current.key)
        current = current.next
    return result


# 手工串一条链表：head <-> 1 <-> 2 <-> 3 <-> tail
head = DLinkedNode()
tail = DLinkedNode()
head.next = tail
tail.prev = head

node1 = DLinkedNode(1, 10)
node2 = DLinkedNode(2, 20)
node3 = DLinkedNode(3, 30)

# 依次插到头部：插入顺序是 1、2、3，所以 3 在最前面（最近使用）
for node in [node1, node2, node3]:
    node.prev = head
    node.next = head.next
    head.next.prev = node
    head.next = node

print("初始顺序（左为最近使用）：", list_order(head, tail))

# 模拟 get(1)：把 1 摘下来，插到头部
node = node1
node.prev.next = node.next
node.next.prev = node.prev

node.prev = head
node.next = head.next
head.next.prev = node
head.next = node
print("get(1) 之后：", list_order(head, tail))

# 模拟 put(4)：容量是 3，先把新节点插到头部
node4 = DLinkedNode(4, 40)
node4.prev = head
node4.next = head.next
head.next.prev = node4
head.next = node4
print("插入 4 之后：", list_order(head, tail))

# 再淘汰尾部（最久未使用）的节点
last = tail.prev
print("要淘汰的尾部节点是：", last.key)
last.prev.next = last.next
last.next.prev = last.prev
print("淘汰之后：", list_order(head, tail))
````

## 参考答案：LRU 缓存

先自己写完再看。重点对照三处：

1. `add_to_head` 和 `remove_node` 的指针顺序。
2. `put` 里“已存在”和“不存在”两条分支的区别。
3. 淘汰时是不是同时删了链表节点和字典项。


````python
class DLinkedNode:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}
        self.head = DLinkedNode()
        self.tail = DLinkedNode()
        self.head.next = self.tail
        self.tail.prev = self.head
        self.size = 0

    def get(self, key):
        if key not in self.cache:
            return -1
        node = self.cache[key]
        self.move_to_head(node)
        return node.value

    def put(self, key, value):
        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self.move_to_head(node)
        else:
            node = DLinkedNode(key, value)
            self.cache[key] = node
            self.add_to_head(node)
            self.size = self.size + 1
            if self.size > self.capacity:
                last = self.tail.prev
                self.remove_node(last)
                del self.cache[last.key]
                self.size = self.size - 1

    def add_to_head(self, node):
        node.prev = self.head
        node.next = self.head.next
        self.head.next.prev = node
        self.head.next = node

    def remove_node(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev

    def move_to_head(self, node):
        self.remove_node(node)
        self.add_to_head(node)


cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))    # 期望 1
cache.put(3, 3)
print(cache.get(2))    # 期望 -1
cache.put(4, 4)
print(cache.get(1))    # 期望 -1
print(cache.get(3))    # 期望 3
print(cache.get(4))    # 期望 4
````

面试表达：

我用哈希表加双向链表。哈希表存 key 到节点的映射，保证 O(1) 定位；双向链表维护使用顺序，头部是最近使用，尾部是最久未使用。为了省掉空链表和头尾的边界判断，我加了一对虚拟头尾节点。`get` 命中后把节点从原位置摘下来插到头部；`put` 时如果 key 已存在就更新值并移到头部，不存在就新建节点插到头部，然后如果超出容量，把尾部的前一个节点摘掉，同时从哈希表里删除。每个操作都是常数次指针修改加一次字典操作，所以时间 O(1)，空间 O(capacity)。

常见追问：

- **为什么不用单链表？** 淘汰要删尾节点，单链表找前驱是 O(n)，双向链表有 `prev` 才能 O(1)。
- **能不能用 Python 的 `OrderedDict`？** 可以，它内部就是哈希表加双向链表，`move_to_end` 和 `popitem(last=False)` 正好对应这两个动作。但面试一般希望手写，我会按手写版回答。
- **为什么节点里要存 key？** 因为淘汰时是从链表拿到节点的，只有节点里存了 key，才能反过来 `del cache[key]`。

## 这一题的节点类和建表函数

`ListNode` 只有 `val` 和 `next`，表达不了 `random` 指针，所以这一题必须新建节点类。为了和前面的 `ListNode` 区分，这里叫 `RandomNode`（LeetCode 官方类名是 `Node`，字段一样）。

建表仍然沿用 `dummy + tail + for` 逐个创建节点的方法，只在最后多一步：按 `random_indices` 把 `random` 指针连上。旧方法不够的地方只有“多一个字段”，所以只做这一点最小扩展。

- `build_random_list(values, random_indices)`：`values` 是节点值列表，`random_indices` 是每个节点 `random` 指向的下标，`None` 表示指向空。
- `random_list_values(head)`：打印成 `[(值, random 指向的值或 None), ...]`，方便对比原链表和复制结果。


````python
class RandomNode:
    def __init__(self, val=0, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random


def build_random_list(values, random_indices):
    """创建带 random 指针的链表。

    values 是节点值列表；
    random_indices 是每个节点 random 指向的下标，用 None 表示指向空。
    建表仍然用 dummy + tail + for 逐个创建节点，只在最后多连一遍 random。
    """
    if len(values) == 0:
        return None

    nodes = []
    dummy = RandomNode()
    tail = dummy
    for value in values:
        tail.next = RandomNode(value)
        tail = tail.next
        nodes.append(tail)

    for i in range(len(nodes)):
        index = random_indices[i]
        if index is not None:
            nodes[i].random = nodes[index]

    return dummy.next


def random_list_values(head):
    """输出 [(值, random 指向的值或 None), ...]，方便观察结果。"""
    result = []
    current = head
    while current is not None:
        if current.random is None:
            random_value = None
        else:
            random_value = current.random.val
        result.append((current.val, random_value))
        current = current.next
    return result


head = build_random_list([7, 13, 11, 10, 1], [None, 0, 4, 2, 0])
print(random_list_values(head))
````

# 题目 3：随机链表的复制

LeetCode 138. Copy List with Random Pointer

## 题目描述（改写版）

链表的每个节点除了 `next` 指针，还多了一个 `random` 指针，它可以指向链表中的任意节点，也可以指向 `None`。

请你**深拷贝**这条链表：返回一条全新的链表，新链表里每个节点的值、`next` 走向和 `random` 指向，都要和原来一一对应。

关键要求：新链表里的 `next` 和 `random` 必须指向**新链表自己的节点**，不能指向原链表的节点。

## 输入

- `head`：带 `random` 指针的链表头节点，节点个数在 0 到 1000 之间，节点值在 -10000 到 10000 之间。
- `random` 要么指向链表中的某个节点，要么是 `None`。

## 输出

返回复制出来的新链表头节点。

## 示例

链表 `7 -> 13 -> 11 -> 10 -> 1`，其中：

- 7 的 `random` 是 `None`
- 13 的 `random` 指向 7
- 11 的 `random` 指向 1
- 10 的 `random` 指向 11
- 1 的 `random` 指向 7

复制后，新链表的形状要和上面完全一致，而且新节点的 `val` 可以相等，但节点对象必须是新的。

## 易漏细节

- 空链表要返回 `None`。
- 复制 `random` 的时候，目标新节点可能还没有创建出来，这就是需要分两遍遍历的原因。
- 判断是否复制成功，不要只看值，要检查 `new_head is not head`、`new_head.next is not head.next` 这类对象身份。


## 解法名称

**哈希表映射（Hash Map Mapping）**，也叫“两次遍历 + 旧新节点映射”。

## 暴力思路

先照着 `next` 复制一条新链表，再对每个节点处理 `random`：从原链表头开始数，看 `random` 指向第几个节点，然后走到新链表的对应位置。

每个节点的 `random` 都要重新走一遍链表，总共 O(n²)，太慢。

## 优化思路

慢的原因：复制 `random` 时，目标新节点可能还没有创建出来，所以只能靠位置去数。

解决：先用字典建立“旧节点 -> 新节点”的映射，之后所有定位都是 O(1)。

分两遍遍历：

1. **第一遍**：只复制值。每遇到一个旧节点，就造一个新节点，写进 `mapping[旧节点] = 新节点`。
2. **第二遍**：补指针。`新节点的 next = mapping[旧节点的 next]`，`新节点的 random = mapping[旧节点的 random]`。遇到 `None` 就跳过。

第二遍能直接连 `random`，是因为第一遍结束时，所有新节点都已经创建好了。

时间 O(n)，空间 O(n)（字典和复制出来的节点）。


````python
class RandomNode:
    def __init__(self, val=0, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random


def build_random_list(values, random_indices):
    """创建带 random 指针的链表。

    values 是节点值列表；
    random_indices 是每个节点 random 指向的下标，用 None 表示指向空。
    建表仍然用 dummy + tail + for 逐个创建节点，只在最后多连一遍 random。
    """
    if len(values) == 0:
        return None

    nodes = []
    dummy = RandomNode()
    tail = dummy
    for value in values:
        tail.next = RandomNode(value)
        tail = tail.next
        nodes.append(tail)

    for i in range(len(nodes)):
        index = random_indices[i]
        if index is not None:
            nodes[i].random = nodes[index]

    return dummy.next


def random_list_values(head):
    """输出 [(值, random 指向的值或 None), ...]，方便观察结果。"""
    result = []
    current = head
    while current is not None:
        if current.random is None:
            random_value = None
        else:
            random_value = current.random.val
        result.append((current.val, random_value))
        current = current.next
    return result


def copy_random_list_trace(head):
    """带打印版本：看清两遍遍历分别做了什么。"""
    if head is None:
        return None

    mapping = {}
    current = head
    print("第一遍：只复制值，建立 旧节点 -> 新节点 的映射")
    while current is not None:
        mapping[current] = RandomNode(current.val)
        print("    旧节点", current.val, "-> 新节点", mapping[current].val)
        current = current.next

    current = head
    print("第二遍：补 next 和 random")
    while current is not None:
        if current.next is not None:
            mapping[current].next = mapping[current.next]
        if current.random is not None:
            mapping[current].random = mapping[current.random]
        print("    处理完值为", current.val, "的节点")
        current = current.next

    return mapping[head]


head = build_random_list([7, 13, 11], [None, 0, 0])
new_head = copy_random_list_trace(head)
print("原链表：", random_list_values(head))
print("新链表：", random_list_values(new_head))
print("是不是全新的头节点：", new_head is not head)
````

````python
class RandomNode:
    def __init__(self, val=0, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random


def build_random_list(values, random_indices):
    """创建带 random 指针的链表。

    values 是节点值列表；
    random_indices 是每个节点 random 指向的下标，用 None 表示指向空。
    建表仍然用 dummy + tail + for 逐个创建节点，只在最后多连一遍 random。
    """
    if len(values) == 0:
        return None

    nodes = []
    dummy = RandomNode()
    tail = dummy
    for value in values:
        tail.next = RandomNode(value)
        tail = tail.next
        nodes.append(tail)

    for i in range(len(nodes)):
        index = random_indices[i]
        if index is not None:
            nodes[i].random = nodes[index]

    return dummy.next


def random_list_values(head):
    """输出 [(值, random 指向的值或 None), ...]，方便观察结果。"""
    result = []
    current = head
    while current is not None:
        if current.random is None:
            random_value = None
        else:
            random_value = current.random.val
        result.append((current.val, random_value))
        current = current.next
    return result


def copy_random_list_answer(head):
    if head is None:
        return None

    mapping = {}
    current = head
    while current is not None:
        mapping[current] = RandomNode(current.val)
        current = current.next

    current = head
    while current is not None:
        if current.next is not None:
            mapping[current].next = mapping[current.next]
        if current.random is not None:
            mapping[current].random = mapping[current.random]
        current = current.next

    return mapping[head]


head = build_random_list([7, 13, 11, 10, 1], [None, 0, 4, 2, 0])
new_head = copy_random_list_answer(head)
print(random_list_values(new_head))
print("是深拷贝：", new_head is not head and new_head.next is not head.next)
print("空链表返回 None：", copy_random_list_answer(build_random_list([], [])))
````

## 第二层答案：不用字典的 O(1) 空间做法

面试里如果被追问“能不能不用额外的字典”，可以答这个思路，分三遍：

1. **第一遍**：在每个原节点后面插入它的副本，形成 `A -> A' -> B -> B' -> ...`。
2. **第二遍**：`A'.random = A.random.next`。因为副本就贴在原节点后面，原节点的 `random` 的下一个一定是它的副本。
3. **第三遍**：把两条链表拆开，奇数位置是原链表，偶数位置是副本链表。

时间 O(n)，额外空间 O(1)。代码比哈希表版长，第一遍先掌握哈希表版。


````python
class RandomNode:
    def __init__(self, val=0, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random


def build_random_list(values, random_indices):
    """创建带 random 指针的链表。

    values 是节点值列表；
    random_indices 是每个节点 random 指向的下标，用 None 表示指向空。
    建表仍然用 dummy + tail + for 逐个创建节点，只在最后多连一遍 random。
    """
    if len(values) == 0:
        return None

    nodes = []
    dummy = RandomNode()
    tail = dummy
    for value in values:
        tail.next = RandomNode(value)
        tail = tail.next
        nodes.append(tail)

    for i in range(len(nodes)):
        index = random_indices[i]
        if index is not None:
            nodes[i].random = nodes[index]

    return dummy.next


def random_list_values(head):
    """输出 [(值, random 指向的值或 None), ...]，方便观察结果。"""
    result = []
    current = head
    while current is not None:
        if current.random is None:
            random_value = None
        else:
            random_value = current.random.val
        result.append((current.val, random_value))
        current = current.next
    return result


def copy_random_list_second(head):
    if head is None:
        return None

    # 第一遍：在每个原节点后面插入副本
    current = head
    while current is not None:
        new_node = RandomNode(current.val)
        new_node.next = current.next
        current.next = new_node
        current = new_node.next

    # 第二遍：设置副本的 random
    current = head
    while current is not None:
        if current.random is not None:
            current.next.random = current.random.next
        current = current.next.next

    # 第三遍：拆成两条链表
    dummy = RandomNode()
    tail = dummy
    current = head
    while current is not None:
        tail.next = current.next
        tail = tail.next
        current.next = current.next.next
        current = current.next

    return dummy.next


head = build_random_list([7, 13, 11, 10, 1], [None, 0, 4, 2, 0])
new_head = copy_random_list_second(head)
print(random_list_values(new_head))
print("原链表没有被破坏：", random_list_values(head))
````

面试表达：

我用哈希表做旧节点到新节点的映射，分两遍遍历。第一遍只复制节点的值，同时建立映射；第二遍再补 `next` 和 `random`，因为这时所有新节点都已经创建好了，查表就能直接拿到目标节点。时间 O(n)，空间 O(n)。

如果要求额外空间 O(1)，我会用“拼接再拆分”：先把每个节点的副本插在它自己后面，这样 `原节点.random.next` 就是副本的 `random`，最后再把两条链表拆开。

常见追问：**为什么不能用节点的值做字典的键？** 因为值可能重复，重复的值会互相覆盖，必须用节点对象本身做键。


## 题目 4 做题前先补：归并排序和快慢指针找中点

**归并排序**分三步：把序列分成两半，各自排好序，再合并两个有序序列。

它特别适合链表：数组归并需要额外空间来拷贝，而链表合并只需要改指针（Day 06 已经学过“合并两个有序链表”）。

要在链表上用归并排序，先要解决“找到中点并断开”。

**快慢指针找中点**：慢指针每次走 1 步，快指针每次走 2 步。快指针走到头时，慢指针正好在中间。

有两个细节必须记牢：

1. 快指针从 `head.next` 开始，不是 `head`。
2. 找到中点后要执行 `slow.next = None` 把链表断开。

下面的格子跑一遍，看清这两点为什么重要。


````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


head = build_list([4, 2, 1, 3])
print("原链表：", list_values(head))

slow = head
fast = head.next
while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next

print("slow 停在：", slow.val)
mid = slow.next
slow.next = None
print("断开后左半段：", list_values(head))
print("断开后右半段：", list_values(mid))

# 反面例子：如果快指针从 head 开始，两个节点时会出错
head2 = build_list([1, 2])
slow2 = head2
fast2 = head2
while fast2 is not None and fast2.next is not None:
    slow2 = slow2.next
    fast2 = fast2.next.next
print("从 head 出发时，slow 停在：", slow2.val, "，它的下一个是：", slow2.next)
````

# 题目 4：排序链表

LeetCode 148. Sort List

## 题目描述（改写版）

给你一条单链表的头节点 `head`，请把它按节点值**从小到大**排好序，返回排序后的头节点。

进阶要求：时间复杂度 `O(n log n)`，额外空间尽量少。

## 输入

- `head`：单链表头节点，节点个数在 0 到 50000 之间，节点值在 -100000 到 100000 之间。

## 输出

返回排序后的链表头节点。

## 示例

示例 1：`4 -> 2 -> 1 -> 3`，结果是 `1 -> 2 -> 3 -> 4`。

示例 2：`-1 -> 5 -> 3 -> 4 -> 0`，结果是 `-1 -> 0 -> 3 -> 4 -> 5`。

示例 3：空链表，返回空。

## 易漏细节

- 只有 0 个或 1 个节点时，直接返回，这就是递归的终止条件。
- 用快慢指针找中点后，必须把链表**断开**（`slow.next = None`），否则递归不会结束。
- 快指针要从 `head.next` 开始，而不是 `head`，否则偶数个节点时中点会偏后，导致某一段一直是空。


## 解法名称

**自顶向下归并排序（Top-Down Merge Sort）**。

## 暴力思路

把所有值读进列表，用 `sorted` 排好，再写回节点。

能过题，但改的是值不是节点；而且面试官通常希望看到“链表自己的排序”，所以只作为兜底。

## 优化思路

三步走：

1. **分**：用快慢指针找中点，`slow.next = None` 断开，得到左右两段。
2. **治**：递归排序左右两段。
3. **合**：用 Day 06 的“合并两个有序链表”把两段接起来。

递归的终止条件：`head is None or head.next is None`，也就是 0 个或 1 个节点时天然有序，直接返回。

复杂度：每次把链表分成两半，一共分 log n 层，每层合并要走一遍 n 个节点，所以时间 O(n log n)；空间是递归调用栈，O(log n)。

## 画图跟踪

以 `4 -> 2 -> 1 -> 3` 为例：

```text
第一层：找中点断开      [4, 2] 和 [1, 3]
第二层：继续断开        [4] [2]  和  [1] [3]
第三层：合并            [2, 4]    和  [1, 3]
第四层：合并            [1, 2, 3, 4]
```


````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def merge_two_lists(list1, list2):
    """合并两条有序链表，来自 Day 06 的写法。"""
    dummy = ListNode()
    tail = dummy

    while list1 is not None and list2 is not None:
        if list1.val < list2.val:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next

    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return dummy.next


def sort_list_trace(head, depth=0):
    """带打印版本：看清“分”和“合”的层次。"""
    space = "    " * depth
    if head is None or head.next is None:
        if head is not None:
            print(space + "只有", head.val, "，天然有序")
        return head

    print(space + "当前要排序：", list_values(head))

    slow = head
    fast = head.next
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    mid = slow.next
    slow.next = None
    print(space + "分成：", list_values(head), "和", list_values(mid))

    left = sort_list_trace(head, depth + 1)
    right = sort_list_trace(mid, depth + 1)
    merged = merge_two_lists(left, right)
    print(space + "合并成：", list_values(merged))
    return merged


head = build_list([4, 2, 1, 3])
result = sort_list_trace(head)
print("最终结果：", list_values(result))
````
