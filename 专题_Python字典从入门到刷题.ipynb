# Python 字典：从入门到刷题

字典 `dict` 是 Hot 100 中最重要的 Python 工具之一。

这份专题不要求一次背完所有方法，而是解决三个问题：

1. 字典中的“键”和“值”到底是什么？
2. `[]`、`get()`、`in`、`items()` 分别什么时候用？
3. 刷题时怎样想到“这里应该使用字典”？

学习顺序：

```text
创建和查询 -> 新增和修改 -> 遍历和删除
          -> 计数 -> 分组 -> 下标映射
          -> 两数之和、前缀和等题目
```

建议分两次学习，每次 40～60 分钟。

## 1. 字典到底是什么？

字典保存的是**键和值的对应关系**：

```python
student = {
    "name": "小明",
    "age": 22,
    "score": 95,
}
```

可以把它理解成通讯录：

```text
姓名 -> 电话号码
```

在 Python 字典中：

```text
key（键） -> value（值）
```

例如 `"name"` 是键，`"小明"` 是它对应的值。

专业名称：**映射（Mapping）**、**键值对（Key-Value Pair）**、**哈希表（Hash Table）**。

````python
student = {
    "name": "小明",
    "age": 22,
    "score": 95,
}

print(student)
print("姓名：", student["name"])
print("分数：", student["score"])

assert student["age"] == 22
````

## 2. 字典与列表有什么区别？

列表主要通过位置查找：

```python
scores = [90, 95, 88]
scores[1]  # 按下标查找
```

字典主要通过键查找：

```python
scores = {"小明": 90, "小红": 95, "小刚": 88}
scores["小红"]  # 按键查找
```

| 需求 | 更适合 |
|---|---|
| 按顺序保存一组数据 | 列表 |
| 按下标访问第几个元素 | 列表 |
| 保存“谁对应什么” | 字典 |
| 快速查找某个键是否出现 | 字典 |
| 保存某个数字对应的下标 | 字典 |

刷题时看到“对应关系”，要主动想到字典。

## 3. 键和值有什么要求？

### 键必须可哈希

常见可作为键的类型：

- 整数 `int`
- 字符串 `str`
- 元组 `tuple`（内部元素也要可哈希）
- 布尔值 `bool`

列表、字典、集合通常不能作为字典键：

```python
bad = {[1, 2]: "value"}  # TypeError: unhashable type: 'list'
```

### 值几乎可以是任意对象

```python
data = {
    "count": 3,
    "positions": [0, 2, 5],
    "info": {"name": "小明"},
}
```

刷题中，值常保存次数、下标、列表或另一个字典。

````python
valid_keys = {
    10: "整数键",
    "name": "字符串键",
    (1, 2): "元组键",
}

print(valid_keys[10])
print(valid_keys[(1, 2)])

assert valid_keys["name"] == "字符串键"
````

# 第一部分：创建与查询

## 4. 创建字典

### 直接创建

```python
scores = {"小明": 90, "小红": 95}
```

### 创建空字典

```python
scores = {}
scores = dict()
```

### 从键值对创建

```python
pairs = [("a", 1), ("b", 2)]
mapping = dict(pairs)
```

### 根据两个序列创建

```python
names = ["小明", "小红"]
scores = [90, 95]
mapping = dict(zip(names, scores))
```

````python
pairs = [("a", 1), ("b", 2)]
mapping_from_pairs = dict(pairs)

names = ["小明", "小红"]
scores = [90, 95]
score_map = dict(zip(names, scores))

print(mapping_from_pairs)
print(score_map)

assert score_map["小红"] == 95
````

## 5. 使用 `[]` 查询

```python
scores = {"小明": 90}
scores["小明"]  # 90
```

如果键一定存在，使用 `[]` 最直接。

但键不存在时会触发 `KeyError`：

```python
scores["小红"]  # KeyError
```

因此在使用 `[]` 前，需要知道：

- 题目是否保证键存在？
- 前面是否已经用 `in` 检查？
- 是否更适合使用 `get()`？

````python
scores = {"小明": 90, "小红": 95}

print(scores["小明"])

try:
    print(scores["小刚"])
except KeyError as error:
    print("键不存在：", error)
````

## 6. 使用 `get()` 安全查询

```python
scores.get("小明")      # 90
scores.get("小刚")      # None
scores.get("小刚", 0)   # 0
```

`get(key, default)` 的规则：

- 键存在：返回对应值；
- 键不存在：返回默认值；
- 没写默认值：默认返回 `None`。

计数时最常见：

```python
counts[x] = counts.get(x, 0) + 1
```

这句话翻译成：

> 先读取 `x` 原来的次数；如果还没有，就当作 0；然后加 1。

````python
scores = {"小明": 90}

print(scores.get("小明"))
print(scores.get("小刚"))
print(scores.get("小刚", 0))

assert scores.get("小明", 0) == 90
assert scores.get("小刚", 0) == 0
````

## 7. `[]` 与 `get()` 怎么选？

| 场景 | 推荐 |
|---|---|
| 键必须存在，不存在说明程序有问题 | `d[key]` |
| 键可能不存在，希望提供默认值 | `d.get(key, default)` |
| 需要判断键是否出现过 | `key in d` |
| 计数 | `d.get(key, 0)` |

不要无脑全部使用 `get()`。

例如学生数据必须包含姓名：

```python
name = student["name"]
```

如果缺少姓名，直接报错可能比默默得到 `None` 更容易发现数据问题。

## 8. `in` 判断的是键

```python
scores = {"小明": 90, "小红": 95}

"小明" in scores  # True
90 in scores       # False
```

`in` 默认检查字典的键，不检查值。

检查值：

```python
90 in scores.values()
```

刷题中最常用：

```python
if need in seen:
    ...
```

它在问：“我需要的这个键以前出现过吗？”

````python
scores = {"小明": 90, "小红": 95}

print("小明" in scores)
print(90 in scores)
print(90 in scores.values())

assert "小明" in scores
assert 90 not in scores
````

# 第二部分：新增、修改与删除

## 9. 新增和修改使用同一种写法

```python
d[key] = value
```

如果键不存在，就是新增：

```python
scores["小刚"] = 88
```

如果键已经存在，就是修改：

```python
scores["小明"] = 100
```

一个字典中同一个键只能保留一个值。再次赋值会覆盖旧值。

````python
scores = {"小明": 90}

scores["小红"] = 95
scores["小明"] = 100

print(scores)
assert scores == {"小明": 100, "小红": 95}
````

## 10. `update()`：一次更新多个键

```python
scores.update({"小明": 100, "小刚": 88})
```

也可以使用关键字形式：

```python
config.update(timeout=30, debug=True)
```

如果键已经存在，`update()` 会覆盖旧值。

注意与 `Counter.update()` 的区别：普通字典的 `update()` 是覆盖；`Counter.update()` 是累加计数。

````python
scores = {"小明": 90, "小红": 95}
scores.update({"小明": 100, "小刚": 88})

print(scores)
assert scores["小明"] == 100
assert scores["小刚"] == 88
````

## 11. `setdefault()`：不存在时才创建

```python
d.setdefault(key, default)
```

- 键存在：返回原来的值，不修改；
- 键不存在：写入默认值，并返回它。

分组时常见：

```python
groups = {}
groups.setdefault(first_letter, []).append(word)
```

它等价于：

```python
if first_letter not in groups:
    groups[first_letter] = []
groups[first_letter].append(word)
```

入门阶段两种都可以。展开写更容易理解，`setdefault` 更紧凑。

````python
words = ["apple", "ant", "banana", "book"]
groups = {}

for word in words:
    first_letter = word[0]
    groups.setdefault(first_letter, []).append(word)

print(groups)
assert groups == {"a": ["apple", "ant"], "b": ["banana", "book"]}
````

## 12. 删除字典元素

### `pop`

```python
score = scores.pop("小明")
score = scores.pop("不存在", 0)
```

删除键并返回对应值。提供默认值后，键不存在也不会报错。

### `del`

```python
del scores["小明"]
```

键不存在会报错。

### `popitem`

```python
key, value = scores.popitem()
```

现代 Python 中删除并返回最后插入的键值对。

### `clear`

删除所有键值对：`scores.clear()`。

````python
scores = {"小明": 90, "小红": 95, "小刚": 88}

removed = scores.pop("小红")
missing = scores.pop("不存在", 0)

print("删除的分数：", removed)
print("不存在时：", missing)
print("剩余字典：", scores)

assert removed == 95
assert missing == 0
````

# 第三部分：遍历字典

## 13. 遍历键

下面两种写法等价：

```python
for key in d:
    print(key)

for key in d.keys():
    print(key)
```

通常直接写 `for key in d` 更简洁。

`.keys()` 返回的是动态视图，不是普通列表。如果确实需要列表：

```python
keys = list(d.keys())
```

````python
scores = {"小明": 90, "小红": 95}

for name in scores:
    print(name)

print(list(scores.keys()))
assert list(scores.keys()) == ["小明", "小红"]
````

## 14. 遍历值和键值对

遍历值：

```python
for value in d.values():
    print(value)
```

同时获得键和值：

```python
for key, value in d.items():
    print(key, value)
```

`.items()` 返回的每一项可以看作 `(key, value)` 二元组，所以能够直接解包。

````python
scores = {"小明": 90, "小红": 95, "小刚": 88}

total = 0
for name, score in scores.items():
    print(name, score)
    total += score

print("平均分：", total / len(scores))
assert total == 273
````

## 15. 遍历时不要改变字典大小

危险写法：

```python
for key in d:
    if should_delete(key):
        del d[key]
```

这会在遍历过程中改变字典大小，通常触发运行时错误。

安全方式之一：遍历键的副本。

```python
for key in list(d.keys()):
    if should_delete(key):
        del d[key]
```

或者使用字典推导式创建新字典：

```python
filtered = {k: v for k, v in d.items() if keep(k, v)}
```

````python
scores = {"小明": 90, "小红": 55, "小刚": 88}

passed = {
    name: score
    for name, score in scores.items()
    if score >= 60
}

print(passed)
assert passed == {"小明": 90, "小刚": 88}
````

## 16. 字典保留插入顺序，但不会自动排序

现代 Python 字典会保留键的插入顺序：

```python
d = {}
d["b"] = 2
d["a"] = 1
```

遍历顺序是 `b`、`a`，不会自动按字母排序。

按键排序遍历：

```python
for key in sorted(d):
    print(key, d[key])
```

按值排序：

```python
sorted(d.items(), key=lambda item: item[1])
```

排序结果是键值对列表，不再是字典。

````python
scores = {"小明": 90, "小红": 95, "小刚": 88}

by_score = sorted(scores.items(), key=lambda item: item[1], reverse=True)
print(by_score)

assert by_score[0] == ("小红", 95)
````

# 第四部分：刷题中的四种角色

## 17. 角色一：计数器

问题特征：

- 每个元素出现多少次？
- 字符频次是否相同？
- 哪个元素出现最多？

标准写法：

```python
counts = {}

for value in nums:
    counts[value] = counts.get(value, 0) + 1
```

每轮分两步理解：

1. `counts.get(value, 0)`：取出旧次数；
2. `+ 1` 后存回 `counts[value]`。

````python
nums = [2, 1, 2, 3, 2, 1]
counts = {}

for value in nums:
    old_count = counts.get(value, 0)
    new_count = old_count + 1
    counts[value] = new_count
    print(f"value={value}, 旧次数={old_count}, 新字典={counts}")

assert counts == {2: 3, 1: 2, 3: 1}
````

## 18. 角色二：存在性记录

问题特征：

- 某个元素以前出现过吗？
- 是否有重复？
- 当前需要的值已经见过吗？

```python
seen = {}

for value in nums:
    if value in seen:
        return True
    seen[value] = True
```

如果只需要判断是否存在，不需要保存额外信息，`set` 往往更直接：

```python
seen = set()
```

字典适合“存在的同时还要保存次数、下标或状态”。

````python
def contains_duplicate_with_dict(nums):
    seen = {}
    for value in nums:
        if value in seen:
            return True
        seen[value] = True
    return False


assert contains_duplicate_with_dict([1, 2, 3, 1]) is True
assert contains_duplicate_with_dict([1, 2, 3]) is False
````

## 19. 角色三：下标映射

问题特征：

- 某个数以前出现在哪个位置？
- 找到值以后需要返回下标。

```python
position = {}

for index, value in enumerate(nums):
    position[value] = index
```

此时：

```text
数字 -> 下标
```

这正是“两数之和”哈希表解法的核心。

````python
nums = [10, 20, 30]
position = {}

for index, value in enumerate(nums):
    position[value] = index

print(position)
assert position == {10: 0, 20: 1, 30: 2}
````

## 20. 角色四：分组容器

问题特征：

- 按某个特征把元素放到同一组；
- 一个键对应多个值；
- 按首字母、类别、排序结果等分组。

普通字典写法：

```python
groups = {}

for word in words:
    key = word[0]
    if key not in groups:
        groups[key] = []
    groups[key].append(word)
```

这里字典的值不是单个数字，而是列表。

````python
words = ["apple", "ant", "banana", "book", "cat"]
groups = {}

for word in words:
    key = word[0]
    if key not in groups:
        groups[key] = []
    groups[key].append(word)

print(groups)
assert groups["a"] == ["apple", "ant"]
````

## 21. 选择普通 dict、defaultdict 还是 Counter？

| 需求 | 推荐 |
|---|---|
| 自己控制键和值的内容 | `dict` |
| 键不存在时自动创建空列表或 0 | `defaultdict` |
| 只统计元素次数 | `Counter` |
| 判断是否存在，不保存附加信息 | `set` |

三种计数方式：

```python
# dict
counts[x] = counts.get(x, 0) + 1

# defaultdict
counts[x] += 1

# Counter
counts = Counter(nums)
```

面试中三种都可以。普通字典最能展示你理解了计数逻辑。

# 第五部分：字典键的设计

## 22. 什么信息能够代表“同一组”？

分组题的关键常常不是字典语法，而是如何设计键。

例如字母异位词：

```text
eat -> 排序后 aet
tea -> 排序后 aet
ate -> 排序后 aet
```

三个单词排序结果相同，可以使用排序后的字符元组作为键：

```python
key = tuple(sorted(word))
```

为什么使用元组？因为列表不能作为字典键，元组可以。

````python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = {}

for word in words:
    key = tuple(sorted(word))
    groups.setdefault(key, []).append(word)

answer = list(groups.values())
print(answer)

assert ["eat", "tea", "ate"] in answer
assert ["tan", "nat"] in answer
````

## 23. 同一个键再次出现会覆盖旧值

```python
position = {}
position[5] = 0
position[5] = 3
```

最终只保留 `{5: 3}`。

如果需要保存同一个值的所有位置，值应该使用列表：

```python
positions = {}
positions.setdefault(5, []).append(0)
positions.setdefault(5, []).append(3)
```

结果：

```python
{5: [0, 3]}
```

设计字典前要问：一个键只对应一个值，还是可能对应多个值？

````python
nums = [5, 2, 5, 3, 5]
positions = {}

for index, value in enumerate(nums):
    positions.setdefault(value, []).append(index)

print(positions)
assert positions[5] == [0, 2, 4]
````

# 第六部分：常见高级操作

## 24. 字典推导式

```python
squares = {number: number * number for number in nums}
```

带筛选：

```python
even_squares = {
    number: number * number
    for number in nums
    if number % 2 == 0
}
```

交换键和值：

```python
reversed_map = {value: key for key, value in d.items()}
```

只有值唯一时，反转才不会发生覆盖。

````python
nums = [1, 2, 3, 4]
squares = {number: number * number for number in nums}
even_squares = {number: square for number, square in squares.items() if number % 2 == 0}

print(squares)
print(even_squares)

assert even_squares == {2: 4, 4: 16}
````

## 25. 嵌套字典

```python
students = {
    "S01": {"name": "小明", "score": 90},
    "S02": {"name": "小红", "score": 95},
}
```

读取：

```python
students["S01"]["score"]
```

嵌套太深时代码会难读。刷题时可以使用元组、列表或自定义对象简化结构。

安全读取嵌套值：

```python
students.get("S03", {}).get("score", 0)
```

但如果数据结构本应完整，直接索引并让错误暴露也可能更合理。

````python
students = {
    "S01": {"name": "小明", "score": 90},
    "S02": {"name": "小红", "score": 95},
}

print(students["S01"]["score"])
print(students.get("S03", {}).get("score", 0))

assert students["S02"]["name"] == "小红"
````

## 26. 复制字典

```python
copy_a = original.copy()
copy_b = dict(original)
```

它们是浅拷贝：最外层字典是新的，但内部可变对象仍可能共享。

```python
original = {"items": [1, 2]}
copy_a = original.copy()
copy_a["items"].append(3)
```

此时 `original["items"]` 也会变化。

嵌套结构完全独立时：

```python
import copy
deep_copy = copy.deepcopy(original)
```

````python
import copy

original = {"items": [1, 2]}
shallow = original.copy()
deep = copy.deepcopy(original)

shallow["items"].append(3)

print("原字典：", original)
print("浅拷贝：", shallow)
print("深拷贝：", deep)

assert original["items"] == [1, 2, 3]
assert deep["items"] == [1, 2]
````

# 第七部分：复杂度与面试知识

## 28. 字典操作复杂度

平均情况下：

| 操作 | 时间复杂度 |
|---|---|
| 查询 `d[key]` | `O(1)` |
| 判断 `key in d` | `O(1)` |
| 新增或修改 | `O(1)` |
| 删除 | `O(1)` |
| 遍历全部键值对 | `O(n)` |
| 复制字典 | `O(n)` |

这里的 `O(1)` 是平均复杂度。极端哈希冲突等情况下可能退化。

空间复杂度：保存 `n` 个不同键值对通常需要 `O(n)` 空间。

## 29. 为什么字典查询通常很快？

字典会对键计算哈希值，再根据哈希值定位存储位置。

可以粗略理解成：

```text
key -> hash(key) -> 存储位置 -> value
```

它不需要像列表那样从头逐个比较所有元素，因此平均查询是 `O(1)`。

面试回答：

> Python 字典底层使用哈希表。查询时通过键的哈希值定位槽位，因此平均查询、插入和删除复杂度为 `O(1)`；极端哈希冲突情况下最坏可能退化。

# 第八部分：三道典型算法题

## 30. 两数之和：先查再存

```python
def two_sum(nums, target):
    seen = {}

    for i, number in enumerate(nums):
        need = target - number

        if need in seen:
            return [seen[need], i]

        seen[number] = i
```

字典角色：**数字 → 以前出现的下标**。

为什么先查再存？

因为同一个元素不能使用两次。先检查以前的元素，再记录当前元素，可以自然避免把当前下标和自己配对。

````python
def two_sum(nums, target, verbose=False):
    seen = {}

    for i, number in enumerate(nums):
        need = target - number

        if verbose:
            print(f"i={i}, number={number}, need={need}, seen={seen}")

        if need in seen:
            return [seen[need], i]

        seen[number] = i

    return None


assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
print(two_sum([3, 3], 6, verbose=True))
````

## 32. 和为 K 的子数组：前缀和计数

```python
def subarray_sum(nums, k):
    prefix_count = {0: 1}
    prefix = 0
    answer = 0

    for number in nums:
        prefix += number
        answer += prefix_count.get(prefix - k, 0)
        prefix_count[prefix] = prefix_count.get(prefix, 0) + 1

    return answer
```

字典角色：**前缀和 → 以前出现的次数**。

这里不是只判断某个前缀和是否存在，而是需要知道出现了多少次，所以值保存次数。

````python
def subarray_sum(nums, k):
    prefix_count = {0: 1}
    prefix = 0
    answer = 0

    for number in nums:
        prefix += number
        answer += prefix_count.get(prefix - k, 0)
        prefix_count[prefix] = prefix_count.get(prefix, 0) + 1

    return answer


assert subarray_sum([1, 1, 1], 2) == 2
assert subarray_sum([1, -1, 0], 0) == 3
print(subarray_sum([1, 2, 3], 3))
````

# 第九部分：常见错误

## 33. 十个高频误区

1. 把 `=` 当成比较，比较相等应使用 `==`。
2. 访问可能不存在的键却直接写 `d[key]`。
3. 以为 `x in d` 检查的是值，实际上检查键。
4. 计数时忘记把不存在的键当作 0。
5. 同一个键重复赋值，旧值被覆盖却没有意识到。
6. 想保存多个结果，值却没有设计成列表。
7. 使用列表作为字典键，触发不可哈希错误。
8. 遍历字典时同时增加或删除键。
9. 使用浅拷贝后修改内部列表，以为原字典不会变化。
10. 两数之和先存再查，可能错误地重复使用当前元素。

## 34. 调试字典的方式

不要只在循环结束后打印字典。应该观察每一轮：

```python
for value in nums:
    old = counts.get(value, 0)
    counts[value] = old + 1

    print(
        "当前值=", value,
        "旧次数=", old,
        "更新后字典=", counts,
    )
```

刷题调试时重点看：

1. 当前正在查询哪个键？
2. 查询发生在存储之前还是之后？
3. 值保存的是次数、下标还是列表？
4. 当前键是否可能覆盖旧值？

````python
def trace_counts(nums):
    counts = {}
    for value in nums:
        old = counts.get(value, 0)
        counts[value] = old + 1
        print(f"value={value}, old={old}, counts={counts}")
    return counts


assert trace_counts([1, 2, 1]) == {1: 2, 2: 1}
````

## 专项练习参考答案

先独立完成，再运行下一单元。

````python
def frequency_map_answer(nums):
    counts = {}
    for value in nums:
        counts[value] = counts.get(value, 0) + 1
    return counts


def first_duplicate_answer(nums):
    seen = {}
    for value in nums:
        if value in seen:
            return value
        seen[value] = True
    return None


def all_positions_answer(nums):
    positions = {}
    for index, value in enumerate(nums):
        positions.setdefault(value, []).append(index)
    return positions


def top_student_answer(scores):
    return max(scores, key=scores.get) if scores else None


def merge_counts_answer(left, right):
    result = left.copy()
    for key, count in right.items():
        result[key] = result.get(key, 0) + count
    return result


def two_sum_answer(nums, target):
    seen = {}
    for index, value in enumerate(nums):
        need = target - value
        if need in seen:
            return [seen[need], index]
        seen[value] = index
    return None


assert frequency_map_answer([1, 2, 1, 3, 2, 1]) == {1: 3, 2: 2, 3: 1}
assert first_duplicate_answer([3, 1, 2, 1, 3]) == 1
assert first_duplicate_answer([1, 2, 3]) is None
assert all_positions_answer([5, 2, 5, 3, 5]) == {5: [0, 2, 4], 2: [1], 3: [3]}
assert top_student_answer({"小明": 90, "小红": 95}) == "小红"
assert top_student_answer({}) is None
assert merge_counts_answer({"a": 2}, {"a": 1, "b": 3}) == {"a": 3, "b": 3}
assert two_sum_answer([2, 7, 11, 15], 9) == [0, 1]
print("字典专项练习参考答案测试通过")
````

## 35. 如何说明为什么使用字典？

> 题目需要快速查询某个值以前是否出现，并且还要保存它对应的下标，因此我使用哈希表建立“值到下标”的映射。遍历数组时，每次哈希查询和写入平均是 `O(1)`，总时间复杂度为 `O(n)`；哈希表最多保存 `n` 个元素，所以空间复杂度为 `O(n)`。
