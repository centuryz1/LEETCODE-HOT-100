# 双指针专题：为什么是 `while left < right`？

## 学习目标

学完后，你应该能够：

1. 从题目需求推导出 `left` 和 `right` 的职责。
2. 解释为什么这里使用 `while`，而不是 `if`。
3. 解释为什么条件是 `left < right`，而不是 `left <= right`。
4. 根据总和与目标值的关系正确移动指针。
5. 理解三数之和中的三处去重。
6. 发现“函数传入 `target`，代码却一直和 0 比较”的问题。

专业名称：**排序 + 双指针**，英文是 **Sorting + Two Pointers**。

## 先记住一句话

```python
while left < right:
```

可以直接翻译成：

> 只要左边界还在右边界左侧，就还有两个不同位置可以组合，继续搜索。

这里包含两个信息：

- `while`：需要反复尝试多组左右位置。
- `left < right`：两个指针必须指向不同下标。

# 第一部分：三数之和怎样变成两数之和？

原问题要求寻找：

```text
nums[i] + nums[left] + nums[right] = target
```

先用 `for` 固定第一个数 `nums[i]`，那么问题就变成：

```text
在 i 后面的范围中，寻找两个数
使 nums[left] + nums[right] = target - nums[i]
```

所以代码分工是：

```python
for i in range(n - 2):       # 固定第一个数
    left = i + 1             # 剩余范围最左边
    right = n - 1            # 剩余范围最右边

    while left < right:      # 在剩余范围中反复寻找两个数
        ...
```

`for` 负责换第一个数，`while` 负责寻找另外两个数。

# 第二部分：left 和 right 分别代表什么？

数组排序后：

```text
下标：  0    1    2    3    4    5
数值： -4   -1   -1    0    1    2
        i    left                 right
```

固定 `i` 后：

- `left` 从 `i + 1` 开始，指向剩余范围最小的候选数。
- `right` 从末尾开始，指向剩余范围最大的候选数。
- `[left, right]` 表示还没有搜索完的候选范围。

每比较一次，至少移动一个指针，候选范围就会缩小。

# 第三部分：为什么使用 while？

第一次计算的左右组合不一定就是答案。

例如固定 `-1`，剩余数字是 `[0, 1, 2]`，目标是 `0`：

```text
第一次：-1 + 0 + 2 = 1，太大
移动 right
第二次：-1 + 0 + 1 = 0，找到
```

这里至少尝试了两组位置。实际题目中可能需要尝试更多组，而且次数事先不知道。

- `if left < right` 只会检查一组。
- `while left < right` 会不断检查，直到没有两个不同位置。

对应控制流口诀：

> `if` 决定做不做一次，`while` 决定还要不要继续做。

````python
nums = [-1, 0, 1, 2]
target = 0
i = 0
left = i + 1
right = len(nums) - 1

while left < right:
    total = nums[i] + nums[left] + nums[right]
    print(f'i={i}, left={left}, right={right}, 三个数={nums[i], nums[left], nums[right]}, total={total}')

    if total < target:
        left += 1
    elif total > target:
        right -= 1
    else:
        print('找到答案')
        left += 1
        right -= 1
````

# 第四部分：为什么是 `<`，不是 `<=`？

题目要求选择三个不同下标。

`i` 已经固定，`left` 和 `right` 还必须指向两个不同位置。

当：

```python
left == right
```

两个指针指向同一个元素。如果继续计算，就相当于把同一个下标使用两次。

因此合法状态必须是：

```python
left < right
```

一旦 `left >= right`：

- 相等：只剩一个位置，凑不出两个不同元素。
- 交叉：候选范围已经搜索完。

所以循环结束。

````python
nums = [1, 2, 3]
left = 1
right = 1

print('left == right：', left == right)
print('两个指针指向的下标：', left, right)
print('实际上只有一个位置，不能把 nums[1] 使用两次。')
````

# 第五部分：为什么总和小就移动 left？

双指针能这样移动，是因为数组已经排序。

```python
nums.sort()
```

排序后，越往右数值通常越大：

- `total < target`：总和太小，需要换更大的数，因此 `left += 1`。
- `total > target`：总和太大，需要换更小的数，因此 `right -= 1`。
- `total == target`：找到答案，记录后左右都移动。

这是一种利用单调性的搜索。

如果数组没有排序，移动 `left` 不一定让总和增大，移动 `right` 也不一定让总和减小，规则就失效了。

## 指针移动表

| 比较结果 | 含义 | 操作 |
|---|---|---|
| `total < target` | 总和太小 | `left += 1` |
| `total > target` | 总和太大 | `right -= 1` |
| `total == target` | 找到答案 | 记录；左右都移动 |

每一轮至少移动一个指针非常重要，否则相同状态会被反复计算，形成死循环。

# 第六部分：三数之和的三处去重

## 1. 固定数 i 去重

```python
if i > 0 and nums[i] == nums[i - 1]:
    continue
```

相同的第一个数已经处理过，再固定一次只会产生重复答案，所以跳过当前 `i`。

## 2. 找到答案后，left 去重

```python
while left < right and nums[left] == nums[left - 1]:
    left += 1
```

`left` 已经先向右移动一步，再跳过与刚才相同的值。

## 3. 找到答案后，right 去重

```python
while left < right and nums[right] == nums[right + 1]:
    right -= 1
```

`right` 已经先向左移动一步，再跳过与刚才相同的值。

这些 `while` 表示：重复值可能连续出现多个，要一直跳过。

## 总和不等于目标时必须立刻去重吗？

不必须。

当 `total < target` 时，标准写法通常只做：

```python
left += 1
```

当 `total > target` 时，通常只做：

```python
right -= 1
```

即使遇到重复值，多计算几次也不会产生重复答案；真正需要保证答案不重复的是“找到答案以后”的左右去重。

在总和不等于目标时提前跳过重复值也可以，但代码更长。初学阶段建议先掌握标准简洁写法。

# 第七部分：你代码里的 target 问题

你的函数定义是：

```python
def three_sum(nums, target):
```

说明目标值由 `target` 参数传进来。

但内部写的是：

```python
if total < 0:
    ...
elif total > 0:
    ...
```

这会忽略 `target`，始终寻找和为 0 的三元组。

如果函数支持任意目标值，应写成：

```python
if total < target:
    ...
elif total > target:
    ...
```

如果做 LeetCode 15，目标固定为 0，则可以删除 `target` 参数，明确函数只解决三数之和为 0。

# 第八部分：支持任意 target 的标准版本

````python
def three_sum_target(nums, target):
    nums = sorted(nums)
    n = len(nums)
    answer = []

    for i in range(n - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left = i + 1
        right = n - 1

        while left < right:
            total = nums[i] + nums[left] + nums[right]

            if total < target:
                left += 1
            elif total > target:
                right -= 1
            else:
                answer.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1

                while left < right and nums[left] == nums[left - 1]:
                    left += 1
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1

    return answer


assert three_sum_target([-1, 0, 1, 2, -1, -4], 0) == [[-1, -1, 2], [-1, 0, 1]]
assert three_sum_target([0, 1, 2, 3, 4], 6) == [[0, 2, 4], [1, 2, 3]]
print('任意 target 版本测试通过')
````

## LeetCode 15 的标准版本

LeetCode 15 的目标固定为 0，可以让函数只接收 `nums`：

````python
def three_sum_zero(nums):
    nums = sorted(nums)
    answer = []

    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left, right = i + 1, len(nums) - 1

        while left < right:
            total = nums[i] + nums[left] + nums[right]

            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                answer.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1

                while left < right and nums[left] == nums[left - 1]:
                    left += 1
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1

    return answer


assert three_sum_zero([-1, 0, 1, 2, -1, -4]) == [[-1, -1, 2], [-1, 0, 1]]
assert three_sum_zero([0, 0, 0, 0]) == [[0, 0, 0]]
print('LeetCode 15 版本测试通过')
````

# 第九部分：怎样自己推导 while 条件？

以后不要直接背 `while left < right`，按下面四步推导：

1. 固定一个元素后，还需要从剩余范围选择几个元素？
2. `left` 和 `right` 是否必须代表两个不同位置？
3. 第一次组合不成功时，是否还需要继续尝试？
4. 什么情况下剩余范围已经不足以组成答案？

对三数之和：

- 还需要选择两个元素。
- 两个下标必须不同。
- 需要反复尝试多组位置。
- 左右相遇或交叉时，只剩不到两个候选位置。

所以自然得到：

```python
while left < right:
```

# 第十部分：常见错误

1. **写成 `if left < right`**：只检查一组左右位置。
2. **写成 `while left <= right`**：可能重复使用同一个下标。
3. **某个分支不移动指针**：状态不变，可能死循环。
4. **数组没有排序**：无法根据总和大小决定移动方向。
5. **传入 target 却和 0 比较**：函数名义上支持任意目标，实际上只找 0。
6. **找到答案后不移动指针**：同一个答案会被反复加入。
7. **不去重**：结果中可能出现重复三元组。
8. **`for i in range(n)`**：虽然不一定报错，但最后两个位置不可能再组成三元组，写 `range(n - 2)` 更准确。

# 第十二部分：练习参考答案

````python
def pair_sum_answer(nums, target):
    left = 0
    right = len(nums) - 1

    while left < right:
        total = nums[left] + nums[right]

        if total < target:
            left += 1
        elif total > target:
            right -= 1
        else:
            return [left, right]

    return None


assert pair_sum_answer([1, 2, 4, 6, 10], 8) == [1, 3]
assert pair_sum_answer([1, 2, 4, 6, 10], 20) is None
print('两数之和练习测试通过')
````

# 最后总结

`while left < right` 不是固定咒语，而是由问题条件推导出来的：

- `while`：还需要反复尝试。
- `left < right`：还存在两个不同下标。
- `left += 1`：排序后，总和太小就换大一点的数。
- `right -= 1`：排序后，总和太大就换小一点的数。
- `left >= right`：候选范围不足两个位置，搜索结束。

最短记忆版：

> 左右还没碰头，就还有一对候选；碰头以后，只剩一个位置，不能再组成一对。
