## 今天最重要的事：把回溯模板背下来

回溯就是“**一条路走到黑，走不通就退回来换一条**”。写代码时永远是这三件事：

```python
def backtrack(路径, 选择列表):
    if 满足结束条件:
        收集答案
        return
    for 选择 in 选择列表:
        path.append(选择)          # 1. 做选择
        backtrack(新的路径, 新的选择列表)   # 2. 递归
        path.pop()                 # 3. 撤销选择
```

三条铁律：

1. **`path.append` 和 `path.pop` 必须成对出现**，中间夹着递归。这就是“撤销选择”。
2. **收集答案时要写 `path[:]`**，也就是复制一份。因为 `path` 是同一个列表，后面还会被修改；如果直接 `result.append(path)`，最后结果里全是同一个列表（而且都是空的）。
3. 结束条件写清楚：什么时候收集答案、什么时候直接返回。

下面用一个最小例子看清 `path[:]` 为什么必须复制。


````python
path = [1, 2]
result_wrong = []
result_wrong.append(path)
path.append(3)
print("不复制的结果：", result_wrong)      # 被后面的修改带跑了

path = [1, 2]
result_right = []
result_right.append(path[:])
path.append(3)
print("复制一份的结果：", result_right)    # 保存的是当时的样子
````

规律：

- **已经在 `path` 里的数字不能再选**（这就是判断条件 `if num in path: continue`）；
- 当 `path` 的长度等于数组长度时，就是一个完整排列，收集答案。

自己动手画一遍这个树，比看代码有用得多。


````python
# 用打印看清回溯的进入和退出
nums = [1, 2]
path = []


def demo():
    if len(path) == len(nums):
        print("    收集答案：", path[:])
        return
    for num in nums:
        if num in path:
            print("    ", num, "已经用过了，跳过")
            continue
        path.append(num)
        print("选择", num, "，当前 path =", path)
        demo()
        removed = path.pop()
        print("撤销", removed, "，当前 path =", path)


demo()
````

## 参考答案：全排列

```python
def permute_answer(nums):
    result = []
    path = []

    def backtrack():
        if len(path) == len(nums):
            result.append(path[:])
            return
        for num in nums:
            if num in path:
                continue
            path.append(num)
            backtrack()
            path.pop()

    backtrack()
    return result
```

面试表达：

我用回溯。`path` 记录当前已经排好的部分，`result` 收集完整答案。每层循环遍历数组里的每个数字，如果它已经在 `path` 里就跳过，保证不重复使用；否则把它加入 `path`，递归填下一个位置，递归返回后再把它弹出来，也就是撤销选择，这样回到上层时状态是干净的。当 `path` 长度等于数组长度时就得到一个完整排列，这时要复制一份再保存，因为 `path` 后面还会被修改。时间 O(n 乘 n 的阶乘)，空间 O(n)。


````python


def permute_answer(nums):
    result = []
    path = []

    def backtrack():
        if len(path) == len(nums):
            result.append(path[:])
            return
        for num in nums:
            if num in path:
                continue
            path.append(num)
            backtrack()
            path.pop()

    backtrack()
    return result


print(len(permute_answer([1, 2, 3])))          # 6
print(sorted(permute_answer([1, 2, 3])))
print(permute_answer([1]))                     # [[1]]
print(sorted(permute_answer([0, 1])))          # [[0, 1], [1, 0]]
````

## 题目 2 做题前先补：组合问题里为什么要 start

全排列是“每个位置都从所有数字里挑”，子集和组合不一样：它们关心的是“从数组里挑出哪些数”，**顺序不重要**。

比如 `[1, 2]` 和 `[2, 1]` 是同一个子集，如果还按全排列那样每层遍历整个数组，就会重复。

解决办法：加一个参数 `start`，表示“这一层只能从下标 `start` 开始往后挑”。

```python
for i in range(start, len(nums)):
    path.append(nums[i])
    backtrack(i + 1)      # 下一个位置从 i+1 开始，保证不回头
    path.pop()
```

这样 `[1, 2]` 只会以一种顺序被选出来，不会出现 `[2, 1]`。

**记住这条：组合、子集类问题都要带 `start`；排列类问题不用。**


````python
# 看清 start 的作用
nums = [1, 2, 3]


def demo(start, path):
    print("当前 path =", path, "，这一层可以从下标", start, "开始挑")
    for i in range(start, len(nums)):
        path.append(nums[i])
        demo(i + 1, path)
        path.pop()


demo(0, [])
````

## 题目 3 做题前先补：可以重复选，怎么控制

组合总和和子集很像，但要处理两个新问题：

**一、元素可以重复选。** `[2,2,3]` 是合法的（2 用了两次）。所以递归时不能传 `i + 1`，要传 `i`，表示“下一个位置还可以从自己开始挑”。

**二、要凑出目标和。** 每选一个数，就把目标减去它；当剩余目标正好等于 0 时收集答案；如果剩余目标小于 0，说明选多了，直接返回（这叫剪枝）。

还有一个小优化：先给数组排序，那么一旦 `candidates[i] > remain`，后面的更大了，可以直接 `break` 跳出循环。

**注意：`[2,3]` 和 `[3,2]` 算同一个组合**，所以仍然要用 `start` 控制顺序，避免重复。


````python
# 手工看一次“剩余目标”的变化
candidates = [2, 3, 6, 7]
target = 7


def demo(start, remain, path):
    print("path =", path, "，剩余目标 =", remain)
    if remain == 0:
        print("    找到一个组合")
        return
    if remain < 0:
        print("    超过目标了，剪枝返回")
        return
    for i in range(start, len(candidates)):
        path.append(candidates[i])
        demo(i, remain - candidates[i], path)
        path.pop()


demo(0, target, [])
````

## 参考答案：组合总和

```python
def combination_sum_answer(candidates, target):
    result = []
    path = []
    candidates.sort()

    def backtrack(start, remain):
        if remain == 0:
            result.append(path[:])
            return
        if remain < 0:
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remain:
                break
            path.append(candidates[i])
            backtrack(i, remain - candidates[i])
            path.pop()

    backtrack(0, target)
    return result
```

面试表达：

我用回溯。`path` 是当前选中的组合，`remain` 是还差多少才能凑到目标。每次从 `start` 开始循环，选中一个数后递归，并把剩余目标减去这个数；因为数字可以重复使用，递归时传的是 `i` 而不是 `i + 1`，这样下一层还能继续选当前这个数。如果剩余目标正好等于 0，就收集一份结果；如果小于 0 说明选多了，直接返回。为了剪枝，我先给数组排序，一旦发现当前数字已经大于剩余目标，后面的数字更大，就直接跳出循环。用 `start` 保证组合内部下标递增，避免出现重复的组合。


````python


def combination_sum_answer(candidates, target):
    result = []
    path = []
    candidates.sort()

    def backtrack(start, remain):
        if remain == 0:
            result.append(path[:])
            return
        if remain < 0:
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remain:
                break
            path.append(candidates[i])
            backtrack(i, remain - candidates[i])
            path.pop()

    backtrack(0, target)
    return result


print(sorted(combination_sum_answer([2, 3, 6, 7], 7)))     # [[2, 2, 3], [7]]
print(sorted(combination_sum_answer([2, 3, 5], 8)))       # [[2,2,2,2],[2,3,3],[3,5]]
print(combination_sum_answer([2], 1))                     # []
print(combination_sum_answer([3], 3))                     # [[3]]
````

```python
def letter_combinations_answer(digits):
    if len(digits) == 0:

    phone = {

```

面试表达：

我用回溯。先建立数字到字母串的映射表。递归函数带一个 `index` 参数，表示当前处理到第几个数字。如果 `index` 已经等于数字串长度，说明所有位置都选好了，就把 `path` 里的字符拼成字符串收集起来。否则取出当前数字对应的字母串，逐个做成选择、递归到下一个数字、再撤销。因为每个数字的字母都会全部尝试，所以能覆盖所有组合。时间上界是 O(4 的 n 次方乘 n)，空间 O(n)。


````python


def letter_combinations_answer(digits):
    if len(digits) == 0:
        return []

    phone = {
        "2": "abc", "3": "def", "4": "ghi", "5": "jkl",
        "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz",
    }

    result = []
    path = []

    def backtrack(index):
        if index == len(digits):
            result.append("".join(path))
            return
        letters = phone[digits[index]]
        for ch in letters:
            path.append(ch)
            backtrack(index + 1)
            path.pop()

    backtrack(0)
    return result


print(len(letter_combinations_answer("23")))       # 9
print(letter_combinations_answer(""))              # []
print(sorted(letter_combinations_answer("2")))     # ['a', 'b', 'c']
print(len(letter_combinations_answer("79")))       # 16
````

# 今日小结

回溯模板就三段：**做选择、递归、撤销选择**。今天的三道题都是它的变形：

1. **全排列**：每层从所有数字里挑，跳过已选的，到底部才收集答案。
2. **子集**：带 `start` 保证不回头，每进一个节点就收集答案。
3. **组合总和**：带 `start` + `remain`，递归传 `i` 允许重复选，`remain < 0` 剪枝。
4. **字母组合**：递归参数换成 `index`，处理“第几个数字”。

要记牢的两条：

- **收集答案一定写 `path[:]`**；
- **排列不用 `start`，组合和子集必须用 `start`**。
