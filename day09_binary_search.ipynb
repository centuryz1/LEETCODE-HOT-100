## 题目 1 做题前先补：闭区间二分模板

设搜索区间是 `[left, right]`，左右端点都可能是答案：

```python
while left <= right:
    middle = (left + right) // 2
```

- 当前值小于目标：`left = middle + 1`，排除中点及左侧。
- 当前值大于目标：`right = middle - 1`，排除中点及右侧。

循环结束时 `left = right + 1`，`left` 正好是目标应该插入的位置。

````python
nums = [1, 3, 5, 6]
left, right = 0, len(nums) - 1
target = 2
while left <= right:
    middle = (left + right) // 2
    print(left, middle, right, nums[middle])
    if nums[middle] < target:
        left = middle + 1
    else:
        right = middle - 1
print("插入位置：", left)  # 1
````

LeetCode 35. Search Insert Position

## 题目描述（改写版）

给你一个严格递增的整数数组 `nums` 和整数 `target`。如果目标存在，返回它的下标；否则返回把它插入后仍保持递增顺序的位置。

### 输入、输出与限制

- 数组非空且没有重复值。
- 返回一个下标，可能等于 `len(nums)`。
- 要求使用 `O(log n)` 的算法。

### 示例

`nums=[1,3,5,6]`：`target=5` 返回 2；`target=2` 返回 1；`target=7` 返回 4。

## 解法名称

专业名称：**标准二分查找（Standard Binary Search）**。

## 线性扫描

从左到右找到第一个大于等于目标的位置，时间 `O(n)`。

## 二分思路

在有序数组上每次排除一半搜索区间。找到目标直接返回；未找到时，循环结束后的 `left` 就是第一个大于目标的位置或数组末尾。

时间复杂度 `O(log n)`，额外空间复杂度 `O(1)`。

````python
nums, target = [1, 3, 5, 6], 2
left, right = 0, len(nums) - 1
while left <= right:
    middle = (left + right) // 2
    print(f"区间 [{left},{right}]，middle={middle}，值={nums[middle]}")
    if nums[middle] == target:
        break
    if nums[middle] < target:
        left = middle + 1
    else:
        right = middle - 1
print("答案位置：", left)
````


# 题目 2：搜索二维矩阵

LeetCode 74. Search a 2D Matrix

## 题目描述（改写版）

给你一个非空矩阵：每行从左到右递增，并且每一行的第一个数都大于上一行的最后一个数。判断整数 `target` 是否存在。

### 输入、输出与限制

- 返回布尔值。
- 矩阵整体按逐行展开后严格递增。
- 目标时间复杂度为 `O(log(mn))`。

### 示例

矩阵 `[[1,3,5,7],[10,11,16,20],[23,30,34,60]]` 中，目标 3 返回 `True`，目标 13 返回 `False`。

## 解法名称

专业名称：**一维化二分查找（Flattened Binary Search）**。

## 两次二分

先二分定位可能的行，再在行内二分，可以做到 `O(log m + log n)`。

## 一维化思路

由于逐行展开后整体递增，直接在虚拟下标 `0` 到 `m*n-1` 上二分，再用除法和取余访问实际元素。

时间复杂度 `O(log(mn))`，额外空间复杂度 `O(1)`。

````python
matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
target = 13
rows, cols = len(matrix), len(matrix[0])
left, right = 0, rows * cols - 1
while left <= right:
    middle = (left + right) // 2
    row, col = middle // cols, middle % cols
    value = matrix[row][col]
    print(f"虚拟下标 {middle} -> ({row},{col}) -> {value}")
    if value < target:
        left = middle + 1
    elif value > target:
        right = middle - 1
    else:
        break
````

## 题目 3 做题前先补：找“第一个大于等于目标”的位置

普通二分找到任意一个目标就返回，但边界二分要继续收缩：

- `nums[middle] >= target`：中点可能就是左边界，记录方向并继续向左找。
- `nums[middle] < target`：左边都太小，只能去右边。

最终 `left` 是第一个大于等于 `target` 的位置。类似地，可以写出第一个**大于** `target` 的位置，它减 1 就是最后一个目标位置。

````python
nums, target = [5, 7, 7, 8, 8, 10], 8
left, right = 0, len(nums)
while left < right:
    middle = (left + right) // 2
    if nums[middle] < target:
        left = middle + 1
    else:
        right = middle
print("第一个大于等于 8 的位置：", left)  # 3
````

# 题目 3：在排序数组中查找元素的第一个和最后一个位置

LeetCode 34. Find First and Last Position of Element in Sorted Array

## 题目描述（改写版）

给你一个按非递减顺序排列的整数数组 `nums` 和整数 `target`，返回目标第一次和最后一次出现的下标 `[first, last]`。如果目标不存在，返回 `[-1,-1]`。

### 输入、输出与限制

- 数组可能为空，也可能含大量重复值。
- 必须达到 `O(log n)` 时间复杂度。
- 不能找到一个目标后再线性向两边扩展。

### 示例

`nums=[5,7,7,8,8,10]`、`target=8` 返回 `[3,4]`；目标 6 返回 `[-1,-1]`。

## 解法名称

专业名称：**边界二分查找（Boundary Binary Search）**。

## 线性扩展

先找到一个目标，再向左右扫描；极端情况下全是目标，退化为 `O(n)`。

## 优化思路

写两个二分：`lower_bound` 找第一个大于等于目标的位置；`upper_bound` 找第一个大于目标的位置。若左边界有效，右边界就是 `upper_bound - 1`。

时间复杂度 `O(log n)`，额外空间复杂度 `O(1)`。

````python
nums, target = [5,7,7,8,8,10], 8
left, right = 0, len(nums)
while left < right:
    middle = (left + right) // 2
    print(f"找左界：[{left},{right}) middle={middle} value={nums[middle]}")
    if nums[middle] < target:
        left = middle + 1
    else:
        right = middle
print("左边界：", left)
````

## 参考答案：在排序数组中查找元素的第一个和最后一个位置

````python
def search_range_answer(nums, target):
    def lower_bound(value):
        left, right = 0, len(nums)
        while left < right:
            middle = (left + right) // 2
            if nums[middle] < value:
                left = middle + 1
            else:
                right = middle
        return left

    first = lower_bound(target)
    if first == len(nums) or nums[first] != target:
        return [-1, -1]
    last = lower_bound(target + 1) - 1
    return [first, last]
````

# 题目 4：搜索旋转排序数组

LeetCode 33. Search in Rotated Sorted Array

这是今天的加练题。

## 题目描述（改写版）

一个原本严格递增且没有重复值的数组，在某个位置被旋转。给你旋转后的数组 `nums` 和整数 `target`，返回目标下标；不存在时返回 -1。

### 输入、输出与限制

- 数组非空、元素互不相同。
- 旋转次数未知，也可能等于没有旋转。
- 要求 `O(log n)` 时间复杂度。

### 示例

`nums=[4,5,6,7,0,1,2]`：目标 0 返回 4，目标 3 返回 -1。

## 解法名称

专业名称：**旋转数组二分查找（Binary Search on Rotated Array）**。

## 线性扫描

逐个比较能在 `O(n)` 找到，但没有达到题目要求。

## 二分思路

每轮先判断左半还是右半有序，再检查目标是否位于有序半段的闭区间中，据此保留可能包含答案的一半。

时间复杂度 `O(log n)`，额外空间复杂度 `O(1)`。

````python
nums, target = [4,5,6,7,0,1,2], 0
left, right = 0, len(nums) - 1
while left <= right:
    middle = (left + right) // 2
    print(f"[{left},{right}] middle={middle} value={nums[middle]}")
    if nums[middle] == target:
        print("找到：", middle)
        break
    if nums[left] <= nums[middle]:
        print("左半有序")
        if nums[left] <= target < nums[middle]:
            right = middle - 1
        else:
            left = middle + 1
    else:
        print("右半有序")
        if nums[middle] < target <= nums[right]:
            left = middle + 1
        else:
            right = middle - 1
````
