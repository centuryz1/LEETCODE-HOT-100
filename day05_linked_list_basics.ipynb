## 题目 1 做题前先补：链表节点、`None` 和 `is`

链表不是用下标连接元素，而是每个节点通过 `next` 指向下一个节点。

- `node.val`：当前节点保存的值。
- `node.next`：下一个节点；没有下一个节点时是 `None`。
- `p is q`：判断两个变量是否指向**同一个节点对象**。
- `p.val == q.val`：只比较值相等，不能证明两个链表真的相交。

本题判断的是节点身份，不是节点数值，所以必须使用 `is`。

## 建表流程拆解（为什么这两条链表"相交"）

上面的建表用嵌套 `ListNode(...)` 手搭，为的是精确控制哪些节点共用：

```python
shared = ListNode(8, ListNode(10))                      # 公共尾巴：8 -> 10
head_a = ListNode(4, ListNode(1, shared))               # A：4 -> 1 -> [8 -> 10]
head_b = ListNode(5, ListNode(6, ListNode(1, shared)))  # B：5 -> 6 -> 1 -> [8 -> 10]
```

### 第 1 步：`ListNode(8, ListNode(10))`

嵌套构造是"从里往外"建链：先创建值 10 的节点，再创建值 8 的节点并把它的 `next` 指向 10。`shared` 最终指向值 8 的节点。

```
10 号节点先独立存在 → 8 号节点.next 指向它 → shared 指向 8 号节点
```

### 第 2 步：`ListNode(1, shared)`

造一个值 1 的**新节点**，`next` 指向 `shared`（8 号节点）。

### 第 3 步：`ListNode(4, ...)`

再包一层，`next` 指向刚才的 1 号节点。A 链成型：

```
head_a:  4 → 1 → shared(8) → 10
```

### 第 4 步：B 链

同样手法再造一个值 1 的**新节点**（注意：和 A 的那个 1 不是同一个对象），再依次包上 6、5：

```
head_a:  4 → 1 → shared(8) → 10
head_b:  5 → 6 → 1 → shared(8) → 10
```

### 整体结构图

```
  4 → 1 ──────────────┐
                      ↓
                      8 → 10        ← 从这里开始共用
                      ↑
  5 → 6 → 1 ──────────┘
```

### 为什么这就是"相交"

链表的"相交"判断的是**节点身份**（`is`），不是值：

- `head_a.next.next is shared` → `True`：8 号节点是同一个对象，从这里开始共用。
- `head_b.next.next.next is shared` → `True`：同上。
- A 的 1 号节点和 B 的 1 号节点值都是 1，但**不是同一个对象**——这正好演示了本题的坑：只看值会误判相交。

所以本题的答案（交点）是 8 号节点。

> 补充：`build_list` 是另一种通用建表方式——用 `dummy` 哨兵节点循环追加；本题不用它，是因为需要让两条链精确共用同一条尾巴。

### 示例

## 解法名称

指针 `p` 走完 A 后转去 B，指针 `q` 走完 B 后转去 A。两者最终都走了 `A 独有部分 + B 独有部分 + 公共部分`，长度差被抵消，因此会在交点相遇；不相交时会同时变成 `None`。

时间复杂度 `O(m+n)`，额外空间复杂度 `O(1)`。

## 参考答案：相交链表

````python
# 完整参考答案：相交链表
# 本格可以单独运行，不依赖前面的任何代码格。

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def attach_shared_tail(prefix_values, shared_head):
    """创建前缀，并把前缀尾节点连接到公共链表。"""
    head = build_list(prefix_values)
    if head is None:
        return shared_head
    tail = head
    while tail.next is not None:
        tail = tail.next
    tail.next = shared_head
    return head


def build_intersecting_lists(prefix_a, prefix_b, shared_values):
    """创建两条真正共享同一批节点对象的相交链表。"""
    shared_head = build_list(shared_values)
    head_a = attach_shared_tail(prefix_a, shared_head)
    head_b = attach_shared_tail(prefix_b, shared_head)
    return head_a, head_b, shared_head


def get_intersection_node_answer(head_a, head_b):
    p, q = head_a, head_b
    while p is not q:
        p = head_b if p is None else p.next
        q = head_a if q is None else q.next
    return p


head_a, head_b, shared_head = build_intersecting_lists(
    [4, 1], [5, 6, 1], [8, 4, 5]
)
assert get_intersection_node_answer(head_a, head_b) is shared_head
assert get_intersection_node_answer(build_list([1, 2]), build_list([1, 2])) is None
print('相交节点：', get_intersection_node_answer(head_a, head_b).val)
````

## 题目 2 做题前先补：变量保存的是节点引用

执行 `current.next = previous` 会改变链表箭头。如果没有先保存原来的下一个节点，后面的链表就找不到了。

因此每轮必须按这个顺序：

1. `next_node = current.next`：先保存后路。
2. `current.next = previous`：反转当前箭头。
3. `previous = current`：移动已反转部分的头。
4. `current = next_node`：继续处理剩余链表。

## 题目 3 做题前先补：快慢指针和安全判断

慢指针每次走一步，快指针每次走两步：

```python

fast = fast.next.next
```

但执行 `fast.next.next` 前，必须保证 `fast` 和 `fast.next` 都不是 `None`，所以循环条件写成：

```python

```

Python 的 `and` 从左向右判断，左边已经为假时不会再访问右边，因此不会发生空引用错误。

# 题目 3：环形链表

## 题目描述（改写版）

### 输入、输出与限制

- 输入可能是空链表。
- 返回布尔值 `True` 或 `False`。
- 节点值可能重复，因此不能根据 `val` 判断是否回到了旧节点。
- 不需要找环的入口，本题只判断有没有环。

### 示例

`3 -> 2 -> 0 -> -4`，若 `-4` 的 `next` 指回节点 2，则返回 `True`。

## 解法名称

专业名称：**Floyd 快慢指针判环（Floyd's Cycle Detection）**，也叫龟兔赛跑算法。

遍历时把节点对象放入集合；遇到重复对象就有环。时间 `O(n)`，空间 `O(n)`。

慢指针每次一步，快指针每次两步。有环时快指针会在环内追上慢指针；无环时快指针会先到 `None`。

时间复杂度 `O(n)`，额外空间复杂度 `O(1)`。

## 题目 4 做题前先补：怎样找到链表中点

## 参考答案：回文链表

````python
# 完整参考答案：回文链表
# 本格可以单独运行，不依赖前面的任何代码格。

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    """根据普通列表从头创建一条无环单链表。"""
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    """把无环链表转成普通列表，方便观察结果。"""
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values


def is_palindrome_list_answer(head):
    slow, fast = head, head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    previous = None
    current = slow
    while current is not None:
        next_node = current.next
        current.next = previous
        previous = current
        current = next_node

    left, right = head, previous
    while right is not None:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True


assert is_palindrome_list_answer(build_list([1, 2, 2, 1])) is True
assert is_palindrome_list_answer(build_list([1, 2])) is False
assert is_palindrome_list_answer(build_list([1])) is True
print(is_palindrome_list_answer(build_list([1, 2, 3, 2, 1])))
````

我用快慢指针找到中点，反转后半链表，再从链表头和后半段头部逐个比较。这样时间 O(n)、额外空间 O(1)；如果业务要求不改变输入，我会在比较后恢复后半链表。
