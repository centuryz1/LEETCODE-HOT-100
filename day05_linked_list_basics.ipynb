## 题目 1 做题前先补：链表节点、`None` 和 `is`

链表不是用下标连接元素，而是每个节点通过 `next` 指向下一个节点。

- `node.val`：当前节点保存的值。
- `node.next`：下一个节点；没有下一个节点时是 `None`。
- `p is q`：判断两个变量是否指向**同一个节点对象**。
- `p.val == q.val`：只比较值相等，不能证明两个链表真的相交。

本题判断的是节点身份，不是节点数值，所以必须使用 `is`。

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = ListNode()
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def list_values(head, limit=30):
    values = []
    current = head
    while current is not None and len(values) < limit:
        values.append(current.val)
        current = current.next
    return values

shared = ListNode(8, ListNode(10))
head_a = ListNode(4, ListNode(1, shared))
head_b = ListNode(5, ListNode(6, ListNode(1, shared)))

print(head_a.next.next.val)       # 8
print(head_a.next.next is shared) # True
print(head_b.next.next.next is shared)  # True
````

# 题目 1：相交链表

LeetCode 160. Intersection of Two Linked Lists

## 题目描述（改写版）

给你两个单链表的头节点 `head_a` 和 `head_b`。如果两个链表从某个节点开始共用同一段节点，请返回第一个公共节点；如果完全不相交，返回 `None`。

### 输入、输出与限制

- 输入是两个链表头节点，任意一个都可能是 `None`。
- 返回的是公共节点对象，不是节点的值。
- 相交后，后面的节点必须全部共用；题目保证链表中没有环。
- 不能为了得到答案而破坏原链表。

### 示例

链表 A 为 `4 -> 1 -> 8 -> 4 -> 5`，链表 B 为 `5 -> 6 -> 1 -> 8 -> 4 -> 5`。如果两个值为 8 的位置其实是同一个节点对象，就返回这个节点。

## 解法名称

专业名称：**双指针路径拼接（Two Pointers with Path Switching）**。

## 暴力思路

把链表 A 的所有节点对象放进集合，再遍历链表 B，第一次遇到集合中的节点就是交点。时间 `O(m+n)`，空间 `O(m)`。

## 优化思路

指针 `p` 走完 A 后转去 B，指针 `q` 走完 B 后转去 A。两者最终都走了 `A 独有部分 + B 独有部分 + 公共部分`，长度差被抵消，因此会在交点相遇；不相交时会同时变成 `None`。

时间复杂度 `O(m+n)`，额外空间复杂度 `O(1)`。

````python
p, q = head_a, head_b
step = 0
while p is not q:
    print(f"第 {step} 步：p={None if p is None else p.val}, q={None if q is None else q.val}")
    p = head_b if p is None else p.next
    q = head_a if q is None else q.next
    step += 1
print("相交节点：", None if p is None else p.val)
````

## 题目 2 做题前先补：变量保存的是节点引用

执行 `current.next = previous` 会改变链表箭头。如果没有先保存原来的下一个节点，后面的链表就找不到了。

因此每轮必须按这个顺序：

1. `next_node = current.next`：先保存后路。
2. `current.next = previous`：反转当前箭头。
3. `previous = current`：移动已反转部分的头。
4. `current = next_node`：继续处理剩余链表。

````python
head = build_list([1, 2, 3])
previous = None
current = head

next_node = current.next
current.next = previous
previous = current
current = next_node

print("已反转部分：", list_values(previous))  # [1]
print("还未处理部分：", list_values(current))  # [2, 3]
````

- 节点本身可以复用，只改变 `next` 指向，不需要创建一整条新链表。

### 示例

## 解法名称

专业名称：**迭代指针反转（Iterative Pointer Reversal）**。

先把所有值存入列表，再倒序创建新链表，需要 `O(n)` 额外空间，而且没有真正复用原节点。

时间复杂度 `O(n)`，额外空间复杂度 `O(1)`。

````python
head = build_list([1, 2, 3, 4])
previous, current = None, head
while current is not None:
    next_node = current.next
    current.next = previous
    previous = current
    current = next_node
    print("已反转：", list_values(previous), "待处理：", list_values(current))
````

## 题目 3 做题前先补：快慢指针和安全判断

慢指针每次走一步，快指针每次走两步：

```python
slow = slow.next
fast = fast.next.next
```

但执行 `fast.next.next` 前，必须保证 `fast` 和 `fast.next` 都不是 `None`，所以循环条件写成：

```python
while fast is not None and fast.next is not None:
```

Python 的 `and` 从左向右判断，左边已经为假时不会再访问右边，因此不会发生空引用错误。

````python
head = build_list([3, 2, 0, -4])
tail = head
while tail.next is not None:
    tail = tail.next
tail.next = head.next  # 尾节点重新指向值为 2 的节点，形成环

slow = head.next
fast = head.next.next
print("slow：", slow.val, "fast：", fast.val)
````

# 题目 3：环形链表

LeetCode 141. Linked List Cycle

## 题目描述（改写版）

给你一个单链表头节点 `head`，判断链表中是否存在环。如果沿着 `next` 不断前进，最终会再次到达之前访问过的节点，就说明有环。

### 输入、输出与限制

- 输入可能是空链表。
- 返回布尔值 `True` 或 `False`。
- 节点值可能重复，因此不能根据 `val` 判断是否回到了旧节点。
- 不需要找环的入口，本题只判断有没有环。

### 示例

`3 -> 2 -> 0 -> -4`，若 `-4` 的 `next` 指回节点 2，则返回 `True`。

## 解法名称

专业名称：**Floyd 快慢指针判环（Floyd's Cycle Detection）**，也叫龟兔赛跑算法。

## 哈希集合思路

遍历时把节点对象放入集合；遇到重复对象就有环。时间 `O(n)`，空间 `O(n)`。

## 优化思路

慢指针每次一步，快指针每次两步。有环时快指针会在环内追上慢指针；无环时快指针会先到 `None`。

时间复杂度 `O(n)`，额外空间复杂度 `O(1)`。

````python
slow, fast = head, head
for step in range(8):
    if fast is None or fast.next is None:
        print("到达链表末尾，没有环")
        break
    slow = slow.next
    fast = fast.next.next
    print(f"第 {step + 1} 步：slow={slow.val}, fast={fast.val}")
    if slow is fast:
        print("两个指针相遇，存在环")
        break
````

## 题目 4 做题前先补：怎样找到链表中点

让 `slow` 每次走一步、`fast` 每次走两步。当 `fast` 到达末尾时，`slow` 就在链表中间附近。

- 偶数长度：`slow` 到达后半部分第一个节点。
- 奇数长度：`slow` 到达中间节点；中间值不会影响回文判断。

找到中点后，可以复用上一题的反转方法反转后半段，再从两端向中间比较。

````python
head = build_list([1, 2, 3, 2, 1])
slow, fast = head, head
while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next
print("中点附近的值：", slow.val)  # 3
````

时间复杂度 `O(n)`，额外空间复杂度 `O(1)`。

````python
sample = build_list([1, 2, 2, 1])
slow, fast = sample, sample
while fast is not None and fast.next is not None:
    slow = slow.next
    fast = fast.next.next
print("前半起点：", sample.val, "后半起点：", slow.val)
print("接下来需要反转后半段并逐个比较")
````
