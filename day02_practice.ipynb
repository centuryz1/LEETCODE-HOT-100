## 最长连续序列

LeetCode 128. Longest Consecutive Sequence

给你一个没有排序的整数列表 `nums`，请你找出其中最长的连续数字序列的长度。返回长度；原数组无序；重复数字不影响答案。示例：`[100, 4, 200, 1, 3, 2]` 返回 `4`。


```python
# 题目：最长连续序列
# 解法：哈希集合 + 序列起点判断（Hash Set with Sequence Start Detection）
# 输入：见函数参数。
# 目标：完成题目要求。
# 输出：见题目描述。
# 注意：题目限制和去重/连续/窗口条件。

nums = [100, 4, 200, 1, 3, 2]

def longest_cs(nums):
    nums_set = set(nums)
    length = 1
    ans = 0
   
    for i in nums:
        if i-1 not in nums_set:
            cur = i
            length = 1
            while cur+1 in nums_set:
                cur += 1
                length += 1

            ans = max(ans,length)


    return ans





print(longest_cs(nums))

```

## 三数之和

LeetCode 15. 3Sum

给你一个整数列表 `nums`，请你找出所有不重复的三元组 `[a, b, c]`，使得 `a + b + c == 0`。返回数字三元组，不是下标；结果不能重复。


```python
# 题目：三数之和
# 解法：排序 + 双指针 + 去重（Sorting + Two Pointers with Deduplication）
# 输入：见函数参数。
# 目标：完成题目要求。
# 输出：见题目描述。
# 注意：题目限制和去重/连续/窗口条件。

def three_sum(nums,target):
    ans = []
    nums.sort()
    n = len(nums)

    for i in range(n):
        if i>0 and nums[i] == nums[i-1]:
            continue

        left = i+1
        right = n-1

        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total > 0:
                right -= 1
                while left<right and nums[right] ==nums[right+1]:
                    right -= 1
            elif total<0:
                left += 1
                while left<right and nums[left] == nums[left-1]:
                    left += 1
            else:
                ans.append([nums[i],nums[left],nums[right]])
                right -= 1
                left += 1
                while left<right and nums[left] == nums[left-1]:
                    left += 1
                while left<right and nums[right] == nums[right+1]:
                    right -= 1

    

    return ans
    


```

```python
print(three_sum([-1, 0, 1, 2, -1, -4],0))
print(three_sum([0, 1, 1],0))  # 期望 []
print(three_sum([0, 0, 0],0))  # 期望 [[0, 0, 0]]
```

## 无重复字符的最长子串

LeetCode 3. Longest Substring Without Repeating Characters

给你一个字符串 `s`，请你找出其中不包含重复字符的最长连续子串长度。子串必须连续；返回长度；空字符串答案是 0。


```python
# 题目：无重复字符的最长子串
# 解法：滑动窗口 + 哈希集合（Sliding Window + Hash Set）
# 输入：见函数参数。
# 目标：完成题目要求。
# 输出：见题目描述。
# 注意：题目限制和去重/连续/窗口条件。
s = 'abcabcbb'


def length_of_longest_substring(s):
    seen = set()
    left = 0
    ans = 0

    for right,x in enumerate(s):

        while x in seen:
            seen.remove(s[left])
            left += 1

            
        seen.add(x)

        ans = max(ans,right-left+1)   
    return(ans)






    

print(length_of_longest_substring(s))
```

## 找到字符串中所有字母异位词

LeetCode 438. Find All Anagrams in a String

给你两个字符串 `s` 和 `p`，请你在 `s` 中找到所有子串，使这些子串是 `p` 的字母异位词，并返回这些子串的起始下标。


```python
# 题目：找到字符串中所有字母异位词
# 解法：固定长度滑动窗口 + 字符频次计数（Fixed-size Sliding Window + Frequency Count）
# 输入：见函数参数。
# 目标：完成题目要求。
# 输出：见题目描述。
# 注意：题目限制和去重/连续/窗口条件。
from collections import Counter

def find_anagrams(s, p):
    need = Counter(p)
    window = Counter()
    ans = []
    k = len(p)


    for right,ch in enumerate(s):
        window[ch] += 1

        if right-k+1>0:
            left_ch = s[right-k]
            window[left_ch] -= 1
            if window[left_ch] == 0:
                del window[left_ch]

        if window == need:
            ans.append(right-k+1)


    return ans



        






print(find_anagrams('cbaebabacd', 'abc'))  # 期望 [0, 6]
print(find_anagrams('abab', 'ab'))          # 期望 [0, 1, 2]
```
