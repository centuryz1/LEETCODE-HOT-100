- `while`：**条件循环**，英文是 **While Loop / Conditional Loop**。

## 先给结论

记住最重要的一句话：

- `if`：**条件成立，就执行一次。**
- `while`：**只要条件还成立，就继续执行。**

它们都会判断条件，但执行完代码块以后不同：

- `if` 执行完就往下走，不会自动回来重新判断。
- `while` 执行完会回到条件处，再判断一次。

所以真正要问的是：

> 这个条件成立后，我只处理一次，还是要一直处理到条件不成立？

# 第一部分：认识 `if`

## `if` 的工作方式

```text
判断条件
  ├─ 成立：执行一次 if 代码块
  └─ 不成立：跳过 if 代码块
然后继续执行后面的代码
```

`if` 适合一次性判断：

- 如果今天下雨，就带伞。
- 如果分数大于等于 60，就判定及格。
- 如果当前数字比最大值大，就更新最大值。
- 如果一个数字是连续序列的起点，就开始检查这条序列。

````python
score = 75

if score >= 60:
    print('及格')

print('这句话继续执行')
````

### `if` 不会自动重复

下面的代码即使条件成立，也只会让 `number` 加一次：

````python
number = 1

if number < 5:
    number += 1
    print('if 内部：', number)

print('if 结束后：', number)
# 最终是 2，不会自动继续变成 3、4、5。
````

# 第二部分：认识 `while`

## `while` 的工作方式

```text
判断条件
  ├─ 不成立：结束循环
  └─ 成立：执行代码块
             ↓
          回去重新判断条件
```

`while` 适合“重复次数事先不确定，但知道停止条件”的事情：

- 只要数字还小于 5，就继续加一。
- 只要左边字符仍然重复，就继续移动左指针。
- 只要下一个连续数字存在，就继续向后寻找。
- 只要队尾元素不符合要求，就继续删除。

````python
number = 1

while number < 5:
    number += 1
    print('while 内部：', number)

print('while 结束后：', number)
````

## `while` 的每一轮

| 判断次数 | 判断前的 `number` | `number < 5` | 是否执行 | 执行后的 `number` |
|---|---:|---|---|---:|
| 1 | 1 | True | 是 | 2 |
| 2 | 2 | True | 是 | 3 |
| 3 | 3 | True | 是 | 4 |
| 4 | 4 | True | 是 | 5 |
| 5 | 5 | False | 否，循环结束 | 5 |

循环体执行了 4 次，但条件判断了 5 次。最后还要判断一次 False，程序才知道应该停下。

# 第三部分：你的代码为什么必须用 `while`

你的目标不是只检查“下一个数”，而是要沿着连续序列一直向后找，直到下一个数不存在。

假设：

```python
nums = [100, 4, 200, 1, 3, 2]
nums_set = {1, 2, 3, 4, 100, 200}
```

当 `i = 1` 时，正确过程应该是：

```text
从 1 开始
2 存在：走到 2，长度变成 2
3 存在：走到 3，长度变成 3
4 存在：走到 4，长度变成 4
5 不存在：停止
```

这里要走几步事先不知道，可能一步都不用走，也可能走很多步，因此要使用 `while`。

## 第一版：使用 `if`

`if` 只检查一次 `cur + 1`。即使后面还有 `3`、`4`，执行完一次以后也会直接离开代码块。

````python
def longest_cs_if(nums):
    nums_set = set(nums)
    ans = 0

    for i in nums:
        if i - 1 not in nums_set:
            cur = i
            length = 1
            print(f'\n从起点 {i} 开始')

            if cur + 1 in nums_set:
                cur += 1
                length += 1
                print(f'if 只前进一步：cur={cur}, length={length}')

            ans = max(ans, length)

    return ans

nums = [100, 4, 200, 1, 3, 2]
print('\n第一版答案：', longest_cs_if(nums))  # 错误地得到 2
````

## 正确版：使用 `while`

`while` 每前进一步，都会回去重新检查新的 `cur + 1` 是否存在，因此能走完整条连续序列。

````python
def longest_cs_while(nums):
    nums_set = set(nums)
    ans = 0

    for i in nums:
        if i - 1 not in nums_set:
            cur = i
            length = 1
            print(f'\n从起点 {i} 开始')

            while cur + 1 in nums_set:
                cur += 1
                length += 1
                print(f'while 继续前进：cur={cur}, length={length}')

            print(f'{cur + 1} 不存在，停止这条序列')
            ans = max(ans, length)

    return ans

nums = [100, 4, 200, 1, 3, 2]
print('\n正确答案：', longest_cs_while(nums))  # 4
````

## 为什么外层是 `if`，内层是 `while`？

```python
if i - 1 not in nums_set:
    while cur + 1 in nums_set:
        ...
```

外层 `if` 回答：当前数字是不是一条连续序列的起点？对当前 `i` 只判断一次，所以用 `if`。

内层 `while` 回答：下一个连续数字还存在吗？存在就继续，直到不存在，所以用 `while`。

- `if` 负责**决定要不要开始**。
- `while` 负责**决定要不要继续**。

# 第四部分：三问判断法

## 问题 1：条件成立后，只需要处理一次吗？

如果是，优先考虑 `if`。例如：当前数字比最大值大时，更新一次最大值。

## 问题 2：执行一次后，需要回去重新检查同一个条件吗？

如果是，考虑 `while`。例如：移走一个重复字符后，窗口里可能仍有重复，所以要重新检查。

## 问题 3：循环体有没有让条件逐渐变成 False？

如果使用 `while`，必须能回答“它以后为什么会停”。通常需要更新某个变量：

- `cur += 1`
- `left += 1`
- `count -= 1`
- 从集合、列表或队列中删除元素

如果条件相关变量一直不变，就可能出现死循环。

## 快速选择表

| 场景 | 使用 | 原因 |
|---|---|---|
| 判断一次是否及格 | `if` | 只做一次决定 |
| 当前数更大就更新答案 | `if` | 当前轮只更新一次 |
| 一直加到数字达到目标 | `while` | 重复次数未知 |
| 一直删除重复字符 | `while` | 删除一次后可能仍重复 |
| 遍历列表中的每个元素 | `for` | 遍历对象已经明确 |
| 从序列起点一直走到终点 | `if + while` | `if` 判断起点，`while` 负责延伸 |

一句实用口诀：

> 判断一次用 `if`，反复判断用 `while`，逐个遍历用 `for`。

# 第五部分：`for`、`if`、`while` 的分工

它们不是互相排斥的，可以组合使用：

- `for`：把每个候选对象看一遍。
- `if`：判断这个对象是否需要处理。
- `while`：处理开始后判断是否还要继续。

```python
for i in nums:                      # 看每个数字
    if i - 1 not in nums_set:       # 判断是不是起点
        while cur + 1 in nums_set:  # 沿序列不断向后
            ...
```

专业说法：**遍历 + 条件筛选 + 条件循环**。

# 第六部分：常见错误

## 错误 1：需要重复，却写成 `if`

代码只处理一步，遇到长度超过 2 的连续情况就出错。

## 错误 2：只需要判断一次，却随手写成 `while`

可能重复执行本来只需要一次的操作，甚至死循环。

## 错误 3：`while` 中忘记更新变量

下面的代码不要运行：

```python
number = 1
while number < 5:
    print(number)
```

`number` 永远是 1，条件永远为 True，所以不会停止。

## 错误 4：混淆执行次数和判断次数

`while` 退出前还会多做一次条件判断，发现 False 后才结束。
