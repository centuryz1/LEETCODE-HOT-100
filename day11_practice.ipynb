请你**深拷贝**这条链表：返回一条全新的链表，新链表里每个节点的值、`next` 走向和 `random` 指向，都要和原来一一对应。

关键要求：新链表里的 `next` 和 `random` 必须指向**新链表自己的节点**，不能指向原链表的节点。

## 输入

- `head`：带 `random` 指针的链表头节点，节点个数在 0 到 1000 之间，节点值在 -10000 到 10000 之间。
- `random` 要么指向链表中的某个节点，要么是 `None`。

## 输出

返回复制出来的新链表头节点。

## 示例

链表 `7 -> 13 -> 11 -> 10 -> 1`，其中：

- 7 的 `random` 是 `None`
- 13 的 `random` 指向 7
- 11 的 `random` 指向 1
- 10 的 `random` 指向 11
- 1 的 `random` 指向 7

复制后，新链表的形状要和上面完全一致，而且新节点的 `val` 可以相等，但节点对象必须是新的。

## 易漏细节

- 空链表要返回 `None`。
- 复制 `random` 的时候，目标新节点可能还没有创建出来，这就是需要分两遍遍历的原因。
- 判断是否复制成功，不要只看值，要检查 `new_head is not head`、`new_head.next is not head.next` 这类对象身份。


````python
class RandomNode:
    def __init__(self, val=0, next=None, random=None):
        self.val = val
        self.next = next
        self.random = random


def build_random_list(values, random_indices):
    """创建带 random 指针的链表。

    values 是节点值列表；
    random_indices 是每个节点 random 指向的下标，用 None 表示指向空。
    建表仍然用 dummy + tail + for 逐个创建节点，只在最后多连一遍 random。
    """
    if len(values) == 0:
        return None

    nodes = []
    dummy = RandomNode()
    tail = dummy
    for value in values:
        tail.next = RandomNode(value)
        tail = tail.next
        nodes.append(tail)

    for i in range(len(nodes)):
        index = random_indices[i]
        if index is not None:
            nodes[i].random = nodes[index]

    return dummy.next


def random_list_values(head):
    """输出 [(值, random 指向的值或 None), ...]，方便观察结果。"""
    result = []
    current = head
    while current is not None:
        if current.random is None:
            random_value = None
        else:
            random_value = current.random.val
        result.append((current.val, random_value))
        current = current.next
    return result


# 题目：随机链表的复制
# 解法：哈希表映射（Hash Map Mapping）
# 输入：带 random 指针的链表头节点 head，节点个数 0 到 1000，random 指向链表中任意节点或 None。
# 目标：深拷贝整条链表，新链表的 next 和 random 必须指向新链表自己的节点。
# 输出：返回新链表的头节点；空链表返回 None。
# 注意：要分两遍遍历，第一遍只建映射，第二遍才连 next 和 random；None 要单独判断。


def copy_random_list(head):
    # 在这里写你的代码
    pass
````
